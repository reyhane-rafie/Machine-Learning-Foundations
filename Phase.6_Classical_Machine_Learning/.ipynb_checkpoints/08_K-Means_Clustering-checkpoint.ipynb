# Lesson 8 — K-Means Clustering

> **Goal:** Understand K-Means as an **unsupervised learning** algorithm that automatically groups similar observations, and understand why it can be useful for discovering patterns in Remote Sensing data.

---

# 1. Why should I care?

So far, most of the algorithms you've learned needed **labels** such as "Forest," "Water," or "Urban." K-Means is different: it can find groups **without being given the correct classes beforehand**.

This is especially interesting in Remote Sensing because satellite data contains huge amounts of information, and sometimes we don't already know what the natural groups in that data are.

---

# 2. Core Concept

**K-Means clustering** groups observations according to their similarity.

You tell the algorithm how many groups you want:

```text
K = 3
````

Then K-Means tries to divide the data into **3 clusters**.

The basic process is:

```text
1. Choose K
       ↓
2. Place K centers
       ↓
3. Assign each point to its nearest center
       ↓
4. Move each center to the average
       ↓
5. Repeat
       ↓
6. Stop when the groups stabilize
```

The center of a cluster is called a **centroid**.

The important idea is:

> **Points within the same cluster should be relatively similar to each other, while points in different clusters should be relatively different.**

Unlike supervised learning, there is no `y` label telling the algorithm what each group means.

---

# 3. Simple Example

Imagine you have people described by:

* Height
* Weight

But you don't know their groups beforehand.

You might see:

```text
        Weight
          ↑
          │        ● ● ●
          │       ● ●
          │
          │
          │  × ×
          │ × × ×
          │
          └────────────────→ Height
```

K-Means might discover:

```text
Cluster 1 → shorter/lighter people
Cluster 2 → taller/heavier people
```

But here's something very important:

**K-Means doesn't know what the clusters mean.**

It simply says:

```text
Cluster 1
Cluster 2
```

You have to interpret them.

---

# 4. What Does `K` Mean?

`K` = number of clusters you want.

For example:

```text
K = 2
```

means:

> Find 2 groups.

```text
K = 5
```

means:

> Find 5 groups.

Choosing `K` is an important part of using K-Means.

---

# 5. What Is a Centroid?

A centroid is the **center of a cluster**.

Imagine three points:

```text
●
   ●
      ●
```

The centroid is approximately their average position:

```text
●
   ●  ← centroid
      ●
```

Mathematically, for each feature, the centroid is basically the **mean** of the points in that cluster.

This connects directly to your earlier Statistics lesson:

> **K-Means uses the idea of the mean to find cluster centers.**

---

# 6. Remote Sensing / Environmental Example

Suppose you have satellite pixels described by:

```text
NDVI
NDWI
NDBI
```

But you don't have land-cover labels.

You ask K-Means to create:

```text
K = 4
```

clusters.

The algorithm might produce:

```text
Cluster 1
High NDVI
Low NDBI

Cluster 2
Low NDVI
High NDBI

Cluster 3
High NDWI

Cluster 4
Moderate NDVI
Moderate NDBI
```

You might then interpret them as:

```text
Cluster 1 → Vegetation
Cluster 2 → Urban/built-up
Cluster 3 → Water
Cluster 4 → Agriculture/Bare land
```

**But remember:** the algorithm itself did not know those names.

You interpreted the clusters using environmental knowledge.

---

# 7. K-Means vs Classification

This is extremely important.

### Supervised Classification

You provide labels:

```text
Pixel → Forest
Pixel → Water
Pixel → Urban
```

The model learns to reproduce those labels.

### K-Means

You provide:

```text
Pixel features
```

but no labels.

The algorithm discovers:

```text
Cluster 1
Cluster 2
Cluster 3
```

Then **you interpret the clusters**.

So:

```text
Supervised Learning
Features + Labels
        ↓
Learn classes

K-Means
Features only
        ↓
Discover groups
```

---

# 8. What Problem Does It Solve?

K-Means is mainly used for:

## Clustering

Finding groups in unlabeled data.

Examples:

* Customer segmentation
* Image segmentation
* Land-cover exploration
* Environmental zone identification
* Grouping similar pixels
* Finding patterns in climate data

It is an **unsupervised learning algorithm**.

---

# 9. When Should I Use It?

Use K-Means when:

### 1. You don't have labels

For example:

You have a satellite image but no land-cover training samples.

---

### 2. You want to discover hidden groups

You might want to ask:

> "Are there naturally different environmental zones in this region?"

---

### 3. You want a simple clustering method

K-Means is relatively easy to understand and implement.

---

### 4. You have numerical features

For example:

```text
NDVI
NDWI
Temperature
Elevation
Rainfall
```

---

# 10. What Assumptions Does It Make?

K-Means has some important assumptions.

### 1. Similarity can be represented by distance

K-Means generally uses distance to determine which points belong together.

So the meaning of "similar" must make sense in your feature space.

---

### 2. Clusters are reasonably compact

K-Means works best when clusters are roughly:

* Compact
* Relatively spherical
* Separated from each other

It struggles with unusual cluster shapes.

For example:

```text
      ● ● ●
    ●       ●
   ●         ●
    ●       ●
      ● ● ●
```

K-Means may not handle this kind of structure well.

---

### 3. You need to choose `K`

The algorithm does not automatically know how many clusters you actually want.

---

### 4. Feature scale matters

This is another connection to your Feature Engineering lesson.

Suppose:

```text
NDVI       → 0–1
Elevation  → 0–3000
```

Elevation can dominate the distance calculation.

Therefore, scaling is often important before K-Means.

---

# 11. Advantages

### ✅ Simple

The basic idea is easy to understand.

### ✅ Fast

K-Means is relatively computationally efficient.

### ✅ Works well with large numerical datasets

This can be useful for large environmental datasets.

### ✅ Doesn't require labels

This is its biggest advantage compared with supervised algorithms.

### ✅ Useful for exploration

It can help you discover patterns before building a supervised model.

---

# 12. Limitations

### ❌ You must choose `K`

You need to decide how many clusters to create.

---

### ❌ Sensitive to initialization

Different starting centroids can sometimes produce different results.

Scikit-learn therefore uses multiple initializations by default to improve robustness.

---

### ❌ Sensitive to feature scaling

Large numerical ranges can dominate the distance calculation.

---

### ❌ Sensitive to outliers

An extreme observation can pull a centroid away from the center of its normal group.

---

### ❌ Assumes relatively simple cluster shapes

It doesn't naturally handle every possible shape.

---

### ❌ Clusters don't automatically have meaningful names

K-Means might give you:

```text
Cluster 0
Cluster 1
Cluster 2
```

You need domain knowledge to decide whether these correspond to:

```text
Forest
Water
Agriculture
```

---

# 13. How Do We Evaluate K-Means?

This is different from supervised learning.

We don't necessarily have true labels to compare against.

One common metric is:

## Inertia

It measures how close observations are to their assigned cluster centers.

Generally:

> **Lower inertia = points are closer to their cluster centers.**

But there's a problem.

If we increase `K`:

```text
K = 2 → higher inertia
K = 3 → lower inertia
K = 4 → even lower inertia
...
```

Eventually, if every point gets its own cluster, inertia becomes extremely small.

So we can't simply say:

> "Choose the K with the lowest inertia."

We need to balance compactness with an appropriate number of clusters.

---

# 14. The Elbow Method

A common method for choosing `K` is the **Elbow Method**.

You calculate inertia for different values:

```text
K = 2
K = 3
K = 4
K = 5
K = 6
...
```

Then plot them.

Typically, you look for a point where adding another cluster gives **much less improvement**.

Conceptually:

```text
Inertia
  ↑
  │\
  │ \
  │  \
  │   \__
  │      \__
  │         \_
  └────────────────→ K
          ↑
        Elbow
```

The "elbow" can provide a useful starting point for choosing `K`.

It is not an absolute rule.

---

# 15. Silhouette Score

Another useful metric is the **silhouette score**.

It considers:

> How similar is a point to its own cluster compared with other clusters?

The intuition:

```text
Good clustering:

Within cluster → very similar
Between clusters → very different
```

A higher silhouette score generally indicates better-separated clusters.

Again, it should be treated as a guide rather than a guarantee that the clusters have real-world meaning.

---

# 16. Apply It to a Real Dataset

We'll use the **Iris dataset**, but this time we'll pretend we **don't know the flower species**.

That's important.

We will use only:

```text
X = Measurements
```

and NOT:

```text
y = Species
```

This makes the exercise unsupervised.

---

## Step 1 — Load the Data

```python
from sklearn.datasets import load_iris

data = load_iris()

X = data.data
```

Notice:

```text
X → Features
```

We don't use `data.target`.

---

## Step 2 — Scale the Features

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

We scale because K-Means is distance-based.

---

## Step 3 — Create K-Means

```python
from sklearn.cluster import KMeans

model = KMeans(
    n_clusters=3,
    random_state=42
)
```

We tell the model:

```text
K = 3
```

because we want three clusters.

---

## Step 4 — Fit the Model

```python
model.fit(X_scaled)
```

The algorithm:

1. Starts with cluster centers.
2. Assigns points to the nearest center.
3. Recalculates the centers.
4. Reassigns points.
5. Repeats until the solution stabilizes.

---

## Step 5 — Get Cluster Assignments

```python
clusters = model.labels_

print(clusters)
```

You might see:

```text
0 0 1 1 2 2 0 1 ...
```

These numbers are the **cluster IDs**.

They are not automatically:

```text
0 = Setosa
1 = Versicolor
2 = Virginica
```

That's an important distinction.

---

## Step 6 — Look at the Centroids

```python
print(model.cluster_centers_)
```

These are the centers of the three clusters in the **scaled feature space**.

---

# 17. Remote Sensing Workflow

Imagine you have Sentinel-2 data.

```text
Sentinel-2 Image
       ↓
Extract Features
       ↓
NDVI / NDWI / NDBI / Bands
       ↓
Scale Features
       ↓
Choose K
       ↓
K-Means
       ↓
Cluster Assignment
       ↓
Cluster Map
       ↓
Environmental Interpretation
```

You might obtain:

```text
Cluster 0
Cluster 1
Cluster 2
Cluster 3
```

Then investigate their characteristics:

| Cluster | NDVI   | NDWI | NDBI | Possible Interpretation |
| ------- | ------ | ---- | ---- | ----------------------- |
| 0       | High   | Low  | Low  | Vegetation              |
| 1       | Low    | High | Low  | Water                   |
| 2       | Low    | Low  | High | Urban                   |
| 3       | Medium | Low  | Low  | Agriculture/Bare land   |

The final interpretation requires **domain knowledge and validation**.

---

# 18. Very Important: K-Means Does NOT Know What "Forest" Means

Suppose K-Means produces:

```text
Cluster 0
Cluster 1
Cluster 2
```

It has not discovered "Forest" in the human sense.

It has discovered groups of observations that are similar according to the features you provided.

You then investigate:

```text
Cluster 0 → What are its spectral characteristics?
Cluster 1 → Where does it occur?
Cluster 2 → What environmental conditions does it have?
```

Then you assign meaningful interpretations.

This distinction is extremely important in Remote Sensing research.

---

# 19. Important Points

* K-Means is an **unsupervised learning** algorithm.
* It groups similar observations into `K` clusters.
* `K` is the number of clusters you choose.
* The center of each cluster is called a **centroid**.
* K-Means is distance-based, so **feature scaling matters**.
* It works best for relatively compact, separated clusters.
* The resulting clusters require **interpretation and validation**.

---

# 20. Practice Together

### Question 1

You have 100,000 satellite pixels with:

```text
NDVI
NDWI
Temperature
Elevation
```

but **you have no land-cover labels**.

You want to discover groups of similar pixels.

Would you choose:

**A. Logistic Regression**

**B. Random Forest**

**C. K-Means**

**D. Linear Regression**

Choose one and explain **why**.

I'll check your reasoning before we move to Question 2.

---

# 21. Mini Quiz — Don't Answer Yet

## Conceptual

**1.** What is the difference between supervised learning and K-Means clustering?

**2.** What does `K` represent in K-Means?

**3.** What is a centroid?

---

## Multiple Choice

**4.** Which statement about K-Means is correct?

A. It requires labeled training data.

B. It predicts a continuous numerical value.

C. It groups observations based on similarity.

D. It always automatically finds the correct number of clusters.

---

## Practical Scenario

**5.** You have a Sentinel-2 image and want to explore the natural groups of pixels using:

* Red
* NIR
* NDVI
* NDWI
* NDBI

You don't have training labels.

Answer:

1. Why could K-Means be useful?
2. Why might you scale the features?
3. What does `K = 4` mean?
4. After obtaining four clusters, does the algorithm automatically know which one is "Forest"?
5. Name one method you could use to help choose an appropriate `K`.

---

# 22. Ready to Move On?

Before moving forward, you should be able to explain:

> **"K-Means is an unsupervised algorithm that groups similar observations into K clusters by repeatedly assigning observations to nearby centroids and updating those centroids."**

You should be able to:

* Explain **clustering vs classification**.
* Explain what `K` means.
* Explain what a **centroid** is.
* Describe the basic K-Means iteration.
* Explain why scaling matters.
* Explain why choosing `K` is important.
* Understand the **Elbow Method** and **Silhouette Score** at an intuitive level.
* Explain why K-Means clusters need **interpretation and validation**.
