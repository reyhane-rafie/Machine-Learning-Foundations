# 1. Why Should You Care?

Most real-world data is stored in tables, such as **CSV** or **Excel** files.

Weather records, NDVI measurements, air quality observations, and machine learning datasets are all examples of **tabular data**.

**Pandas** is the most widely used Python library for organizing, cleaning, and analyzing this type of data before building machine learning models.

---

# 2. Core Concept (Short)

## What is Pandas?

**Pandas** is a Python library for working with tabular data.

The two main data structures are:

- **Series** → One column of data.
- **DataFrame** → A complete table with rows and columns.

Think of it like this:

### Series

```text
NDVI
0.72
0.54
0.81
```

### DataFrame

| Location | NDVI | Year |
|----------|-----:|-----:|
| Forest A | 0.72 | 2025 |
| Forest B | 0.54 | 2025 |
| Forest C | 0.81 | 2025 |

Most of the time, you will work with **DataFrames**.

# 4. Remote Sensing / Environmental Example

Imagine you measured NDVI at different locations.

| Location | NDVI | Land Cover |
|----------|-----:|------------|
| Forest A | 0.82 | Forest |
| Farm B | 0.45 | Agriculture |
| City C | 0.12 | Urban |

This table can be stored as a **Pandas DataFrame**.

Now you can:

- Find all forests.
- Sort locations by NDVI.
- Calculate the average NDVI for each land cover type.
- Merge this table with rainfall data.

These are common tasks in environmental research.

---

# 5. Important Concepts

## A) DataFrame

A **DataFrame** is a table.

Example:

```python
import pandas as pd

data = {
    "Location": ["Forest A", "Farm B", "City C"],
    "NDVI": [0.82, 0.45, 0.12]
}

df = pd.DataFrame(data)

print(df)
```

Output:

```text
   Location   NDVI
0  Forest A   0.82
1    Farm B   0.45
2    City C   0.12
```

---

## B) Series

Selecting one column gives a **Series**.

```python
print(df["NDVI"])
```

Output:

```text
0    0.82
1    0.45
2    0.12
```

---

## C) Filtering

Filtering selects only rows that meet a condition.

Example:

```python
healthy = df[df["NDVI"] > 0.5]

print(healthy)
```

Output:

```text
   Location   NDVI
0  Forest A   0.82
```

---

## D) Sorting

Sorting arranges data.

```python
sorted_df = df.sort_values("NDVI")
```

The lowest NDVI appears first.

To sort from highest to lowest:

```python
sorted_df = df.sort_values("NDVI", ascending=False)
```

---

## E) Grouping

Grouping combines rows with the same category.

Example data:

| Land Cover | NDVI |
|------------|-----:|
| Forest | 0.82 |
| Forest | 0.79 |
| Urban | 0.14 |
| Urban | 0.18 |

Average NDVI by land cover:

```python
df.groupby("Land Cover")["NDVI"].mean()
```

Result:

```text
Forest    0.805
Urban     0.160
```

---

## F) Merging

Merging joins two tables using a common column.

### Table 1

| Location | NDVI |
|----------|-----:|
| A | 0.72 |
| B | 0.55 |

### Table 2

| Location | Rainfall |
|----------|---------:|
| A | 420 |
| B | 610 |

Merge the tables:

```python
merged = pd.merge(df1, df2, on="Location")
```

Result:

| Location | NDVI | Rainfall |
|----------|-----:|---------:|
| A | 0.72 | 420 |
| B | 0.55 | 610 |

---

## G) Missing Values

Sometimes data is incomplete.

Example:

| Location | NDVI |
|----------|-----:|
| A | 0.72 |
| B | *(missing)* |
| C | 0.81 |

Find missing values:

```python
df.isna()
```

Remove rows with missing values:

```python
df.dropna()
```

Fill missing values:

```python
df.fillna(0)
```

---

# 6. Important Points

- A **DataFrame** is a table.
- A **Series** is one column from a DataFrame.
- Filtering selects rows based on conditions.
- Sorting changes the order of rows.
- Grouping summarizes data by categories.
- Merging combines tables.
- Missing values should be identified and handled before analysis.

# 8. Mini Coding Example

```python
import pandas as pd

data = {
    "Location": ["Forest A", "Farm B", "City C"],
    "NDVI": [0.82, 0.45, 0.12]
}

df = pd.DataFrame(data)

high_ndvi = df[df["NDVI"] > 0.5]

print(high_ndvi)
```

---

# Line by Line Explanation

## Import Pandas

```python
import pandas as pd
```

Imports the **Pandas** library.

---

## Create the Data

```python
data = {
    "Location": ["Forest A", "Farm B", "City C"],
    "NDVI": [0.82, 0.45, 0.12]
}
```

---

## Create a DataFrame

```python
df = pd.DataFrame(data)
```

---

## Display the Result

```python
print(high_ndvi)
```

Displays the filtered table.
