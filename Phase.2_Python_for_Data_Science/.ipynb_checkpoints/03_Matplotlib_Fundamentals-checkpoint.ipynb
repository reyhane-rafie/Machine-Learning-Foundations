# 1. Why Should You Care?

Data visualization helps humans understand patterns that numbers alone cannot show.

In **AI, Machine Learning, GIS, and Remote Sensing**, visualization is essential for exploring data, finding trends, checking errors, and communicating scientific results.

Examples:

- Showing NDVI changes over time.
- Comparing temperature between cities.
- Understanding the distribution of satellite measurements.
- Visualizing model predictions.

# 2. Core Concept (Short)

**Matplotlib** is a Python library used to create graphs and charts.

Basic workflow:

```python
import matplotlib.pyplot as plt
```

`plt` is a short name for Matplotlib's plotting tools.

Then:

```python
plt.plot(x, y)
plt.show()
```

creates and displays a graph.

Think of visualization as:

```text
Raw Data
    ↓
Python
    ↓
Matplotlib
    ↓
Graph
    ↓
Human Understanding
```

# 5. Important Plot Types

## A) Line Plot

### Purpose

Shows changes over a continuous variable.

Common uses:

- Time series
- Climate trends
- NDVI changes

Example:

```python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4]
ndvi = [0.3, 0.4, 0.6, 0.8]

plt.plot(months, ndvi)

plt.xlabel("Month")
plt.ylabel("NDVI")

plt.show()
```

Result:

A line showing vegetation increase.

---

## B) Scatter Plot

### Purpose

Shows relationships between two variables.

Example question:

**Does rainfall affect NDVI?**

Data:

```python
rainfall = [100, 200, 300, 400]

ndvi = [0.2, 0.4, 0.6, 0.8]
```

Plot:

```python
import matplotlib.pyplot as plt

rainfall = [100, 200, 300, 400]

ndvi = [0.2, 0.4, 0.6, 0.8]

plt.scatter(rainfall, ndvi)

plt.xlabel("Rainfall")
plt.ylabel("NDVI")

plt.show()
```

Each point represents one observation.

Used for:

- Correlation analysis
- Machine learning feature relationships

---

## C) Histogram

### Purpose

Shows the distribution of data.

Example NDVI values:

```text
0.2
0.3
0.5
0.6
0.7
0.8
```

Question:

**Are most areas low vegetation or high vegetation?**

Code:

```python
import matplotlib.pyplot as plt

ndvi = [0.2, 0.3, 0.5, 0.6, 0.7, 0.8]

plt.hist(ndvi)

plt.xlabel("NDVI")
plt.ylabel("Frequency")

plt.show()
```

Used for:

- Understanding data distribution
- Detecting unusual values

---

## D) Bar Chart

### Purpose

Compare categories.

Example:

| Land Cover | NDVI |
|------------|-----:|
| Forest | 0.8 |
| Agriculture | 0.5 |
| Urban | 0.1 |

Code:

```python
import matplotlib.pyplot as plt

land = ["Forest", "Agriculture", "Urban"]
ndvi = [0.8, 0.5, 0.1]

plt.bar(land, ndvi)

plt.ylabel("NDVI")

plt.show()
```

Used for:

- Comparing groups
- Presenting research results

---

# 6. Important Points

- Matplotlib creates visualizations from data.
- Line plots show changes over time.
- Scatter plots show relationships between variables.
- Histograms show data distribution.
- Bar charts compare categories.
- Visualization is important before machine learning because it helps understand the data.

# 7. Mini Coding Example

Let's visualize NDVI over seasons.

```python
import matplotlib.pyplot as plt

season = ["Spring", "Summer", "Autumn", "Winter"]

ndvi = [0.7, 0.9, 0.5, 0.3]

plt.plot(season, ndvi)

plt.xlabel("Season")
plt.ylabel("NDVI")

plt.title("Seasonal NDVI Change")

plt.show()
```
