`````markdown
# Lesson 4 — Jupyter Notebook Fundamentals

**Goal:** Learn how to use Jupyter Notebook as a tool for learning, experimenting, documenting, and presenting Python analysis.

Today we will understand:

- Markdown
- Cells
- Documentation

---

# 1. Why Should You Care?

Jupyter Notebook is one of the most important tools for **Data Science, AI, and Remote Sensing** workflows.

Researchers use it to combine:

- Python code
- Results
- Graphs
- Explanations

Instead of only writing code, you create an **interactive scientific report** where others can understand your analysis.

Examples:

- Processing satellite images.
- Exploring NDVI datasets.
- Training machine learning models.
- Documenting research experiments.

---

# 2. Core Concept (Short)

## What is Jupyter Notebook?

Jupyter Notebook is an interactive environment where you can write and run Python code step by step.

A notebook consists of:

```text
Notebook
   │
   ├── Code Cells
   │
   ├── Markdown Cells
   │
   └── Outputs (Results, Graphs, Tables)
```

Instead of running a complete Python file:

```bash
python analysis.py
```

you can run small parts and immediately see the results.

---

# 3. Cells

A **cell** is a block of content inside a notebook.

There are two main types:

---

## A) Code Cell

Used for writing Python code.

Example:

```python
temperature = 25

print(temperature)
```

Output:

```text
25
```

You can run the cell and immediately see the result.

---

## B) Markdown Cell

Used for writing explanations and documentation.

Markdown:

```markdown
# NDVI Analysis

This notebook calculates vegetation changes using satellite data.
```

Result:

# NDVI Analysis

This notebook calculates vegetation changes using satellite data.

---

# 4. Markdown

Markdown is a simple language for formatting text.

It helps you create readable documentation.

---

## Headings

Markdown:

```markdown
# Main Title

## Section

### Subsection
```

Result:

```text
Main Title
Section
Subsection
```

---

## Bold Text

Markdown:

```markdown
**Important**
```

Result:

**Important**

---

## Lists

Markdown:

```markdown
- Sentinel-2
- Landsat
- MODIS
```

Result:

- Sentinel-2
- Landsat
- MODIS

---

## Code Blocks

Markdown:

````markdown
```python
print("Hello")
```
`````

Result:

```python
print("Hello")
```

# 5. Simple Example

Imagine you are analyzing NDVI.

A good notebook structure:

### Markdown Cell

```markdown
# NDVI Analysis of Isfahan

Goal:
Calculate vegetation changes using Sentinel-2 data.
```

### Code Cell

```python
import numpy as np

ndvi = np.array([0.2, 0.5, 0.8])

print(ndvi.mean())
```

Output:

```text
0.5
```

### Markdown Cell

```markdown
The average NDVI is **0.5**, which indicates moderate vegetation.
```

This creates a complete scientific workflow.

# 6. Documentation

Documentation means explaining:

* What your code does.
* Why you wrote it.
* What the results mean.

Good researchers do not only write code.

They explain their thinking.

---

## Without Documentation

```python
x = np.array([0.2, 0.5, 0.8])

print(x.mean())
```

A reader doesn't know:

* What is `x`?
* Why calculate the mean?
* What does the result represent?

---

## Better Documentation

Markdown:

```markdown
## Calculate Average NDVI

This section calculates the average vegetation index
from selected pixels.
```

Code:

```python
ndvi = np.array([0.2, 0.5, 0.8])

average_ndvi = ndvi.mean()

print(average_ndvi)
```

Now anyone can understand your work.

# 9. Mini Coding Example

A simple notebook workflow:

### Markdown Cell

```markdown
# Temperature Analysis

This notebook calculates average temperature.
```

### Code Cell

```python
temperatures = [20, 25, 30, 35]

average = sum(temperatures) / len(temperatures)

print(average)
```

Output:

```text
27.5
```

### Markdown Cell

```markdown
The average temperature is **27.5°C**.
```
