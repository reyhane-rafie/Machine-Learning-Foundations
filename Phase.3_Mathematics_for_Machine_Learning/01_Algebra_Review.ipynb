# 2. Core Concept (Short)

## A) Fractions

A **fraction** represents part of a whole.

Example:

$$
\frac{1}{2}
$$

means:

> One out of two equal parts.

Fractions can also represent division.

Example:

$$
\frac{6}{3}=2
$$

---

## B) Exponents

An **exponent** tells us how many times a number is multiplied by itself.

Example:

$$
2^3 = 2 \times 2 \times 2 = 8
$$

Where:

- **2** → Base
- **3** → Exponent

---

## C) Logarithms

A **logarithm** answers the question:

> **"What exponent do I need?"**

Example:

$$
\log_2(8)=3
$$

Because:

$$
2^3=8
$$

Think of exponents and logarithms as opposite operations.

**Exponent**

$$
2^3=8
$$

**Logarithm**

$$
\log_2(8)=3
$$

---

# 3. Simple Example

Imagine folding a piece of paper.

- Fold once → 2 layers
- Fold twice → 4 layers
- Fold three times → 8 layers

This follows powers of 2:

$$
2^1=2
$$

$$
2^2=4
$$

$$
2^3=8
$$

Now ask:

> **"I have 8 layers. How many folds did I make?"**

The answer is:

$$
\log_2(8)=3
$$

# 4. Remote Sensing / Environmental Example

## Fractions

Suppose **80 out of 100 pixels** are vegetation.

Vegetation fraction:

$$
\frac{80}{100}=0.8
$$

Meaning:

> **80% of the image is vegetation.**

---

## Exponents

Satellite data often uses scientific notation.

Example:

$$
6.5 \times 10^6
$$

means:

```text
6,500,000
```

Scientific notation is useful because satellite datasets often contain very large numbers.

---

## Logarithms

Logarithms appear in:

- Machine Learning loss functions
- Information theory
- Radar Remote Sensing (decibel scale)
- Measuring changes across large value ranges

For now, remember:

> **A logarithm tells us which exponent produced a number.**

---

# 5. Important Points

- Fractions represent parts of a whole or division.
- Exponents represent repeated multiplication.
- Logarithms are the inverse of exponents.
- These concepts appear frequently in Machine Learning formulas.
- Understanding the ideas is more important than memorizing equations.

# 6. Mini Coding Example

Python can calculate exponents.

```python
print(2 ** 3)
```

Output:

```text
8
```

### Explanation

```python
2 ** 3
```

means:

$$
2^3
$$

---

Python can also calculate logarithms.

```python
import math

print(math.log2(8))
```

Output:

```text
3.0
```

### Explanation

`log2()` asks:

> **"What power of 2 equals 8?"**

Answer:

$$
2^3=8
$$
