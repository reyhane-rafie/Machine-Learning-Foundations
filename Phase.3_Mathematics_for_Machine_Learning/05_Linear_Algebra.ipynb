# 2. Core Concept

Linear Algebra is the mathematics of **vectors** and **matrices**.

Think of it like this:

```text
Scalar
   ↓
One number

Vector
   ↓
A list of numbers

Matrix
   ↓
A table of numbers
```

Machine Learning datasets are usually matrices where:

- **Rows = Observations (Samples)**
- **Columns = Features (Variables)**

Example:

| NDVI | Temperature | Elevation |
|------:|------------:|----------:|
| 0.72 | 24 | 350 |
| 0.58 | 27 | 410 |
| 0.83 | 22 | 290 |

This table is a **matrix**.

Each row is a **vector**.

---

# B) Vectors

Examples:

```text
[2, 5]
```

or

A vector can represent:

- Position

- Features of one observation

This vector describes one student.

---

# C) Matrices

A **matrix** is a table of numbers.

Example:

$$
A=
\begin{bmatrix}
1 & 2\\
3 & 4\\
5 & 6
\end{bmatrix}
$$

A matrix contains many vectors.

### Everyday Example

| Student | Math | English |
|---------|-----:|--------:|
| A | 80 | 75 |
| B | 92 | 88 |
| C | 65 | 70 |

This table is a matrix.

### Remote Sensing Example

A grayscale image:

$$
\begin{bmatrix}
120 & 118 & 115\\
122 & 125 & 121\\
119 & 117 & 116
\end{bmatrix}
$$

Each number is one pixel value.

The entire image is a matrix.

---

# D) Dot Product

The **dot product** combines two vectors into **one number**.

Multiply corresponding elements, then add the results.

Example:

$$
\mathbf{A}=
\begin{bmatrix}
1\\
2
\end{bmatrix},
\qquad
\mathbf{B}=
\begin{bmatrix}
3\\
4
\end{bmatrix}
$$

Dot product:

$$
\mathbf{A}\cdot\mathbf{B}
=(1\times3)+(2\times4)
=3+8
=11
$$

### Everyday Example

Importance:

```text
[3,2]
```

Ratings:

```text
[5,4]
```

The dot product gives one overall score.

### Remote Sensing Example

Linear Regression predicts values by taking the **dot product** between:

- Feature vector
- Model weights

This operation appears throughout Machine Learning.

---

# E) Matrix Multiplication

Matrix multiplication combines information from two matrices to produce a new matrix.

Suppose

$$
A=
\begin{bmatrix}
1 & 2\\
3 & 4
\end{bmatrix},
\qquad
B=
\begin{bmatrix}
5 & 6\\
7 & 8
\end{bmatrix}
$$

Then

$$
AB=
\begin{bmatrix}
19 & 22\\
43 & 50
\end{bmatrix}
$$

Each value is calculated by multiplying a row from **A** by a column from **B**.

### Everyday Example

One matrix stores product prices.

Another stores quantities sold.

Matrix multiplication helps compute total sales.

### Remote Sensing Example

Many image transformations, coordinate transformations, and neural networks repeatedly use matrix multiplication.

---

# F) Identity Matrix

The **identity matrix** is like the number **1** for matrices.

Example:

$$
I=
\begin{bmatrix}
1 & 0\\
0 & 1
\end{bmatrix}
$$

Multiplying any compatible matrix by the identity matrix leaves it unchanged.

$$
A\,I=A
$$

### Everyday Example

Like multiplying by one:

$$
7\times1=7
$$

### Remote Sensing Example

Identity matrices appear in optimization algorithms and covariance calculations.

---

# G) Transpose

The **transpose** swaps rows and columns.

Original matrix:

$$
A=
\begin{bmatrix}
1 & 2 & 3\\
4 & 5 & 6
\end{bmatrix}
$$

Transpose:

$$
A^T=
\begin{bmatrix}
1 & 4\\
2 & 5\\
3 & 6
\end{bmatrix}
$$

Rows become columns.

Columns become rows.

### Remote Sensing Example

Transpose is frequently used when organizing feature matrices for Machine Learning calculations.

---

# H) Inverse Matrix

The **inverse matrix** "undoes" another matrix.

Think of it like division.

For numbers:

$$
5\times\frac{1}{5}=1
$$

For matrices:

$$
A\,A^{-1}=I
$$

where

- \(A^{-1}\) is the inverse matrix.
- \(I\) is the identity matrix.

### Remote Sensing Example

Regression algorithms often use matrix inverses to estimate model parameters.

---

# I) Eigenvectors *(Intuition Only)*

Don't worry about calculations yet.

The key idea:

An **eigenvector** is a special direction that **does not change direction** when transformed by a matrix.

It may become:

- Longer
- Shorter

But it continues pointing in the same direction.

Mathematically:

$$
A\mathbf{v}
=
\lambda\mathbf{v}
$$

where:

- \(A\) = matrix
- \(\mathbf{v}\) = eigenvector
- \(\lambda\) = eigenvalue

### Everyday Example

Imagine stretching a rubber sheet.

Most arrows rotate.

A few arrows continue pointing in exactly the same direction—they only stretch or shrink.

Those are like eigenvectors.

### Remote Sensing Example

**Principal Component Analysis (PCA)** finds important directions in multispectral data using eigenvectors.

PCA reduces data dimensionality while preserving as much information as possible.

# 7. Mini Coding Example

```python
import numpy as np

# Create two vectors
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Dot product
print("Dot Product:", np.dot(a, b))

# Create a matrix
A = np.array([[1, 2],
              [3, 4]])

# Transpose
print("Transpose:")
print(A.T)

# Identity matrix
I = np.eye(2)

print("Identity Matrix:")
print(I)
```

---

## Explanation

```python
np.array()
```

Creates vectors and matrices.

```python
np.dot(a, b)
```

Computes the dot product.

```python
A.T
```

Returns the transpose.

```python
np.eye(2)
```

Creates a **2 × 2 identity matrix**.

These are core operations you'll use in libraries such as:

- NumPy
- Scikit-learn
- TensorFlow
- PyTorch
