# Lesson 2 — Functions and Their Graphs

## Why should I care?

Functions are one of the most important ideas in mathematics. Almost every Machine Learning algorithm is built from functions.

In Remote Sensing, functions describe relationships such as:

- Temperature changes over time
- NDVI versus rainfall
- Population growth
- Sensor calibration
- Machine Learning prediction models

If you understand functions and their graphs, you'll later understand **regression**, **neural networks**, **optimization**, and **probability** much more easily.

---

# Core Concept

A **function** is simply a rule.

It takes an **input** and produces **exactly one output**.

Think of it as a machine.

```text
Input
  │
  ▼
Function
  │
  ▼
Output
```

### Example

\[
f(x)=x+2
\]

If the input is **3**:

\[
f(3)=5
\]

If the input is **10**:

\[
f(10)=12
\]

- **x** is called the **input (independent variable)**.
- **f(x)** is called the **output (dependent variable)**.

---

# Understanding Graphs

A graph shows **how the output changes as the input changes**.

- **Horizontal axis (x-axis)** → Input
- **Vertical axis (y-axis)** → Output

Every point on the graph represents one input-output pair.

### Example

| x | y = x + 2 |
|---|----------:|
| 0 | 2 |
| 1 | 3 |
| 2 | 4 |
| 3 | 5 |

These points form a **straight line**.

---

# Four Important Types of Functions

## 1. Linear Function

### Equation

\[
y=2x+1
\]

### Graph Shape

```text
y
↑
|
|          /
|        /
|      /
|    /
|  /
|/
+----------------→ x
```

### Characteristics

- Straight line
- Constant slope
- Constant rate of change
- Simplest type of function

### Everyday Example

Hourly salary.

If you earn **$20 per hour**:

| Hours | Money ($) |
|------:|----------:|
| 1 | 20 |
| 2 | 40 |
| 3 | 60 |

Every hour adds the same amount.

### Remote Sensing Example

Suppose a satellite sensor has a calibration equation:

\[
Reflectance = 0.8 \times Digital\ Number + 5
\]

Each increase in the digital number changes reflectance by a fixed amount.

---

## 2. Quadratic Function

### Equation

\[
y=x^2
\]

### Graph Shape

```text
y
↑
|
|        •
|      •   •
|    •       •
|  •           •
| •             •
+--------------------→ x
```

Or more simply:

```text
      ∪
```

### Characteristics

- U-shaped curve (parabola)
- Changes slowly near the middle
- Changes faster farther away
- Has a minimum (or maximum if flipped)

### Everyday Example

A ball thrown into the air.

Its height:

- Increases
- Slows
- Stops
- Falls

Its path is approximately quadratic.

### Remote Sensing Example

Plant growth often increases with temperature until an optimum temperature, then decreases.

---

## 3. Exponential Function

### Equation

\[
y=2^x
\]

### Graph Shape

```text
y
↑
|
|                  •
|             •
|         •
|      •
|   •
| •
+----------------------→ x
```

### Characteristics

- Starts slowly
- Then grows very quickly
- Growth accelerates

### Everyday Example

Imagine bacteria doubling every hour.

```text
1
2
4
8
16
32
64
```

Growth becomes very large very quickly.

### Remote Sensing Example

Some populations or invasive species grow approximately exponentially during their early stages.

---

## 4. Logarithmic Function

### Equation

\[
y=\log_2(x)
\]

### Graph Shape

```text
y
↑
|
| •••••••••
|      •
|    •
|  •
| •
+----------------------→ x
```

### Characteristics

- Rapid increase at first
- Then gradually levels off
- Inverse (opposite) of exponential growth

### Everyday Example

Learning a new skill.

At first you improve quickly.

Later, every improvement requires much more effort.

### Remote Sensing Example

Many environmental variables are transformed using logarithms before building Machine Learning models because it makes skewed data easier to analyze.

---

# Comparison of the Four Graphs

## Linear

```text
y
↑
|        /
|      /
|    /
|  /
|/
+--------------→ x
```

## Quadratic

```text
y
↑
|
|      •   •
|    •       •
|  •           •
| •             •
+----------------→ x
```

## Exponential

```text
y
↑
|
|                •
|            •
|        •
|     •
|  •
|•
+----------------→ x
```

## Logarithmic

```text
y
↑
|
| •••••••••
|      •
|    •
|  •
| •
+----------------→ x
```

---

# Easy Way to Remember

| Function | Shape | Memory Trick |
|----------|-------|--------------|
| Linear | `/` | Constant increase |
| Quadratic | `∪` | Bowl shape |
| Exponential | `J` | Explodes upward |
| Logarithmic | `┐` | Fast at first, then slows |

---

# Simple Example

Imagine driving a car.

- **Linear** → Constant speed
- **Quadratic** → Ball thrown from the car
- **Exponential** → Rabbits reproducing rapidly
- **Logarithmic** → Learning a new language

---

# Remote Sensing / Environmental Example

Imagine predicting vegetation health.

Different relationships can appear:

### Linear

More rainfall → slightly higher NDVI.

### Quadratic

NDVI increases with temperature until an optimum, then decreases.

### Exponential

Insect populations rapidly increase during an outbreak.

### Logarithmic

Sensor measurements or environmental variables are transformed to improve Machine Learning models.

Choosing the correct function helps build more accurate models.

---

# Important Points

- A function maps one input to exactly one output.
- A graph shows how outputs change as inputs change.
- Linear functions have a constant rate of change.
- Quadratic functions form a parabola.
- Exponential functions grow (or decay) rapidly.
- Logarithms are the inverse of exponentials.

# Mini Coding Example

```python
def f(x):
    return x + 5

print(f(3))
print(f(10))
```

### Explanation

- `def f(x):` creates a function named `f`.
- `x` is the input.
- `return x + 5` computes the output.
- `print(f(3))` prints **8**.
- `print(f(10))` prints **15**.

Later in Machine Learning, you'll define functions that take many inputs (features) and produce predictions.
