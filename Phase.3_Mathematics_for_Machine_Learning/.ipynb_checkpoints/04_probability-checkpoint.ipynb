# 2. Core Concept

Probability measures **how likely an event is to happen.**

Its value is always between **0 and 1**.

- **0** → Impossible
- **1** → Certain
- **Between 0 and 1** → Possible

Example:

Rolling a fair six-sided die:

Probability of rolling a **4**:

$$
P(4)=\frac{1}{6}
$$

Probability of rolling an **even number**:

$$
P(\text{Even})=\frac{3}{6}=\frac{1}{2}
$$

To understand probability, we first need a few basic ideas.

---

# A) Sample Space

The **sample space** is the set of **all possible outcomes** of an experiment.

Example:

Rolling one die:

$$
S=\{1,2,3,4,5,6\}
$$

This entire set is called the **sample space**.

### Everyday Example

Flipping one coin:

$$
S=\{\text{Heads},\ \text{Tails}\}
$$

These are all possible outcomes.

### Remote Sensing Example

Suppose you're classifying land cover.

Possible classes:

- Forest
- Water
- Urban
- Agriculture

These possible classes form the sample space for your classifier.

---

# B) Events

An **event** is one or more outcomes from the sample space.

Example:

Sample space:

$$
S=\{1,2,3,4,5,6\}
$$

Event:

"Rolling an even number"

$$
E=\{2,4,6\}
$$

An event is simply a **subset of the sample space**.

### Remote Sensing Example

Suppose a model predicts:

- Forest
- Water
- Urban
- Agriculture

The event **"Vegetation"** could include:

$$
\{\text{Forest},\ \text{Agriculture}\}
$$

---

# C) Conditional Probability

Sometimes probabilities change because **new information becomes available**.

Conditional probability means:

> **The probability of A, given that B has already happened.**

It is written as:

$$
P(A\mid B)
$$

Formula:

$$
P(A\mid B)=\frac{P(A\cap B)}{P(B)}
$$

where:

- \(A \cap B\) means **A and B happen together**.
- \(P(B)\neq0\).

---

### Everyday Example

Without knowing today's weather:

> What is the probability you'll carry an umbrella?

Now suppose someone tells you:

> **"It's raining."**

Your probability changes dramatically.

---

### Remote Sensing Example

Suppose a pixel is known to be **cloudy**.

Given that information, the probability that its **NDVI measurement is unreliable** becomes much higher.

---

# D) Independence

Two events are **independent** if one does **not affect** the probability of the other.

Example:

- Rolling a die
- Flipping a coin

The die result does not change the coin result.

They are independent.

Mathematically:

$$
P(A\cap B)=P(A)\times P(B)
$$

---

### Everyday Example

Today's weather does not affect the result of rolling a fair die.

---

### Remote Sensing Example

If two sensors measure completely unrelated environmental variables, those measurements may be approximately independent.

*(In practice, many environmental variables are **not** independent. For example, rainfall and vegetation are often related.)*

---

# E) Bayes' Theorem

Bayes' Theorem updates a probability when **new evidence** becomes available.

Think of it like this:

Initial belief

↓

Observe new evidence

↓

Update your belief

Formula:

$$
P(A\mid B)=
\frac{P(B\mid A)\,P(A)}
{P(B)}
$$

where:

- **\(P(A)\)** → Prior probability
- **\(P(B|A)\)** → Likelihood
- **\(P(B)\)** → Total probability of the evidence
- **\(P(A|B)\)** → Posterior probability

---

### Everyday Example

Suppose you think there's only a small chance your friend is home.

Then you see their car in the driveway.

You now believe it's much more likely they're home.

Your probability has been **updated using new evidence**.

---

### Remote Sensing Example

An image classifier initially predicts a pixel is **Forest**.

Then it also observes:

- High NDVI
- Low surface temperature
- High moisture

The probability of **Forest** increases after considering this new evidence.

Many probabilistic Machine Learning algorithms use this idea.

---

# F) Probability Distributions

A **probability distribution** tells us how probabilities are assigned to different values.

Example:

Rolling a fair die.

| Outcome | Probability |
|---------:|------------:|
| 1 | 1/6 |
| 2 | 1/6 |
| 3 | 1/6 |
| 4 | 1/6 |
| 5 | 1/6 |
| 6 | 1/6 |

This table is a **probability distribution**.

---

## Common Probability Distributions

As you continue studying Machine Learning, you'll encounter:

- Uniform Distribution
- **Normal (Gaussian) Distribution ⭐**
- Binomial Distribution

We'll study each one in detail later.

# 3. Simple Example

Imagine choosing one marble from a bag.

The bag contains:

- 🔴 3 red
- 🔵 2 blue
- 🟢 1 green

Sample space:

All 6 marbles.

Event:

> "Choose a blue marble."

Conditional probability:

Suppose someone tells you:

> "The marble is **not red**."

Now the probabilities change because new information is available.

Independence:

Drawing one marble **without replacement** changes the next draw.

Therefore, the two draws are **not independent**.

---

# 4. Remote Sensing / Environmental Example

Suppose you're building a model to classify pixels into:

- Forest
- Water
- Urban

The model outputs:

| Class | Probability |
|--------|------------:|
| Forest | 0.82 |
| Water | 0.12 |
| Urban | 0.06 |

The model does **not** say:

> "This is definitely forest."

Instead, it says:

> "There is an **82% probability** this pixel is forest."

This is how many Machine Learning classifiers work.

---

# 5. Important Points

- Probability measures how likely an event is.
- The sample space contains all possible outcomes.
- An event is one or more outcomes from the sample space.
- Conditional probability updates probabilities when new information is known.
- Independent events do not affect each other's probabilities.
- Bayes' theorem updates beliefs using evidence.
- A probability distribution describes how probabilities are assigned across possible values.

# 7. Mini Coding Example

```python
import random

# Roll a die
roll = random.randint(1, 6)

print("You rolled:", roll)

# Check if the event "even number" happened
if roll % 2 == 0:
    print("Event: Even number")
else:
    print("Event: Odd number")
```

---

## Explanation

```python
import random
```

Imports Python's built-in **random** module.

```python
random.randint(1, 6)
```

Simulates rolling a fair six-sided die.

```python
roll % 2 == 0
```

Checks whether the result is even.

This program demonstrates a simple **random experiment** and an **event**.
