# Lesson 6 — Calculus for Machine Learning

## 🎯 Goal

Learn the **intuition** behind calculus—not become a mathematician.

We'll focus only on the concepts that are essential for **Machine Learning** and **Remote Sensing**.

---

# 1. Why Should I Care?

Calculus tells us **how things change**.

In Machine Learning, we want to know how changing a model's parameters affects its predictions and errors.

Almost every Machine Learning algorithm learns by **minimizing error**, and calculus provides the mathematical tools to determine **which direction reduces that error the fastest**.

In Remote Sensing, calculus also appears in:

- Optimization
- Image processing
- Physical models
- Deep learning for satellite image analysis

---

# 2. Core Concept

Calculus is the mathematics of **change**.

There are two major branches:

| Branch | Purpose |
|---------|----------|
| **Differential Calculus** | Studies how fast something changes |
| **Integral Calculus** | Adds up many small pieces |

For Machine Learning, **Differential Calculus** is the important one because it helps answer questions like:

- Is the error increasing or decreasing?
- Which direction should the model move?
- How much should the parameters change?

We'll cover six key ideas:

1. Limits
2. Derivatives
3. Gradient
4. Partial Derivatives
5. Chain Rule
6. Gradient Descent

---

# A. Limits (Basic Idea)

## What is a Limit?

A **limit** asks:

> **What value does a function approach as the input gets closer and closer to a point?**

Notice that we care about **what the function approaches**, not necessarily the value at that exact point.

---

## Everyday Example

Imagine walking toward a closed door.

```
You
  🚶
      🚶
           🚶
                🚪
```

Every step brings you closer.

Even if you haven't touched the door yet, we can still describe the position you're approaching.

---

## Why Limits Matter

Derivatives are built from limits.

Instead of measuring change over a large interval, we let the interval become **smaller and smaller**, approaching zero.

Mathematically,

$$
f'(a)=\lim_{h\rightarrow0}\frac{f(a+h)-f(a)}{h}
$$

This formula defines the derivative.

---

## Example

Let

$$
f(x)=x^2
$$

Find the derivative at

$$
x=1
$$

Using the definition,

$$
f'(1)=
\lim_{h\to0}
\frac{(1+h)^2-1^2}{h}
$$

Expand:

$$
(1+h)^2=1+2h+h^2
$$

Therefore,

$$
f'(1)=
\lim_{h\to0}
\frac{2h+h^2}{h}
=
\lim_{h\to0}(2+h)=2
$$

---

# B. Derivatives

## Definition

A **derivative** measures **how fast a function changes**.

Another way to think about it:

> The derivative is the **slope** of the curve at one point.

---

## Interpretation

| Slope | Meaning |
|--------|---------|
| Positive | Function is increasing |
| Negative | Function is decreasing |
| Zero | Flat point (possible maximum or minimum) |

---

## Example

Consider

$$
y=x^2
$$

Its derivative is

$$
\frac{dy}{dx}=2x
$$

That means:

| x | Derivative | Interpretation |
|---|------------|---------------|
| -2 | -4 | Decreasing |
| 0 | 0 | Flat |
| 3 | 6 | Increasing rapidly |

---

## Everyday Example

Driving a car:

```
Position
     ↓
Speed
```

Position tells you **where you are**.

Speed tells you **how fast your position changes**.

A derivative is like the **speed of a function**.

---

## Remote Sensing Example

Suppose vegetation changes over time.

The derivative tells us how quickly **NDVI** is increasing or decreasing, which can indicate:

- Rapid vegetation growth
- Vegetation stress
- Seasonal changes

---

# C. Gradient

A derivative works for a function with **one variable**.

Machine Learning models usually contain **many variables**.

The **gradient** is simply a collection of derivatives.

---

## Mathematical Definition

For a function

$$
f(x,y)
$$

the gradient is

$$
\nabla f=
\left[
\frac{\partial f}{\partial x},
\frac{\partial f}{\partial y}
\right]
$$

For more variables,

$$
\nabla f=
\left[
\frac{\partial f}{\partial x_1},
\frac{\partial f}{\partial x_2},
\cdots,
\frac{\partial f}{\partial x_n}
\right]
$$

---

## Intuition

Think of the gradient as:

> **An arrow pointing in the direction of the greatest increase.**

---

## Everyday Example

Imagine standing on a hill.

```
          ▲
        ▲
      ▲
    😊
```

The gradient points **uphill**.

To go downhill, you move in the **opposite direction**.

---

## Remote Sensing Example

When training a land-cover classifier, the gradient tells the model how each parameter should change to improve predictions.

---

# D. Partial Derivatives

When a function has several variables, we change **one variable at a time** while keeping the others fixed.

---

## Example

Suppose

$$
\text{Crop Yield}
=
f(\text{Rainfall},\text{Temperature})
$$

A partial derivative answers:

- What happens if rainfall changes?
- Temperature stays fixed.

Or

- What happens if temperature changes?
- Rainfall stays fixed.

---

## Mathematical Notation

Rainfall:

$$
\frac{\partial f}{\partial \text{Rainfall}}
$$

Temperature:

$$
\frac{\partial f}{\partial \text{Temperature}}
$$

---

## Everyday Example

Soup depends on:

- Salt
- Pepper
- Water

A partial derivative asks:

> What happens if I change **only the salt**?

---

## Remote Sensing Example

A Machine Learning model might use:

- NDVI
- Elevation
- Rainfall
- Temperature

Partial derivatives measure the effect of changing **one feature** while keeping the others fixed.

---

# E. Chain Rule

Many Machine Learning models combine several functions together.

The **Chain Rule** tells us how changes move through all of them.

---

## Intuition

A change in one step affects the next.

```
Input
   ↓
Layer 1
   ↓
Layer 2
   ↓
Prediction
```

---

## Mathematical Formula

If

$$
y=f(u)
$$

and

$$
u=g(x)
$$

then

$$
\frac{dy}{dx}
=
\frac{dy}{du}
\cdot
\frac{du}{dx}
$$

---

## Everyday Example

```
Flour
   ↓
Dough
   ↓
Bread
```

Changing the flour changes:

- Dough
- Bread

The effect travels through every step.

---

## Remote Sensing Example

Deep neural networks process satellite images through many layers.

The Chain Rule enables **backpropagation**, allowing the network to determine how early layers influence the final prediction.

---

# F. Gradient Descent (Intuition)

Gradient Descent is the optimization algorithm that allows Machine Learning models to learn.

Its goal is simple:

> Find the parameter values that produce the **smallest error**.

---

## Imagine the Error Surface

```
High Error
      ▲
     / \
    /   \
   /     \
  /       \
 /         \
------------→
 Low Error
```

The model starts somewhere on the hill.

It repeatedly:

1. Computes the gradient
2. Moves a small step in the opposite direction
3. Computes the new gradient
4. Repeats

Eventually it reaches (or gets close to) the minimum.

---

## Mathematical Update Rule

If

$$
\theta
$$

represents the model parameters,

then one Gradient Descent step is

$$
\theta
=
\theta
-
\alpha
\nabla J(\theta)
$$

where

| Symbol | Meaning |
|---------|----------|
| $\theta$ | Parameters |
| $\alpha$ | Learning rate |
| $\nabla J$ | Gradient of the error |

---

## Everyday Example

Imagine hiking in thick fog.

You cannot see the whole mountain.

You simply:

- Feel the slope beneath your feet.
- Take one small downhill step.
- Repeat.

Eventually, you reach the valley.

---

## Remote Sensing Example

Suppose you're training a model to predict forest cover from satellite imagery.

Initially, predictions are poor.

Gradient Descent gradually adjusts the model's parameters until prediction error becomes much smaller.

---

# 3. Simple Example

Imagine riding a bicycle uphill.

| Concept | Interpretation |
|----------|----------------|
| Limit | Getting closer to the top |
| Derivative | How steep the road is |
| Gradient | Steepest uphill direction |
| Partial Derivative | Change if moving only north or only east |
| Chain Rule | Pedaling → Wheel rotation → Speed → Position |
| Gradient Descent | Small downhill steps to reach the valley |

---

# 4. Remote Sensing / Environmental Example

Suppose you're training a Machine Learning model to estimate **soil moisture** using:

- NDVI
- Rainfall
- Elevation
- Temperature

During training:

- The **derivative** measures how prediction error changes.
- **Partial derivatives** measure the effect of each feature.
- The **gradient** combines all those changes.
- The **Chain Rule** propagates information through multiple layers.
- **Gradient Descent** repeatedly updates parameters until the prediction error becomes small.

This is how many Machine Learning models learn from environmental data.

---

# 5. Important Points

| Concept | Summary |
|----------|---------|
| Limit | Describes what a function approaches |
| Derivative | Measures slope or rate of change |
| Gradient | Collection of derivatives |
| Partial Derivative | Changes one variable while others remain fixed |
| Chain Rule | Connects changes through multiple functions |
| Gradient Descent | Optimization algorithm that minimizes error |

# 7. Mini Coding Example

```python
def error(x):
    return (x - 3) ** 2

# Try different values
for x in [0, 1, 2, 3, 4]:
    print(f"x = {x}, Error = {error(x)}")
```

## Explanation

- `error(x)` is a simple error function.
- The smallest error occurs when **x = 3**.
- In Machine Learning, **Gradient Descent** automatically searches for parameter values that minimize an error function instead of trying every possible value.
