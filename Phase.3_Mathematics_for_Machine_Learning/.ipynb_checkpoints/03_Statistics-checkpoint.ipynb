---

# 2. Core Concept

Imagine you measured the NDVI of five locations:

```text
0.2, 0.4, 0.5, 0.6, 0.8
```

Statistics answers questions like:

**What is a typical value?**

→ Mean and Median

**How spread out are the values?**

→ Variance and Standard Deviation

**How are the values distributed?**

→ Distribution

These measures help describe a dataset before you start analyzing or building a Machine Learning model.

---

# A) Mean (Average)

The **mean** is the average value.

Add all numbers and divide by how many there are.

Example:

```text
10, 20, 30
```

Mean:

$$
\frac{10+20+30}{3}=20
$$

### Everyday Example

Suppose your exam scores are:

```text
70, 80, 90
```

Average score:

$$
\frac{70+80+90}{3}=80
$$

### Remote Sensing Example

Suppose the NDVI of five pixels is:

```text
0.30
0.40
0.45
0.50
0.55
```

The **mean NDVI** represents the overall vegetation health of that area.

---

# B) Median

The **median** is the middle value after sorting the data.

Example:

```text
3 5 8 10 12
```

Median:

```text
8
```

If there are an even number of values, average the two middle values.

---

## Why is the Median Useful?

Suppose salaries are:

```text
2000
2200
2400
2600
50000
```

Mean:

```text
≈ 11,840
```

Median:

```text
2,400
```

The **mean** is pulled upward by one unusually high salary.

The **median** better represents a typical salary.

### Remote Sensing Example

A few cloudy or noisy pixels can have extremely high or low values.

The **median** is often more reliable than the mean because it is less affected by outliers.

---

# C) Variance

Variance measures **how spread out the data are**.

- Small variance → values are close together.
- Large variance → values are spread apart.

Example:

Dataset A

```text
49 50 51
```

Dataset B

```text
10 50 90
```

Both have the same mean (50),

but Dataset B has a much larger variance.

### Remote Sensing Example

- A desert usually has similar pixel values → **low variance**.
- An urban area contains buildings, roads, trees, and water → **high variance**.

---

# D) Standard Deviation

Standard deviation is closely related to variance.

Think of it as:

> **The typical distance from the mean.**

- Small standard deviation → values stay close to the mean.
- Large standard deviation → values vary a lot.

Unlike variance, **standard deviation uses the same units as the original data**, making it easier to interpret.

### Remote Sensing Example

If land surface temperature has:

- High standard deviation → temperatures vary greatly.
- Low standard deviation → temperatures are fairly uniform.

---

# E) Distribution

A **distribution** describes how data are spread across values.

Questions it answers:

- Are most values near the center?
- Are there many high values?
- Are there unusual values (outliers)?

A **histogram** is one of the most common ways to visualize a distribution.

---

## Common Distribution Shapes

### Symmetric

Values are balanced around the center.

### Right-skewed

Many small values and a few very large values.

### Left-skewed

Many large values and a few very small values.

### Bimodal

Two clear peaks in the data.

---

# 3. Simple Example

Imagine the heights of students in a class.

- **Mean** → Average height.
- **Median** → Middle height.
- **Variance** → How much heights differ.
- **Standard deviation** → Typical difference from the average.
- **Distribution** → Overall pattern of heights.

---

# 4. Remote Sensing / Environmental Example

Suppose you're analyzing NDVI values from a satellite image.

- **Mean** → Overall vegetation health.
- **Median** → Typical vegetation, even if some pixels are affected by clouds.
- **Variance** → How different vegetation is across the area.
- **Standard deviation** → Typical variation in vegetation.
- **Distribution** → Reveals whether most pixels have similar NDVI or if there are different land-cover types (forest, farmland, water, etc.).

These statistics are often the **first step** before applying Machine Learning.

---

# 5. Important Points

- Mean is the average value.
- Median is the middle value after sorting.
- Variance measures how spread out the data are.
- Standard deviation is the typical distance from the mean.
- A distribution describes the overall pattern of the data.

# 7. Mini Coding Example

```python
import numpy as np

data = [4, 6, 8]

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Variance:", np.var(data))
print("Standard Deviation:", np.std(data))
```

---

## Explanation

```python
import numpy as np
```

Imports the **NumPy** library.

```python
data = [4, 6, 8]
```

Stores the numbers.

```python
np.mean(data)
```

Computes the average.

```python
np.median(data)
```

Finds the middle value.

```python
np.var(data)
```

Computes the variance.

```python
np.std(data)
```

Computes the standard deviation.

These functions are used frequently when analyzing environmental and satellite datasets.
