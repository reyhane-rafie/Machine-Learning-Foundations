# 1. Why Should I Care?

A Machine Learning model can appear to perform well simply because it **memorized the training data**.

We need reliable methods to know whether the model can make predictions on **new, unseen data**.

In **Remote Sensing**, this is extremely important.

For example:

A land-cover classification model trained on one region should ideally work on new satellite images from another area.

Train/test split, cross-validation, and evaluation metrics help us measure this ability.

---

# 2. Core Concept

Building a Machine Learning model has three main stages:

```

Data
↓
Training
↓
Testing
↓
Evaluation

```

The main question is:

> "Does my model understand the pattern, or did it just memorize the examples?"

To answer this question, we separate data and evaluate performance.

---

## The Three Main Concepts

| Concept | Purpose |
|---------|---------|
| **Train/Test Split** | Separate data into learning data and testing data |
| **Cross Validation** | Test the model multiple times using different data splits |
| **Evaluation Metrics** | Measure how good the model's predictions are |

---

# A. Train/Test Split

## What is Train/Test Split?

Train/Test split divides the dataset into two parts:

---

## Training Data

Used to teach the model.

The model learns patterns and relationships from this data.

---

## Testing Data

Used to evaluate the model on examples it has never seen before.

---

## Common Split

```

80% Training Data
20% Testing Data

```

Example:

You have 1000 satellite pixels:

```

800 pixels → Training
200 pixels → Testing

```

The model learns from the 800 pixels and is tested on the 200 unseen pixels.

---

# Everyday Example

Imagine studying for an exam.

## Training Data

- Textbook
- Practice questions
- Homework

## Testing Data

- The final exam

If you only memorize the practice questions, you may perform well during practice but fail the real exam.

---

# Remote Sensing Example

You train a forest classification model.

## Training Data

```

Satellite images from 2020

```

## Testing Data

```

Satellite images from 2021

```

The model should recognize forests in a new year, not just memorize the old images.

---

# B. Why Not Test on Training Data?

Because the model already knows those examples.

Example:

A student memorizes answers:

```

Practice Exam:
100%

Real Exam:
40%

```

The student did not learn the concept.

They only memorized the examples.

This problem is called:

# Overfitting

Overfitting happens when:

- The model performs very well on training data.
- The model performs poorly on new unseen data.

---

# C. Cross Validation

## What is Cross Validation?

Cross-validation is a more reliable way to test a Machine Learning model.

Instead of splitting data once, we split it multiple times.

The most common method is:

# K-Fold Cross Validation

---

## Example: 5-Fold Cross Validation

Dataset:

```

Fold 1
Fold 2
Fold 3
Fold 4
Fold 5

```

The model trains 5 times.

Each time:

- 4 folds → Training
- 1 fold → Testing

---

## Round 1

```

Testing:
Fold 1

Training:
Fold 2, Fold 3, Fold 4, Fold 5

```

---

## Round 2

```

Testing:
Fold 2

Training:
Fold 1, Fold 3, Fold 4, Fold 5

```

---

At the end:

The average performance becomes the final score.

---

# Everyday Example

Instead of judging a student's ability from one exam:

Give them five different exams and calculate the average score.

This gives a more reliable estimate of their ability.

---

# Remote Sensing Example

Suppose you have:

```

10,000 labeled pixels

```

for land-cover classification.

Instead of testing only one random group, cross-validation checks performance across different subsets.

This gives a more trustworthy estimate.

# D. Evaluation Metrics

After training a model, we need to measure performance.

The evaluation metric depends on the problem type.

There are two major categories:

1. Regression Metrics
2. Classification Metrics

---

# 1. Regression Metrics

Used when predicting continuous values.

Examples:

- Temperature
- Rainfall
- Soil moisture
- Surface temperature

---

# Mean Absolute Error (MAE)

MAE measures the average difference between predictions and real values.

Formula:

$$
MAE =
\frac{1}{n}
\sum_{i=1}^{n}
|y_i-\hat{y_i}|
$$

Where:

- $y_i$ = Real value
- $\hat{y_i}$ = Predicted value
- $n$ = Number of observations

---

## Example

Actual temperature:

```

30°C

```

Model prediction:

```

28°C

```

Error:

```

2°C

```

MAE answers:

> "On average, how far are predictions from reality?"

---

# Root Mean Square Error (RMSE)

RMSE is similar to MAE but gives more importance to large errors.

Formula:

$$
RMSE =
\sqrt{
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y_i})^2
}
$$

Useful when large mistakes are especially harmful.

---

## Remote Sensing Example

Predicting land surface temperature:

Actual:

```

35°C

```

Prediction:

```

37°C

```

Error:

```

2°C

```

Metrics tell us how accurate the model is.

---

# 2. Classification Metrics

Used when predicting categories.

Examples:

- Forest
- Water
- Urban

---

# Accuracy

Accuracy is the percentage of correct predictions.

Formula:

$$
Accuracy =
\frac{Correct\ Predictions}
{Total\ Predictions}
$$

Example:

100 pixels classified:

Correct predictions:

```

90 pixels

```

Accuracy:

```

90%

```

---

# Precision

Precision answers:

> "When the model predicts a class, how often is it correct?"

Formula:

$$
Precision =
\frac{True\ Positives}
{True\ Positives + False\ Positives}
$$

Example:

Model predicts:

```

100 forest pixels

```

Actually forest:

```

80 pixels

```

Precision:

```

80%

```

---

# Recall

Recall answers:

> "Of all actual examples of a class, how many did the model find?"

Formula:

$$
Recall =
\frac{True\ Positives}
{True\ Positives + False\ Negatives}
$$

Example:

Actual forest:

```

100 pixels

```

Model detected:

```

90 pixels

```

Recall:

```

90%

```

---

# Confusion Matrix

A confusion matrix shows:

- Correct predictions
- Incorrect predictions
- Which classes are confused

Example:

| | Predicted Forest | Predicted Water |
|-|-|-|
| Actual Forest | 90 | 10 |
| Actual Water | 5 | 95 |

---

# Remote Sensing Example

A classifier may confuse:

```

Forest ↔ Agriculture

```

because both classes may have similar vegetation signals.

A confusion matrix helps us understand these errors.

---

## Explanation

* `train_test_split()` separates learning data and testing data.
* `test_size=0.2` means 20% is used for testing.
* `accuracy_score()` compares predictions with real labels.

In real projects:

1. The model generates predictions.
2. Predictions are compared with actual values.
3. Evaluation metrics are calculated.
