# 2. Core Concept

Before analyzing data or training a Machine Learning model, always ask yourself:

- Is any data missing?
- Are there unusual values?
- Are there errors or duplicates?

This process is called **Data Cleaning**.

Think of it like preparing ingredients before cooking.

> Even the best recipe won't produce a good meal if the ingredients are spoiled.

The three common problems we'll study are:

# A. Cleaning Data

## What is Data Cleaning?

Data cleaning means finding and fixing problems in a dataset before analysis.

### Common Data Cleaning Tasks

- Remove duplicate records
- Correct incorrect values
- Standardize formats
- Handle missing values
- Detect outliers

The goal is to make the dataset **accurate**, **consistent**, and **reliable**.

---

## Everyday Example

### Original Dataset

| Name | Age |
|------|----:|
| Ali | 20 |
| Sara | 21 |
| Ali | 20 |
| Reza | -5 |

### Problems

- 🔁 **Ali appears twice** (duplicate).
- ❌ **Age = -5** is impossible.

After cleaning:

| Name | Age |
|------|----:|
| Ali | 20 |
| Sara | 21 |
| Reza | *Corrected or Removed* |

---

## Remote Sensing Example

A satellite dataset may contain:

- ☁️ Cloud-covered pixels
- 🔁 Duplicate observations
- 🌡️ Invalid temperature values
- 📅 Different date formats

Cleaning these issues improves the quality of later analyses.

---

# B. Outliers

## What is an Outlier?

An **outlier** is a value that is very different from most other values.

### Example

```
21, 22, 20, 23, 22, 105
```

The value **105** is an outlier.

> An outlier does **not** automatically mean the value is wrong.

Sometimes an outlier is:

- ❌ A measurement error
- ✅ A rare but real observation

---

## Everyday Example

Student heights:

```
165, 170, 168, 172, 260
```

The value **260 cm** is unusual.

Possible explanations:

- A typing mistake (`26 → 260`)
- A very rare but correct measurement

> **Always investigate before removing an outlier.**

---

## Remote Sensing Example

Suppose most land surface temperatures are:

```
22°C
24°C
25°C
23°C
24°C
```

One pixel reports:

```
95°C
```

Possible explanations:

- Sensor error
- Wildfire
- Volcanic activity

An outlier may represent an important environmental event rather than a mistake.

# C. Missing Values

## What are Missing Values?

Sometimes data are unavailable.

Instead of a normal value, you may see:

- `NaN`
- `NULL`
- `None`
- *(blank cell)*

These are called **Missing Values**.

---

## Why Do Values Go Missing?

Common causes include:

- Sensor failure
- Cloud cover
- Data transmission problems
- Human error during data entry

---

## Everyday Example

| Student | Score |
|---------|------:|
| Ali | 85 |
| Sara | NaN |
| Reza | 90 |

Sara's score is missing.

---

## Remote Sensing Example

Clouds often block the Earth's surface.

As a result, some satellite pixels contain **no valid measurement**.

---

## How Can We Handle Missing Values?

### 1. Remove Missing Data

Use this when only a small number of values are missing.

---

### 2. Fill (Impute) Missing Values

Replace missing values using statistics such as:

- Mean
- Median
- Mode

Example:

Replace missing temperatures with the **average temperature**.

---

### 3. Leave Them Missing

Some Machine Learning algorithms can work directly with missing values.

The best approach depends on the dataset and the problem you're solving.

---

# 3. Simple Example

Imagine a survey of five people's ages:

```
20
22
21
NaN
120
```

### Problems

- `NaN` → Missing value
- `120` → Possible outlier

Before calculating statistics or training a model, you should decide how to handle these values.

# 7. Mini Coding Example

```python
import pandas as pd
import numpy as np

# Create a simple dataset
data = pd.DataFrame({
    "Temperature": [22, 24, np.nan, 23, 95]
})

# Check for missing values
print(data.isna())

# Count missing values
print(data.isna().sum())

# Fill missing values with the mean
data["Temperature"] = data["Temperature"].fillna(
    data["Temperature"].mean()
)

print(data)
```

## Explanation

- `import pandas as pd` imports the **Pandas** library.
- `import numpy as np` imports **NumPy**.
- `np.nan` represents a missing value.
- `isna()` identifies missing values.
- `isna().sum()` counts how many values are missing.
- `fillna()` replaces missing values with the column mean.

> **Note:** Filling with the mean is a common strategy, but it isn't always the best choice. Depending on the dataset, using the **median**, **interpolation**, or another method may produce better results.
