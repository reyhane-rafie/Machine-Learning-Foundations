## 🎯 Goal

Learn how to prepare data so Machine Learning models can learn **more effectively**.

# 1. Why Should I Care?

A Machine Learning model doesn't automatically know which information is useful or how to interpret it.

The way you **represent your data** has a huge impact on model performance.

In **Remote Sensing**, raw satellite bands are often transformed into more meaningful features such as:

- 🌿 NDVI (Normalized Difference Vegetation Index)
- 💧 NDWI (Normalized Difference Water Index)
- ⛰️ Terrain indices

These engineered features often improve model accuracy more than using the raw satellite bands alone.

---

# 2. Core Concept

Before training a Machine Learning model, we often transform the data into a better form.

This preparation usually includes three important steps:

| Step | Purpose |
|------|---------|
| **Feature Engineering** | Create useful input features from existing data |
| **Encoding** | Convert text (categories) into numbers |
| **Scaling** | Put numerical features on similar scales |

Think of it like preparing ingredients before cooking.

> Better-prepared ingredients usually lead to a better final dish.

---

# A. Feature Engineering

## What is a Feature?

A **feature** is an **input variable** used by a Machine Learning model.

### Examples

| Feature | Example |
|----------|---------|
| Age | 25 |
| Temperature | 22°C |
| Rainfall | 120 mm |
| NDVI | 0.74 |

---

## What is Feature Engineering?

**Feature Engineering** is the process of creating **new, more informative features** from existing data.

The goal is to give the Machine Learning model information that is easier to learn from.

---

## Everyday Example

Suppose you have:

| Height (cm) | Weight (kg) |
|-------------|------------:|
| 170 | 65 |

Instead of using only height and weight, you create a new feature:

**BMI (Body Mass Index)**

BMI often provides more useful information than height or weight alone.

---

## Remote Sensing Example

Suppose a satellite provides:

- Red band
- Near-Infrared (NIR) band

Instead of using these raw bands directly, we calculate the **Normalized Difference Vegetation Index (NDVI)**.

$$
NDVI = \frac{NIR - Red}{NIR + Red}
$$

NDVI is an engineered feature that better represents vegetation health.

Other common engineered features include:

- 🌿 NDVI
- 💧 NDWI (Water Index)
- 🌱 EVI (Enhanced Vegetation Index)
- ⛰️ Slope
- 🧭 Aspect
- 🌊 Distance to rivers

These engineered features often improve Machine Learning performance.

---

# B. Encoding

## Why Do We Need Encoding?

Machine Learning models work with **numbers**, not words.

Encoding converts **categorical data** into numerical values.

---

## Example

### Original Data

| Land Cover |
|------------|
| Forest |
| Water |
| Urban |

A Machine Learning model cannot directly understand these text labels.

---

## Method 1 — Label Encoding

Assign a unique number to each category.

| Land Cover | Code |
|------------|-----:|
| Forest | 0 |
| Water | 1 |
| Urban | 2 |

### Advantages

- Simple
- Memory efficient

> **Important:** These numbers are **labels only**. They do **not** mean that *Urban* is greater than *Water* or *Forest*.

---

## Method 2 — One-Hot Encoding

Instead of one column, create one column for each category.

| Forest | Water | Urban |
|-------:|------:|------:|
| 1 | 0 | 0 |
| 0 | 1 | 0 |
| 0 | 0 | 1 |

This avoids implying any order between categories.

---

## Everyday Example

Favorite fruit:

- 🍎 Apple
- 🍌 Banana
- 🍊 Orange

One-Hot Encoding creates one column for each fruit.

---

## Remote Sensing Example

Land-cover classes:

- Forest
- Water
- Agriculture
- Urban

These categories are usually encoded before training a land-cover classification model.

---

# C. Scaling

## Why Do We Need Scaling?

Different numerical features often have very different ranges.

### Example

| Feature | Value |
|----------|------:|
| Age | 25 |
| Income | 50,000 |

Without scaling, the **Income** feature may dominate simply because its values are much larger.

Scaling makes numerical features comparable by putting them on similar ranges.

---

## Method 1 — Min-Max Scaling

Min-Max Scaling rescales values to a fixed range, usually **0 to 1**.

The formula is

$$
x_{scaled}=
\frac{x-x_{min}}
{x_{max}-x_{min}}
$$

### Example

| Original | Scaled |
|----------:|-------:|
| 20 | 0.00 |
| 40 | 0.25 |
| 60 | 0.50 |
| 80 | 0.75 |
| 100 | 1.00 |

This method is useful when the minimum and maximum values are known.

---

## Method 2 — Standardization (Z-score Scaling)

Standardization transforms data so that:

- Mean = **0**
- Standard Deviation = **1**

The formula is

$$
z=
\frac{x-\mu}{\sigma}
$$

where

- $\mu$ = Mean
- $\sigma$ = Standard Deviation

Unlike Min-Max Scaling, standardized values are **not restricted** to the range 0–1.

This is one of the most common scaling methods in Machine Learning.

---

## Everyday Example

Suppose you're comparing:

- Height (cm)
- Annual income ($)

Without scaling, income could dominate because it has much larger numbers.

Scaling helps both features contribute more fairly.

---

## Remote Sensing Example

Suppose a model uses:

| Feature | Typical Range |
|----------|--------------:|
| NDVI | 0–1 |
| Elevation | 0–3000 m |
| Rainfall | 0–2000 mm |

Without scaling, **Elevation** and **Rainfall** could dominate the learning process simply because of their larger numerical values.

Scaling helps the model treat all features more fairly.

# 4. Remote Sensing / Environmental Example

Suppose you're predicting **drought risk**.

### Original Data

- NDVI
- Rainfall
- Temperature
- Elevation
- Land-cover type

Before training the model:

- 🌿 Create new features such as NDVI anomalies or temperature differences (**Feature Engineering**).
- 🏷️ Encode land-cover categories (**Encoding**).
- 📏 Scale rainfall, elevation, and temperature (**Scaling**).

These preprocessing steps often improve model performance.

# 7. Mini Coding Example

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, MinMaxScaler

# Create a simple dataset
data = pd.DataFrame({
    "LandCover": ["Forest", "Water", "Urban"],
    "Elevation": [250, 1200, 600]
})

# Label Encoding
encoder = LabelEncoder()
data["LandCover_Code"] = encoder.fit_transform(data["LandCover"])

# Min-Max Scaling
scaler = MinMaxScaler()
data["Elevation_Scaled"] = scaler.fit_transform(
    data[["Elevation"]]
)

print(data)
```

## Explanation

- `LabelEncoder()` converts text categories into numbers.
- `fit_transform()` learns the categories and encodes them.
- `MinMaxScaler()` rescales numerical values to the range **0–1**.
- The dataset now contains numerical values that many Machine Learning models can use.

> **Note:** Scaling is especially important for algorithms that rely on distances or gradients, such as **K-Nearest Neighbors (KNN)**, **Support Vector Machines (SVMs)**, and **Neural Networks**.
