# Why Should I Care About How Python Executes Code?

When your program does not work, one of the first questions you should ask is:

> "What was Python doing when it stopped?"

To answer this question, you need to understand how Python reads and executes your code.

This knowledge will make debugging much easier because you can follow the computer's actions step by step.

---

# Core Concept

Python usually executes code:

```
From top to bottom

One line at a time
```

Think of Python as someone reading a recipe.

It does not skip steps or guess what you mean.

It follows:

1. First instruction.
2. Second instruction.
3. Third instruction.
4. Continue until the program finishes or an error occurs.

---

# Everyday Example

Imagine someone gives you these instructions:

1. Put water in a kettle.
2. Turn on the kettle.
3. Pour the water into a cup.

You would not start with step 3.

The steps must happen in the correct order.

Python works the same way.

---

# Simple Python Example

Consider this program:

```python
name = "Rey"

print(name)

print("Welcome!")
```

Python executes it step by step.

---

## Step 1

```python
name = "Rey"
```

Python stores the value `"Rey"` in memory.

Memory:

```
name → "Rey"
```

---

## Step 2

```python
print(name)
```

Python looks for the variable:

```
name
```

It finds:

```
"Rey"
```

Output:

```
Rey
```

---

## Step 3

```python
print("Welcome!")
```

Python displays:

```
Welcome!
```

The program finishes.

---

# What Happens If There Is an Error?

Consider this program:

```python
print(name)

name = "Rey"
```

Python starts from the first line.

It asks:

> "What is name?"

At this moment, the variable does not exist.

Python cannot find it.

So it stops and reports an error.

It does not continue to the second line and think:

> "Maybe the user creates it later."

Computers do not guess.

The order of instructions matters.

---

# Remote Sensing Example

Suppose you create a workflow for calculating NDVI.

Incorrect order:

```
Display NDVI image

        ↓

Calculate NDVI
```

This does not make sense.

The computer cannot display a result that has not been created yet.

Correct order:

```
Load satellite image

        ↓

Select Red and NIR bands

        ↓

Calculate NDVI

        ↓

Display NDVI image
```

The sequence of operations is essential in programming and Remote Sensing workflows.

# Predicting Code Execution

A useful programming skill is learning to ask:

> "What will Python do next?"

Example:

```python
temperature = 30

print(temperature)

temperature = 35

print(temperature)
```

Execution:

Step 1:

```python
temperature = 30
```

Memory:

```
temperature → 30
```

Step 2:

```python
print(temperature)
```

Output:

```
30
```

Step 3:

```python
temperature = 35
```

The value changes.

Memory:

```
temperature → 35
```

Step 4:

```python
print(temperature)
```

Output:

```
35
```

Final output:

```
30
35
```

# Key Takeaways

- Python normally executes code from top to bottom.
- Python runs one statement at a time.
- Variables usually need to exist before they are used.
- Errors stop normal execution.
- The order of instructions matters.
- Predicting what Python will do next is an important programming skill.

# Common Beginner Mistakes

## Mistake 1: Thinking Python Reads the Entire Program Before Executing

❌ Incorrect:

> Python understands the whole file first and then runs it.

✅ Correct:

Python usually executes instructions step by step.

---

## Mistake 2: Using a Variable Before Creating It

Example:

```python
print(age)

age = 22
```

Problem:

The variable `age` does not exist when Python reaches the first line.

Correct order:

```python
age = 22

print(age)
```

---

## Mistake 3: Thinking Python Can Guess Your Intention

If you write code in the wrong order, Python will not rearrange it.

The computer only follows the instructions you provide.
