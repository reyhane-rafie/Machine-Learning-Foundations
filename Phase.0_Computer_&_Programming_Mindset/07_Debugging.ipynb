# Why Should I Care About Debugging?

No programmer writes perfect code.

Even experienced programmers make mistakes every day.

The difference is how they think about mistakes:

Beginners often think:

> "I failed."

Experienced programmers think:

> "I found something I need to fix."

Debugging is the process of **finding and fixing mistakes in a program**.

It is not a sign of failure.

It is a normal part of programming.

---

# Core Concept

A **bug** is a mistake or error in a program.

**Debugging** is the process of finding and fixing that mistake.

Think of debugging like visiting a doctor.

## Example:

Patient:

> "My stomach hurts."

The doctor does not panic.

Instead, the doctor investigates:

- Where does it hurt?
- When did it start?
- What did you eat?
- Are there other symptoms?

Programming works the same way.

The computer tells you:

> "Something is wrong."

Your job is to investigate why.

---

# Everyday Example

Imagine you are baking a cake.

Instead of being soft and fluffy, the cake is burnt.

You do not conclude:

> "I am terrible at baking."

You investigate the possible causes:

- Oven temperature was too high.
- The cake was left in too long.
- The wrong ingredients were used.
- An important step was skipped.

Finding and fixing the cause is debugging.

---

# Remote Sensing Example

Suppose your NDVI map appears completely black.

Do you immediately assume:

> "Google Earth Engine is broken."

No.

You investigate possible problems.

Questions you might ask:

- Did I load the correct satellite image?
- Did I select the correct bands?
- Is my study area correct?
- Is cloud cover affecting the image?
- Did I make a mistake in the NDVI calculation?

Finding the problem is debugging.

# A Simple Debugging Strategy

When something does not work, follow these steps.

---

## Step 1: What Did I Expect?

Ask:

> What should have happened?

Example:

```
I expected the program to print 25.
```

---

## Step 2: What Actually Happened?

Compare your expectation with reality.

Examples:

```
It printed 52.
```

or

```
The program crashed.
```

---

## Step 3: Where Could the Mistake Be?

Check the different parts of the program.

Remember the IPO model:

```
Input
 ↓
Process
 ↓
Output
```

Possible problems:

### Input

Did the program receive the correct information?

Example:

```
Wrong file loaded
```

---

### Process

Did the program perform the correct operation?

Example:

```
Wrong formula used
```

---

### Output

Was the result displayed correctly?

Example:

```
Wrong visualization settings
```

---

## Step 4: Fix One Thing

Change one part of the program.

Do not change ten things at once.

Then test again.

This helps you understand what actually fixed the problem.

---

# Key Takeaways

- Bugs are normal in programming.
- Debugging is the process of solving problems.
- Errors are information that helps you improve your code.
- Always read the error message before changing code.
- Change one thing at a time.
- Every bug teaches you something new.

# Common Beginner Mistakes

## Mistake 1: Changing Many Things at Once

❌ Incorrect:

Changing five parts of the code and running it again.

Problem:

You will not know which change fixed the issue.

✅ Better:

Change one thing, test, and observe the result.

---

## Mistake 2: Ignoring Error Messages

❌ Incorrect:

Ignoring the error and randomly changing code.

✅ Better:

Read the error message carefully.

Most error messages provide clues about the problem.

---

## Mistake 3: Thinking "I'm Not Good at Programming"

Errors do not mean you are bad at programming.

Errors are part of the learning process.

Every programmer spends time debugging.

Remember this throughout your programming journey:

Programming is not about never making mistakes.

When you encounter an error, replace:

> "I'm terrible at this."

with:

> "Interesting. Let's investigate what the program is trying to tell me."
