# Why Should I Learn Programming?

Programming is the process of telling a computer **how to solve a problem**.

As you progress in **Machine Learning**, **Remote Sensing**, and **GIS**, you will write programs to automate tasks such as:

- Calculating **NDVI** from satellite imagery.
- Detecting water bodies.
- Predicting drought using machine learning models.
- Processing thousands of satellite images automatically.

Without programming, these tasks would need to be completed manually, making many real-world projects impractical or impossible.

---

# Core Concept

A computer is **extremely fast**, but it is **not intelligent**.

It does **not** understand your intentions or fill in missing details.

A computer only follows the instructions that you provide—exactly as they are written.

Programming is the process of writing those instructions.

---

# Programming Is Like Writing a Recipe

Think of programming as writing a cooking recipe.

### Example

1. Wash the rice.
2. Add water.
3. Turn on the stove.
4. Wait 20 minutes.
5. Serve the rice.

If one step is missing or unclear, the recipe may fail.

Programs work exactly the same way.

---

# An Important Idea

Programming is **not simply writing code**.

Programming is the process of:

> Breaking a problem into small, logical, and precise steps that a computer can execute.

Code is simply the language we use to communicate those steps to the computer.

Good programmers spend most of their time solving problems—not memorizing syntax.

---

# Everyday Example

Imagine asking a friend:

> "Make me a cup of tea."

A person already understands what making tea involves.

A computer or robot does not.

Instead, you must provide every step.

Example:

1. Pick up the kettle.
2. Fill it with water.
3. Boil the water.
4. Place a tea bag inside a cup.
5. Pour the hot water into the cup.
6. Wait three minutes.
7. Remove the tea bag.

Humans naturally fill in missing details.

Computers never do.

---

# Remote Sensing Example

Suppose you have **5,000 Sentinel-2 satellite images**.

Your goal is to calculate the **NDVI** for every image.

### Manual Approach

- Open an image.
- Calculate NDVI.
- Save the result.
- Repeat 5,000 times.

This is slow, repetitive, and inefficient.

### Programming Approach

```
For each image:
    Read the image
    Calculate NDVI
    Save the result
```

The computer repeats these instructions automatically until all images have been processed.

This ability to automate repetitive work is one of the biggest advantages of programming.

# Your First Python Program

```python
print("Hello, Rey!")
```

### Output

```text
Hello, Rey!
```

### Explanation

- `print()` is a Python function that displays information on the screen.
- `"Hello, Rey!"` is the text that will be displayed.
- Python reads the instruction and executes it exactly as written.

Even this single line is a program because you are giving the computer a clear instruction to perform.

---

# Key Takeaways

- A computer only follows instructions.
- Programming is the process of writing those instructions.
- Computers never guess what you mean.
- Problem-solving is more important than memorizing code.
- Code is simply a language used to express a solution.

# Summary

Programming is the skill of translating a solution into precise instructions that a computer can execute.

Computers are incredibly fast, but they only do exactly what they are told. Because of this, successful programming depends more on careful problem-solving than on writing code itself.

As you continue learning Python, you'll discover that programming allows you to automate repetitive tasks, analyze large datasets, and build powerful applications—skills that are especially valuable in Remote Sensing, GIS, and Machine Learning.
