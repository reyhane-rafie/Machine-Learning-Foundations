# Why Should I Care About Memory?

Whenever you create a variable, the computer needs a place to store its value.

Examples of information that need to be stored while a program is running:

- Your name
- A satellite image
- Today's temperature
- An NDVI value
- A machine learning model

All of these require computer memory.

Without memory, variables would not exist because the computer would have nowhere to keep their values.

---

# Core Concept

Memory is the computer's **temporary workspace**.

When a program runs, the computer usually follows these steps:

1. Receives information.
2. Stores information in memory.
3. Processes the information.
4. Produces a result.

Memory is the place where the computer keeps the data it is currently working with.

---

# Memory as a Workspace

Think of memory like a desk.

Imagine solving a math problem.

Your desk may contain:

- A notebook
- A calculator
- A pencil
- Scratch paper

The desk is your workspace where you keep things while solving the problem.

A computer's memory works in a similar way.

It stores the information that a program currently needs.

---

# Everyday Example

Imagine you are baking a cake.

You place these ingredients on the kitchen counter:

- Eggs
- Flour
- Sugar
- Butter

The counter is your temporary working area.

After finishing the cake, you put everything away.

Memory works similarly:

While the program is running:

```
Data is stored in memory (temporary workspace)
```

After the program finishes:

```
Temporary memory is cleared
```

---

# Remote Sensing Example

Suppose you are calculating NDVI for a satellite image.

During the calculation, the program may temporarily store:

- Red band data
- Near-infrared (NIR) band data
- NDVI results

Example:

```text
Red Band  → stored in memory

NIR Band  → stored in memory

NDVI     → calculated and stored in memory
```

The computer keeps these values available while performing the calculation.

After the program finishes, these temporary values can be removed from memory.

# Variables and Memory

When you create a variable:

```python
temperature = 32
```

Python stores the value `32` somewhere in memory.

Example:

| Variable | Value |
|----------|-------|
| temperature | 32 |
| city | "Isfahan" |

The variable name allows you to access the value stored in memory.

---

# Tiny Python Example

```python
temperature = 32

city = "Isfahan"
```

Conceptually, the computer stores something like this:

```
Memory:

temperature → 32

city → "Isfahan"
```

You do not need to know exactly how the computer organizes memory yet.

The important idea is:

> Variables are stored in memory while the program is running.

---

# Memory vs Permanent Storage

A common beginner mistake is confusing memory with storage.

## Memory (RAM)

- Temporary workspace.
- Used while programs are running.
- Data is usually removed when the program closes.

Example:

```
Running Python program
        ↓
Variables stored in RAM
```

---

## Hard Drive / SSD

- Permanent storage.
- Keeps files even after shutting down the computer.

Examples:

- Python files (`.py`)
- Satellite images
- Documents
- Applications

---

# Common Beginner Mistakes

## Mistake 1: Thinking memory is the same as the hard drive

❌ Incorrect:

> Memory permanently stores files.

✅ Correct:

- Hard drive/SSD stores files permanently.
- Memory (RAM) stores data temporarily while programs run.

---

## Mistake 2: Thinking variables are stored inside your Python file

❌ Incorrect:

> Variables are saved inside the `.py` file.

✅ Correct:

Your Python file contains instructions.

When you run the program, Python creates variables in memory.

Example:

Python file:

```python
temperature = 32
```

During execution:

```
RAM:

temperature → 32
```

---

## Mistake 3: Believing memory keeps everything forever

Memory is temporary.

Usually, when a program ends:

- The memory used by the program is released.
- Other programs can use that memory.

---

# Why Memory Matters in Remote Sensing

Remote sensing often involves very large datasets.

Examples:

- High-resolution satellite images.
- Time series of satellite observations.
- Machine learning training datasets.

Larger datasets require more memory.

A single high-resolution satellite image may require hundreds of megabytes or even several gigabytes of memory during processing.

This is one reason cloud platforms such as **Google Earth Engine** are powerful.

Instead of processing everything on your own computer, much of the computation happens on remote servers, allowing you to work with datasets that may not fit easily into your computer's memory.

---

# Key Takeaways

- Memory is the computer's temporary workspace.
- Variables are stored in memory while programs run.
- Programs read from and write to memory.
- Memory is different from permanent storage.
- Large datasets require more memory.
- Satellite imagery and machine learning often need significant memory resources.
