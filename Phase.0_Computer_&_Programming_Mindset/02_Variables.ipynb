# Why Should I Care About Variables?

Imagine you are calculating the average temperature of a city.

Instead of trying to remember the value:

```
34.6
```

you can give it a meaningful name:

```text
temperature
```

Now, whenever you use `temperature`, the computer knows that you are referring to the value `34.6`.

In Machine Learning and Remote Sensing, you will constantly work with different types of information, such as:

- NDVI values
- Rainfall measurements
- Temperature data
- Latitude and longitude coordinates
- Number of trees
- Water quality measurements

These values need names so that your program can store, access, and process them.

---

# Core Concept

A **variable** is a named place that stores a value.

Think of a variable like a labeled box.

```
📦 Box → stores a value

🏷️ Label → variable name
```

Example:

```
Variable name: temperature

Stored value:
34.6
```

Whenever you use `temperature`, the computer looks inside that "box" and retrieves the stored value.

---

# Important Concept

The variable name is **not the value itself**.

The variable is only a label that points to the stored information.

Example:

```
temperature → 34.6

age → 22

city → "Isfahan"
```

The name helps us understand what the value represents.

---

# Everyday Example

Imagine your kitchen.

Instead of saying:

> "Bring me the sugar that is inside the second cabinet, on the third shelf, inside the blue container."

You simply say:

> "Bring me the sugar."

The word **sugar** is a label that helps you identify something.

Variables work in the same way.

Instead of remembering values directly, we give them meaningful names.

# Remote Sensing Example

Suppose you are calculating NDVI from a satellite image.

Your program may use variables such as:

```python
red_band

nir_band

ndvi
```

The computer does not understand what NDVI means.

It only knows:

> There is a variable called `ndvi`, and it stores a numerical value.

Using meaningful variable names makes code easier to understand.

Compare these two examples:

### Without meaningful variables

```python
0.73 = (0.82 - 0.31) / (0.82 + 0.31)
```

### With meaningful variables

```python
ndvi = (nir_band - red_band) / (nir_band + red_band)
```

The second example is much easier to read because the variable names explain the purpose of each value.

# Creating Variables in Python

In Python, creating a variable is simple:

```python
variable_name = value
```

Example:

```python
name = "Rey"

age = 23

country = "Iran"
```

### Explanation

```python
name = "Rey"
```

The variable `name` stores the text `"Rey"`.

```python
age = 23
```

The variable `age` stores the number `23`.

```python
country = "Iran"
```

The variable `country` stores the text `"Iran"`.

At this stage, you are not performing calculations. You are simply storing information so that you can use it later.

# Key Takeaways

- A variable has a name and a value.
- The variable name should describe what the value represents.
- Meaningful names make programs easier to read and maintain.
- The computer stores the value, and you access it using the variable name.
