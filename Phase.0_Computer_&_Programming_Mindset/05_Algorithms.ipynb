# Why Should I Care About Algorithms?

Before writing any code, programmers ask:

> "What steps should the computer follow to solve this problem?"

Those steps are called an **algorithm**.

Whether you are:

- Finding the largest number.
- Calculating NDVI.
- Detecting floods.
- Training a machine learning model.

You first need a clear method for solving the problem.

The algorithm comes before the code.

---

# Core Concept

An **algorithm** is a clear, ordered sequence of steps used to solve a problem or complete a task.

A good algorithm has three important characteristics:

## 1. Clear

Every step should be easy to understand.

Example:

```
Compare two temperatures.
```

is clearer than:

```
Analyze the data.
```

---

## 2. Ordered

The steps must follow the correct sequence.

Example:

Incorrect:

```
1. Serve the tea.
2. Boil water.
3. Put tea in the cup.
```

Correct:

```
1. Boil water.
2. Put tea in the cup.
3. Pour water.
4. Serve the tea.
```

---

## 3. Finite

The algorithm must eventually finish.

A good algorithm cannot continue forever.

---

# Algorithm vs Programming Language

Think of an algorithm as a recipe.

The recipe describes:

- What steps to follow.
- The order of the steps.
- The goal.

Python is simply the language used to write those instructions.

Example:

```
Algorithm:
Calculate the average temperature.

Python:
Code that performs the calculation.
```

The algorithm is the idea.

The code is the implementation.

---

# Everyday Example

## Problem:

Find the tallest person in a room.

## Algorithm:

1. Look at the first person.
2. Assume they are the tallest.
3. Compare them with the next person.
4. If the next person is taller, remember the new person.
5. Repeat until everyone has been checked.
6. Announce the tallest person.

Notice:

No Python code was used.

This is pure problem-solving.

---

# Remote Sensing Example

## Problem:

Calculate the average temperature from 100 satellite images.

## Algorithm:

1. Load all satellite images.
2. Read the temperature values.
3. Add all temperature values together.
4. Count the number of temperature values.
5. Divide the total temperature by the number of values.
6. Display the average temperature.

Later, Python can automate these steps.

---

# Algorithm and Python Example

Suppose we want to calculate the average of three numbers.

## Algorithm:

1. Read three numbers.
2. Add them together.
3. Divide the sum by 3.
4. Display the result.

## Python Implementation:

```python
a = 10
b = 20
c = 30

average = (a + b + c) / 3

print(average)
```

Output:

```text
20
```

Important idea:

```
Algorithm → Code
```

The algorithm comes first.

The code simply follows the algorithm.

---

# Key Takeaways

- An algorithm is a step-by-step solution to a problem.
- Algorithms are independent of programming languages.
- Good algorithms are clear, ordered, and finite.
- Every computer program follows an algorithm.
- Designing a good algorithm is often more important than writing code quickly.

# Common Beginner Mistakes

## Mistake 1: Thinking an algorithm is the same as code

❌ Incorrect:

> Algorithm = code

✅ Correct:

```
Algorithm = the plan or idea

Code = the implementation of that plan
```

Example:

Algorithm:

```
Calculate NDVI from satellite bands.
```

Code:

```python
ndvi = (nir - red) / (nir + red)
```

---

## Mistake 2: Writing vague steps

Weak algorithm:

```
Process the data.
```

Better algorithm:

```
Compare each temperature value with the current maximum temperature.
```

Clear steps create better algorithms.

---

## Mistake 3: Assuming there is only one correct algorithm

Many problems can have multiple solutions.

Some algorithms are simply:

- Faster.
- More efficient.
- Easier to understand.

Choosing the best algorithm depends on the problem.

# Final Insight: Algorithms in Machine Learning

Machine Learning methods are also called algorithms.

Examples:

- Linear Regression.
- Decision Trees.
- Random Forests.

They are called algorithms because they are step-by-step methods designed to learn patterns from data.

When you study Machine Learning, think:

> "I am learning different algorithms designed to solve different types of problems."

Machine Learning is not magic.

It is a collection of carefully designed methods for finding patterns and making predictions from data.
