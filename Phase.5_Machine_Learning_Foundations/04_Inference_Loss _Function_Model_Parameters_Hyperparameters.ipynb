---

# 2. Core Concept

Let's imagine we have a model predicting **soil moisture** from satellite data.

The basic learning process is:

```text
Satellite Data
      ↓
    Model
      ↓
  Prediction
      ↓
Compare with Actual Value
      ↓
    Loss
      ↓
Update Parameters
      ↓
 Better Model
```

Now let's understand each part.

---

# A. Model

A **model** is the mathematical system that learns a relationship between inputs and outputs.

For example:

```text
NDVI + Rainfall + Temperature
              ↓
            Model
              ↓
        Soil Moisture
```

The model tries to learn:

> "How are these inputs related to soil moisture?"

A model can be simple, such as a linear equation, or complex, such as a neural network.

---

# B. Inference

**Inference** means using a trained model to make a prediction.

During training:

```text
Data
 ↓
Model
 ↓
Prediction
 ↓
Loss
 ↓
Learning
```

After training:

```text
New Data
   ↓
Trained Model
   ↓
Prediction
```

That prediction process is called **inference**.

---

## Example

You trained a model using thousands of satellite observations.

Now you give it a new satellite observation.

The model predicts:

```text
Soil Moisture = 0.31
```

This is **inference**.

---

# C. Loss Function

A **loss function** measures how wrong the model's prediction is.

Suppose:

```text
Actual Soil Moisture = 0.40

Predicted Soil Moisture = 0.31
```

The model made an error.

The loss function converts the model's error into a numerical value.

```text
Small Loss → Better Prediction

Large Loss → Worse Prediction
```

The general goal of training is:

> **Minimize the loss.**

---

## Everyday Example

Imagine throwing darts at a target.

```text
Close to Target
      ↓
 Small Error
```

```text
Far from Target
      ↓
 Large Error
```

The loss function tells the model:

> "How far were you from the correct answer?"

---

# D. Parameters

**Parameters are values that the model learns from the training data.**

Consider the simple equation:

$$
y = wx + b
$$

Where:

| Symbol | Meaning |
|---|---|
| `x` | Input |
| `y` | Prediction |
| `w` | Parameter |
| `b` | Parameter |

During training, the model changes `w` and `b` to reduce the loss.

For example:

```text
Before Training:

w = 0.5
b = 2
```

After training:

```text
w = 1.7
b = 0.8
```

The model **learned** these values from the training data.

---

# E. Hyperparameters

**Hyperparameters are settings that we choose for the Machine Learning algorithm.**

Unlike parameters, hyperparameters are **not normally learned directly from the training data** during the model's parameter-learning process.

Examples include:

- Learning rate
- Number of trees in Random Forest
- Maximum tree depth
- Number of neural-network layers
- Batch size
- Number of training epochs

---

## Everyday Example

Imagine baking a cake.

### Parameters

The parameters are values that are determined by the learning process.

### Hyperparameters

Hyperparameters are settings you choose before or during training, such as:

- Oven temperature
- Baking time
- Amount of ingredients

You choose the settings, and the process produces the final result.

---

# 3. Simple Example

Imagine a student trying to predict exam scores.

## Input

```text
Hours Studied = 5
```

## Model

```text
y = wx + b
```

## Prediction

```text
75
```

## Actual Score

```text
85
```

The prediction is different from the actual value.

```text
Prediction ≠ Actual
        ↓
      Loss
```

During training, the model adjusts its **parameters** (`w` and `b`) to reduce the loss.

Meanwhile, you might choose a **learning rate** that controls how quickly the parameters change.

The learning rate is a **hyperparameter**.

---

## After Training

Once training is finished, you give the model a new input:

```text
Hours Studied = 7
```

The trained model produces a prediction.

This is **inference**.

# 4. Remote Sensing / Environmental Example

Suppose we want to predict **crop yield**.

## Input Features

```text
NDVI
Rainfall
Temperature
Soil Moisture
Elevation
```

---

## Model

The model learns relationships between these variables and crop yield.

```text
Environmental Features
          ↓
        Model
          ↓
      Crop Yield
```

---

## Parameters

The model learns numerical values that determine how strongly different inputs affect its predictions.

---

## Loss Function

The loss function compares:

```text
Predicted Crop Yield
          vs.
Actual Crop Yield
```

It produces a numerical measure of the prediction error.

---

## Training

During training:

```text
Features
   ↓
Model
   ↓
Prediction
   ↓
Loss
   ↓
Parameter Updates
   ↓
Improved Model
```

The model repeatedly adjusts its parameters to reduce the loss.

---

## Hyperparameters

Depending on the algorithm, you might choose settings such as:

```text
Learning Rate = 0.01
Number of Trees = 100
```

These are **hyperparameters**.

---

## Inference

Finally, you provide data from a new agricultural area:

```text
New Satellite + Environmental Data
              ↓
         Trained Model
              ↓
       Predicted Crop Yield
```

This prediction is **inference**.

---

# 5. The Full Learning Loop

This is one of the most important diagrams to remember:

```text
                 TRAINING
                    │
                    ▼
             Input Features
                    │
                    ▼
                  Model
                    │
                    ▼
               Prediction
                    │
                    ▼
          Compare with Actual
                    │
                    ▼
              Loss Function
                    │
                    ▼
            Adjust Parameters
                    │
                    └──────────────┐
                                   │
                                   ▼
                            Repeat Training
```

The model repeats this process during training.

---

## After Training

Once the model has learned useful parameters:

```text
        NEW DATA
           ↓
    Trained Model
           ↓
      Prediction
           ↓
       INFERENCE
```

There is no need to calculate a training loss for the purpose of making each ordinary inference prediction.

---

# 6. The Critical Difference

This distinction is **very important**:

| Concept | Who Determines It? | Purpose |
|---|---|---|
| **Parameters** | Learned by the model | Determine how the model maps inputs to outputs |
| **Hyperparameters** | Chosen by us | Control how the learning algorithm operates |
| **Loss** | Calculated from predictions and targets | Measures model error |
| **Inference** | Performed by the trained model | Produces predictions for new data |
| **Model** | Designed and trained | Learns the relationship between inputs and outputs |

---

## Remember

> **Parameters are learned. Hyperparameters are chosen.**

This distinction will appear constantly when we start building real Machine Learning models.

---

# 7. Important Points

- A **model** learns a relationship between inputs and outputs.
- **Inference** is using a trained model to make predictions on new data.
- A **loss function** measures how wrong the model's prediction is.
- **Parameters** are learned from training data.
- **Hyperparameters** are settings chosen by the practitioner.
- Training generally tries to **reduce the loss by adjusting parameters**.
- A model uses its learned parameters during inference to generate predictions.
- Hyperparameters can strongly affect how well the model learns.

# 9. Mini Coding Example

Here's a tiny example using a simple linear model:

```python
from sklearn.linear_model import LinearRegression

# Training data
X = [[1], [2], [3], [4]]
y = [2, 4, 6, 8]

# Create the model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Make a prediction
prediction = model.predict([[5]])

print(prediction)
```

---

## What's Happening?

### Step 1 — Input Features

```python
X = [[1], [2], [3], [4]]
```

`X` contains the input features.

---

### Step 2 — Correct Answers

```python
y = [2, 4, 6, 8]
```

`y` contains the target values.

---

### Step 3 — Create the Model

```python
model = LinearRegression()
```

We create a Linear Regression model.

---

### Step 4 — Train the Model

```python
model.fit(X, y)
```

The model learns its **parameters** from the training data.

---

### Step 5 — Inference

```python
model.predict([[5]])
```

The trained model uses the new input to make a prediction.

This is **inference**.

The model should predict something close to:

```text
10
```

because the training data follows:

$$
y = 2x
$$
