# Lesson 3 — Features, Labels, Training, and Testing

> **Goal:** Understand the basic structure of a supervised Machine Learning problem: **what goes into the model, what it should predict, how it learns, and how we check it.**

---

# 1. Why Should I Care?

These four concepts are the foundation of almost every supervised Machine Learning project.

If you understand:

```text
Features → Labels → Training → Testing
````

then later topics such as:

* Random Forest
* Support Vector Machines (SVM)
* Neural Networks
* Regression
* Classification

will become much easier.

For Remote Sensing, this is exactly how we turn satellite and environmental data into a Machine Learning problem.

---

# 2. Core Concept

Imagine you want a model to predict whether a satellite pixel is:

* Forest
* Water
* Urban

You need two main things:

## Features — `X`

The information the model uses to make a prediction.

For example:

```text
NDVI
NDWI
Temperature
Elevation
```

These are the **inputs**.

---

## Labels — `y`

The correct answer we want the model to predict.

```text
Forest
Water
Urban
```

These are the **answers**.

---

## Training

We show the model many examples of:

```text
Features → Correct Label
```

The model learns patterns connecting the features to the labels.

---

## Testing

We give the trained model **new examples it has not seen before**.

We then compare its predictions with the true labels.

```text
New Data
   ↓
Trained Model
   ↓
Prediction
   ↓
Compare with True Label
   ↓
Evaluate Performance
```

---

# 3. Simple Example

Imagine teaching someone to recognize animals.

| Weight | Height | Ears    | Animal |
| -----: | -----: | ------- | ------ |
|   5 kg |  30 cm | Pointed | Cat    |
|  20 kg |  50 cm | Floppy  | Dog    |
|   6 kg |  28 cm | Pointed | Cat    |

---

## Features

```text
Weight
Height
Ears
```

These are the information used to make the prediction.

---

## Label

```text
Animal
```

The label is the correct answer.

---

## Training

The learner studies the labeled examples.

For example:

```text
Weight = 5 kg
Height = 30 cm
Ears = Pointed
        ↓
      Cat
```

```text
Weight = 20 kg
Height = 50 cm
Ears = Floppy
        ↓
      Dog
```

The learner tries to discover patterns connecting the features to the correct animal.

---

## Testing

Now you give the model a new animal:

```text
Weight = 5.5 kg
Height = 29 cm
Ears = Pointed
```

The model predicts:

```text
Cat
```

This is prediction on **unseen data**.

---

# 4. Remote Sensing / Environmental Example

Suppose we want to classify satellite pixels.

Our dataset might look like this:

| NDVI | NDWI | Temperature | Elevation | Land Cover  |
| ---: | ---: | ----------: | --------: | ----------- |
| 0.82 | 0.10 |          22 |       450 | Forest      |
| 0.05 | 0.78 |          18 |       120 | Water       |
| 0.12 | 0.02 |          34 |       300 | Urban       |
| 0.65 | 0.15 |          27 |       500 | Agriculture |

---

## Features — `X`

The input features are:

```text
NDVI
NDWI
Temperature
Elevation
```

In Python, one observation could be represented as:

```python
X = [0.82, 0.10, 22, 450]
```

---

## Label — `y`

The target is:

```text
Land Cover
```

For the example above:

```python
y = "Forest"
```

So one complete observation is:

```python
X = [0.82, 0.10, 22, 450]
y = "Forest"
```

Think of it as:

```text
X = What the model sees

y = What the model should predict
```

---

# 5. Training

We give the model many labeled examples:

```text
Satellite Features + Known Land Cover
                 ↓
              ML Model
                 ↓
          Learns Patterns
```

For example, the model might discover that pixels with:

* High NDVI
* Moderate temperature
* Certain elevation ranges

are often associated with forest.

> **Important:** We don't manually tell the model this rule. The model learns the relationship from the training data.

---

# 6. Testing

Now we give the model pixels it **didn't see during training**.

```text
New Satellite Pixel
        ↓
      Model
        ↓
Prediction: Forest
        ↓
Compare with
Actual Label
        ↓
Correct? ✓
```

This tells us whether the model learned a useful pattern rather than simply memorizing the training examples.

---

# 7. Training vs Testing

This distinction is extremely important.

## Training Data

Used to **learn**.

```text
TRAINING DATA
"Learn from these examples."
        ↓
      MODEL
```

---

## Testing Data

Used to **evaluate**.

```text
TESTING DATA
"Now show me what you can do
with examples you haven't seen."
```

A common split is:

```text
80% → Training
20% → Testing
```

However, this is **not a universal rule**.

The appropriate split depends on factors such as:

* Dataset size
* Dataset structure
* Number of samples
* Spatial or temporal relationships
* The Machine Learning problem

In Remote Sensing, random splitting is not always sufficient because nearby pixels can be highly similar.

---

# 8. One Important Idea: `X` and `y`

You'll see this notation constantly in Python and Machine Learning.

## `X`

Usually represents the **features/input data**.

Example:

```python
X = satellite_features
```

---

## `y`

Usually represents the **target/label/output**.

Example:

```python
y = land_cover
```

---

## Easy Way to Remember

> **X = What does the model see?**

> **y = What should the model predict?**

For example:

```text
X
↓
NDVI
NDWI
Temperature
Elevation

        ↓
      Model
        ↓

y
↓
Forest
```

---

# 9. A Complete Machine Learning Workflow

You can now connect the ideas from this lesson:

```text
Raw Data
   ↓
Features (X)
   ↓
Labels (y)
   ↓
Train/Test Split
   ↓
Training Data
   ↓
Machine Learning Model
   ↓
Learn Patterns
   ↓
Testing Data
   ↓
Predictions
   ↓
Compare Predictions with True Labels
   ↓
Evaluate Model
```

For Remote Sensing:

```text
Satellite Images
       ↓
Data Processing
       ↓
Features
(NDVI, NDWI, Temperature, Elevation)
       ↓
Labels
(Forest, Water, Urban, Agriculture)
       ↓
Train/Test Split
       ↓
Train Model
       ↓
Predict New Pixels
       ↓
Evaluate Performance
```

---

# 10. Important Points

* **Features** are the inputs used by the model.
* **Labels** are the correct answers in supervised learning.
* `X` usually represents features.
* `y` usually represents labels or targets.
* **Training** teaches the model using known examples.
* **Testing** evaluates the model using unseen examples.
* A good model should perform well on **new data**, not just its training data.
* A common train/test split is **80% training and 20% testing**, but this is not a universal rule.
* In Remote Sensing, spatial and temporal relationships should be considered when creating training and testing datasets.

---

# 11. Practice Together

## Question 1

Suppose you're predicting **soil moisture** using:

* Rainfall
* Temperature
* NDVI
* Elevation

And your dataset contains the actual measured soil moisture.

Which of these are the **features**?

A. Soil moisture

B. Rainfall, temperature, NDVI, and elevation

C. The Machine Learning model

D. Training and testing data



```
```
