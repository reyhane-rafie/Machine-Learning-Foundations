# Lesson 5 — Overfitting, Underfitting, Bias, and Variance

> **Goal:** Understand why a model can be **too simple or too complicated**, and how **bias and variance** help us think about this problem.

---

## 1. Why should I care?

A good Machine Learning model should learn the **real pattern** in the data, not simply memorize the training examples.

This is especially important in Remote Sensing because satellite datasets can be complex and noisy. A model might perform extremely well on the pixels it trained on but fail when applied to a new image, year, or region.

---

## 2. Core Concept

Imagine you're trying to learn the relationship between **rainfall and vegetation**.

There are three possibilities.

### Underfitting

The model is **too simple**.

It doesn't capture the important patterns in the data.

```text
Model: "More rainfall always means more vegetation."
````

But reality might be more complicated.

**Result:**

> Poor performance on training data **and** testing data.

---

### Good Fit

The model learns the important underlying pattern without simply memorizing individual observations.

**Result:**

> Good performance on both training and unseen data.

---

### Overfitting

The model is **too complicated**.

It learns the real pattern **plus noise and random details** from the training data.

**Result:**

> Excellent training performance but poor testing performance.

---

## A Useful Analogy 🎓

Imagine studying for an exam.

### Underfitting

You barely study.

You don't understand the material.

```text
Practice exam → Poor
Real exam → Poor
```

### Good Fit

You understand the concepts.

```text
Practice exam → Good
Real exam → Good
```

### Overfitting

You memorize the exact practice questions.

```text
Practice exam → Excellent
Real exam → Poor
```

That's the key idea.

---

## 3. Simple Example

Suppose we have data showing the relationship between:

```text
Hours studied → Exam score
```

### Model A — Too Simple

It assumes:

> Every extra hour increases the score by exactly 5 points.

It misses important patterns.

**Underfitting.**

---

### Model B — Appropriate

It captures the general relationship:

> More study usually improves scores, but the relationship isn't perfectly predictable.

**Good fit.**

---

### Model C — Too Complex

The model creates a complicated rule that perfectly follows every student in the training dataset.

It even learns random things like:

> "Students who studied on Tuesday tend to score higher."

That might just be coincidence.

**Overfitting.**

---

## 4. Bias and Variance

Now we introduce two important ideas.

### Bias

**Bias describes how much the model's assumptions are too simplistic.**

A high-bias model tends to miss important patterns.

```text
High Bias
    ↓
Too simple
    ↓
Underfitting
```

**Example:**

Trying to describe a complex environmental relationship using an extremely simple model.

---

### Variance

**Variance describes how sensitive the model is to the particular training data it received.**

A high-variance model changes a lot when the training dataset changes slightly.

```text
High Variance
    ↓
Too sensitive to training data
    ↓
Overfitting
```

---

## 5. The Bias–Variance Trade-off

This is a very important Machine Learning idea.

Generally:

```text
Too Simple                    Too Complex
    ↓                             ↓
High Bias                   High Variance
    ↓                             ↓
Underfitting                 Overfitting
```

We want a balance.

```text
High Bias ←──── Good Balance ────→ High Variance
                         ↑
                      We want
                     around here
```

The goal isn't:

> "Make the model as complicated as possible."

The goal is:

> **Make the model complex enough to learn the real pattern, but not so complex that it memorizes noise.**

---

## 6. Remote Sensing / Environmental Example

Suppose you want to predict **soil moisture** from:

* NDVI
* Temperature
* Rainfall
* Elevation
* Spectral bands

You train a model using satellite observations from 2024.

### Underfitting

Your model is too simple.

It assumes soil moisture depends almost entirely on rainfall.

It ignores:

* Temperature
* Vegetation
* Elevation
* Soil properties

**Result:**

```text
Training performance → Poor
Testing performance  → Poor
```

This is likely **high bias**.

---

### Overfitting

Now imagine using an extremely complex model.

It learns:

* Real relationships
* Sensor noise
* Specific conditions in your training images
* Random patterns unique to 2024

You test it on 2025 satellite data:

```text
Training performance → Excellent
Testing performance  → Poor
```

This is likely **high variance**.

---

### Good Fit

The model learns general relationships between environmental variables and soil moisture.

```text
Training performance → Good
Testing performance  → Good
```

This is what we want.

---

## 7. How Do We Detect Them?

One useful approach is to compare training and testing performance.

| Training Performance | Testing Performance | Likely Problem |
| -------------------- | ------------------- | -------------- |
| Poor                 | Poor                | Underfitting   |
| Excellent            | Poor                | Overfitting    |
| Good                 | Good                | Good fit       |
| Very different       | —                   | Investigate    |

For example:

### Model A

```text
Training accuracy = 60%
Testing accuracy  = 58%
```

Likely **underfitting**.

### Model B

```text
Training accuracy = 99%
Testing accuracy  = 72%
```

Likely **overfitting**.

### Model C

```text
Training accuracy = 92%
Testing accuracy  = 89%
```

Much healthier.

---

## 8. Important Points

* **Underfitting** means the model is too simple to capture important patterns.
* **Overfitting** means the model learns the training data too specifically, including noise.
* **High bias** is commonly associated with underfitting.
* **High variance** is commonly associated with overfitting.
* We want a balance between bias and variance.
* Always evaluate performance on **unseen data**, not just training data.

---

## 9. Practice Together

### Question 1

You train a land-cover classification model.

You get:

```text
Training accuracy = 65%
Testing accuracy  = 63%
```

The model performs poorly on **both** datasets.

Would you suspect:

**A. Overfitting**

**B. Underfitting**

**C. Neither**

Reply with **A, B, or C**.

I'll check your reasoning before giving you Question 2.

---

## 10. Mini Coding Example

We can visualize the idea with polynomial regression.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

These import tools for creating increasingly complex models.

A simple model might look like:

```python
model = LinearRegression()
```

A more complex model can use polynomial features:

```python
poly = PolynomialFeatures(degree=10)
```

Here, `degree=10` allows the model to create a much more complicated relationship.

For example:

```text
degree = 1  → simple model
degree = 2  → more complex
degree = 5  → very complex
degree = 10 → potentially extremely complex
```

If the model becomes too complex, it may start following individual training points instead of learning the general pattern.

That's where **overfitting** can happen.

> You don't need to memorize this code yet. The important concept is that **model complexity can affect underfitting and overfitting**.

---

## 11. Mini Quiz — Don't Answer Yet

### Conceptual

**1.** What is overfitting?

**2.** What is underfitting?

**3.** Explain the difference between bias and variance in your own words.

### Multiple Choice

**4.** A model has:

```text
Training accuracy = 99%
Testing accuracy  = 70%
```

Which is the most likely explanation?

A. High bias / underfitting
B. High variance / overfitting
C. Perfect generalization
D. Missing values

### Practical Scenario

**5.**

You build a Random Forest model to classify forest, agriculture, water, and urban areas from satellite imagery.

You gradually increase the model's complexity:

```text
Model A:
Training accuracy = 75%
Testing accuracy  = 73%

Model B:
Training accuracy = 91%
Testing accuracy  = 88%

Model C:
Training accuracy = 99%
Testing accuracy  = 69%
```

Answer:

1. Which model is likely underfitting?
2. Which model is likely overfitting?
3. Which model would you choose?
4. Explain why.

---

## 12. Ready to Move On?

Before moving on, you should be able to explain this without memorizing definitions:

```text
Too simple
    ↓
High Bias
    ↓
Underfitting


       ↕

  GOOD BALANCE

       ↕


Too sensitive
    ↓
High Variance
    ↓
Overfitting
```

You should also be able to:

* Identify overfitting from training/testing results.
* Identify underfitting from training/testing results.
* Explain bias and variance intuitively.
* Explain why a model with **99% training accuracy isn't necessarily a good model**.
* Connect these ideas to Remote Sensing models and unseen satellite data.

```
```
