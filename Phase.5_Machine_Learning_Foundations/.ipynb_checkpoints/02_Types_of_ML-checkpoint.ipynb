> **Goal:** Understand the three major learning approaches: **Supervised Learning, Unsupervised Learning, and Reinforcement Learning**, and know when each is useful in Remote Sensing and Environmental Monitoring.

---

# 1. Why Should I Care?

Before learning algorithms such as **Linear Regression, Random Forest, K-Means, or Neural Networks**, you need to understand **what kind of learning problem you have**.

The most important question is:

> **Do we have correct answers (labels) for our training data?**

This determines whether we use **Supervised Learning**, **Unsupervised Learning**, or another learning approach.

---

# 2. Core Concept

Machine Learning can learn in different ways depending on the information we provide.

## Supervised Learning

We give the model:

**Data + Correct Answers**

The model learns the relationship between them and predicts answers for new data.

```text
Input → Model → Known Answer
```

Example:

```text
Satellite Image → Model → Forest
```

---

## Unsupervised Learning

We give the model:

**Data without correct answers**

The model tries to discover patterns or groups by itself.

```text
Input → Model → Discovered Groups / Patterns
```

Example:

```text
Satellite Pixels → Model → Group 1, Group 2, Group 3
```

We don't tell the model what those groups mean beforehand.

---

## Reinforcement Learning

An **agent** learns by interacting with an environment.

It:

1. Takes an action.
2. Receives a reward or penalty.
3. Learns which actions are better.
4. Repeats the process.

```text
Environment
     ↓
   Agent
     ↓
   Action
     ↓
Reward / Penalty
     ↓
   Learning
```

This is different from supervised learning because we don't simply give the model the correct answer for every situation.

---

# 3. Simple Example

Imagine teaching a child to recognize fruits.

## Supervised Learning 🍎

You show:

> Apple → "Apple"

> Banana → "Banana"

> Orange → "Orange"

After seeing many labeled examples, the child learns to identify a new fruit.

---

## Unsupervised Learning 🍎🍌

You give the child a basket of mixed fruits and say:

> "Group these however you think makes sense."

The child might create groups based on:

- Color
- Shape
- Size

You never provided the correct group labels.

---

## Reinforcement Learning 🎮

Imagine teaching a child to play a game.

- Good action → Reward ⭐
- Bad action → Penalty ❌

Through repeated attempts, the child learns which actions lead to better outcomes.

---

# 4. Remote Sensing / Environmental Example

## A. Supervised Learning

Suppose you want to classify satellite pixels into:

- Forest
- Water
- Urban
- Agriculture

You have labeled training data:

| NDVI | NDWI | Temperature | Label |
|------|------|-------------|-------|
| 0.75 | 0.20 | 22 | Forest |
| 0.02 | 0.80 | 18 | Water |
| 0.10 | 0.05 | 32 | Urban |
| 0.60 | 0.15 | 27 | Agriculture |

The model learns:

```text
Features → Land-Cover Class
```

This is **Supervised Learning**.

### Common Algorithms

- Linear Regression
- Logistic Regression
- Decision Trees
- Random Forest
- Support Vector Machines (SVM)
- Neural Networks

---

# B. Unsupervised Learning

Now imagine you have millions of satellite pixels but **no land-cover labels**.

You could ask an algorithm to find groups of similar pixels.

For example:

```text
Group 1 → High NDVI, low temperature
Group 2 → Low NDVI, high temperature
Group 3 → Low NDVI, low NDWI
```

You might later interpret these groups as:

```text
Group 1 → Vegetation
Group 2 → Urban / Bare Land
Group 3 → Water
```

The important point is:

> **The algorithm did not know these names beforehand.**

This is **Unsupervised Learning**.

A famous unsupervised learning algorithm is:

## K-Means Clustering

We'll study it later.

---

# C. Reinforcement Learning

Reinforcement Learning is less common in basic satellite-image classification but can be useful for **decision-making and control**.

Imagine an environmental management system controlling water allocation.

```text
Water Reservoir
      ↓
     Agent
      ↓
  Release Water
      ↓
Environmental Conditions
      ↓
Reward / Penalty
      ↓
   Agent Learns
```

The agent might learn a strategy that balances:

- Agricultural demand
- Reservoir levels
- Environmental flows
- Future water availability

Another possible application is **adaptive satellite observation**, where an agent learns which locations or observations are most valuable.

---

# 5. The Most Important Difference

Here's the mental model to remember:

| Type | Do We Have Labels? | What Does the Model Do? |
|------|--------------------|--------------------------|
| **Supervised** | Yes | Learns to predict labels or values |
| **Unsupervised** | No | Finds patterns or groups |
| **Reinforcement** | No direct answers | Learns through rewards |

---

## Think of Them as Three Different Teachers

### Supervised

> "Here are the questions **and** the correct answers. Learn the relationship."

### Unsupervised

> "Here is the data. Find interesting patterns yourself."

### Reinforcement

> "Try things. I'll tell you whether your actions were good or bad."

---

# 6. Important Points

- **Supervised Learning** uses labeled data.
- **Unsupervised Learning** uses unlabeled data to discover patterns.
- **Reinforcement Learning** learns through actions and rewards.
- Land-cover classification with labeled pixels is a **Supervised Learning** problem.
- Clustering unlabeled satellite pixels is an **Unsupervised Learning** problem.
- Decision-making problems can be approached with **Reinforcement Learning**.

# 8. Mini Coding Example

Here's a simple example of **Supervised Learning**:

```python
from sklearn.tree import DecisionTreeClassifier

# Features
X = [
    [0.8, 0.2],
    [0.7, 0.3],
    [0.1, 0.9],
    [0.2, 0.8]
]

# Labels
y = [
    "Forest",
    "Forest",
    "Water",
    "Water"
]

# Create the model
model = DecisionTreeClassifier()

# Train the model
model.fit(X, y)

# Predict a new pixel
prediction = model.predict([[0.75, 0.25]])

print(prediction)
```

---

## What Is Happening?

```text
X = Features
    ↓
[NDVI, NDWI]

y = Correct Answers
    ↓
[Forest, Forest, Water, Water]

       ↓
   ML Model
       ↓
Learns Relationship
       ↓
   New Pixel
[0.75, 0.25]
       ↓
  Prediction
    "Forest"
```

The critical idea is:

> **We provided the correct labels (`y`).**

Therefore, this is **Supervised Learning**.
