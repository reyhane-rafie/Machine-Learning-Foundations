- Artificial Intelligence (AI)
- Machine Learning (ML)
- Deep Learning (DL)
- Data Science (DS)

However, they are different concepts.

For your goal of becoming an **AI researcher in Remote Sensing**, this distinction is important because satellite image analysis combines all of these fields.

---

## How These Fields Work Together

| Field | Main Purpose |
|---|---|
| **Data Science** | Understanding, preparing, and analyzing environmental data |
| **Machine Learning** | Learning patterns from satellite and environmental features |
| **Deep Learning** | Learning complex patterns directly from images |
| **Artificial Intelligence** | Creating intelligent systems that can make decisions |

---

# 2. Core Concept

# Artificial Intelligence (AI)

## Definition

**Artificial Intelligence is the broadest concept.**

It means:

> Creating machines that can perform tasks that normally require human intelligence.

---

## Examples of AI Tasks

- Understanding images
- Recognizing speech
- Making decisions
- Solving problems
- Learning from experience

AI is the **overall goal**.

---

# Machine Learning (ML)

## Definition

Machine Learning is a **subfield of Artificial Intelligence**.

Instead of manually programming every rule, we provide:

- Data
- Examples
- A learning algorithm

The computer discovers patterns automatically.

---

## Traditional Programming

```
Rules + Data
      ↓
    Answer
```

The programmer creates the rules.

---

## Machine Learning

```
Data + Answers
        ↓
 Learning Algorithm
        ↓
 Rules discovered by the model
```

The model learns the rules from examples.

---

## Example

Instead of writing:

```
IF temperature > 35°C
AND rainfall < 200 mm
THEN drought risk = high
```

We provide many examples.

The Machine Learning model discovers the relationship itself.

---

# Deep Learning (DL)

## Definition

Deep Learning is a **subfield of Machine Learning**.

It uses artificial neural networks with many layers.

---

## Deep Learning is Especially Powerful For:

- Images
- Videos
- Audio
- Large datasets

---

## Examples

- Face recognition
- ChatGPT
- Autonomous vehicles
- Satellite image segmentation

---

## Relationship Between AI, ML, and DL

```
Artificial Intelligence
│
└── Machine Learning
        │
        └── Deep Learning
```

Deep Learning is inside Machine Learning, and Machine Learning is inside AI.

---

# Data Science

## Definition

Data Science is a broader field focused on extracting knowledge and value from data.

It combines:

- Statistics
- Programming
- Machine Learning
- Data Visualization
- Domain Knowledge

---

A data scientist asks:

> "What can we learn from this data?"

---

# 3. Simple Example

Imagine a company wants to predict house prices.

---

# Artificial Intelligence

The overall goal:

> Create an intelligent system that can predict house prices.

---

# Machine Learning

The model learns relationships between:

## Input Features

- House size
- Location
- Number of rooms

## Output

- House price

The algorithm discovers patterns from examples.

---

# Deep Learning

A neural network learns from:

- Images of houses
- Satellite images
- Street views

It discovers complex patterns automatically.

---

# Data Science

A data scientist:

- Collects housing data
- Cleans the data
- Visualizes trends
- Builds models
- Explains results

---

# 4. Remote Sensing / Environmental Example

Imagine predicting **forest loss using satellite imagery**.

---

# Data Science

The workflow includes:

- Collecting Sentinel-2 images
- Removing clouds
- Calculating NDVI
- Analyzing trends
- Preparing datasets

Example:

```
Raw satellite images
          ↓
Data cleaning
          ↓
Feature preparation
          ↓
Machine Learning dataset
```

---

# Machine Learning

You train a model using features such as:

## Input

```
NDVI
Rainfall
Temperature
Elevation
Distance from roads
```

## Output

```
Forest loss probability
```

The model learns relationships between environmental factors and forest loss.

---

# Deep Learning

Instead of manually creating features:

Input:

```
Raw satellite image
```

The neural network automatically learns:

- Texture
- Shapes
- Spectral patterns
- Spatial relationships

Examples:

- CNN (Convolutional Neural Network)
- U-Net for image segmentation

---

# Artificial Intelligence

The complete intelligent system:

```
Satellite Images
        ↓
AI System
        ↓
Automatic Deforestation Detection
        ↓
Environmental Management Alert
```

AI combines data, models, and decision-making into a complete system.

---

# 5. Important Points

| Concept | Main Idea |
|---|---|
| **AI** | The broad goal of creating intelligent machines |
| **Machine Learning** | Allows computers to learn patterns from data |
| **Deep Learning** | A powerful type of Machine Learning using neural networks |
| **Data Science** | Extracts knowledge and value from data |
| **Relationship** | AI → Machine Learning → Deep Learning |
| **Remote Sensing** | Uses all four fields together for environmental applications |

# 7. Mini Coding Example

Machine Learning in Python often starts with the idea:

```python
# Input feature
temperature = [20, 25, 30, 35]

# Target/output
vegetation = [0.8, 0.7, 0.5, 0.3]

# A Machine Learning model learns
# the relationship between them
```

---

## Explanation

- `temperature` is an input feature.
- `vegetation` is the target/output.
- A Machine Learning model tries to learn the relationship between them.

Example question:

> "How does vegetation change when temperature increases?"

Later, algorithms will learn these relationships automatically.
