# Why Should You Care?

In real life, we constantly receive **new information** that changes what we believe.

For example:

- What is the probability that it will rain tomorrow?
- Now suppose you see dark clouds. Does that probability change?

Or imagine a satellite image.

- What is the probability that a pixel represents water?
- Now suppose you know the pixel has a **very low NDVI** and is located next to a river. Does that probability change?

The answer is **yes**.

Conditional probability describes **how probabilities change when new information becomes available**.

---

# Intuition

Imagine looking for a friend in a city.

At first, they could be **anywhere**.

Now someone tells you:

> "They are inside the university."

Suddenly, you no longer search the entire city.

Your search space becomes much smaller.

Conditional probability works the same way.

Instead of considering **every possible outcome**, we only consider outcomes that satisfy the information we already know.

---

# What Is Conditional Probability?

Conditional probability is

> **The probability of an event occurring, given that another event has already occurred.**

It is written as

$$
P(A\mid B)
$$

and is read as

> **"The probability of A given B."**

where

- **A** = the event we want to know about.
- **B** = the information we already know.

In other words,

> **What is the probability of A after learning that B is true?**

---

# Mathematical Definition

Conditional probability is defined as

$$
P(A\mid B)=\frac{P(A\cap B)}{P(B)}
$$

where

- \(P(A\cap B)\) is the probability that **both A and B occur**.
- \(P(B)\) is the probability that **B occurs**.

Notice that we divide by \(P(B)\) because we are now working **only inside event B**.

You do **not** need to memorize this formula immediately.

The important idea is that **learning new information changes the sample space**, so we calculate probabilities within this smaller set.

---

# Example 1 – Drawing Cards

Suppose you draw one card from a standard deck of **52 cards**.

What is the probability of drawing a King?

There are four Kings.

$$
P(\text{King})=\frac{4}{52}
$$

Now suppose someone tells you

> "The card is definitely a face card."

Face cards are

- Jack
- Queen
- King

There are **12** face cards.

Now ask again:

> What is the probability that the card is a King?

Instead of considering all 52 cards, we now consider only the 12 face cards.

$$
P(\text{King}\mid\text{Face Card})=\frac{4}{12}=\frac13
$$

Knowing the card is a face card changed the probability.

---

# Example 2 – Weather

Suppose

$$
P(\text{Rain})=0.30
$$

You wake up and notice very dark clouds.

Your estimate may become

$$
P(\text{Rain}\mid\text{Dark Clouds})=0.80
$$

Nothing about nature changed instantly.

Only your **information** changed.

---

# Example 3 – Students

A class has **100 students**.

- 60 are male.
- 40 are female.
- 30 males play football.
- 10 females play football.

The probability that a randomly selected student plays football is

$$
P(\text{Football})=\frac{40}{100}=0.40
$$

Now suppose you know the student is male.

Your sample space becomes only the **60 male students**.

Among them,

30 play football.

Therefore,

$$
P(\text{Football}\mid\text{Male})=\frac{30}{60}=0.50
$$

Again, receiving new information changes the probability.

---

# Machine Learning Example 1 – Email Spam Detection

Suppose the probability that an email is spam is

20%.

Now the model observes words like

- "Congratulations!"
- "Free money!"
- "Winner!"

The probability of spam increases.

The model is estimating

$$
P(\text{Spam}\mid\text{Observed Words})
$$

rather than simply

$$
P(\text{Spam})
$$

---

# Machine Learning Example 2 – Remote Sensing

Suppose a land-cover classifier predicts one of the following classes:

- Forest
- Water
- Urban
- Agriculture

Initially,

$$
P(\text{Water})=0.10
$$

Now the model observes

- Very low NDVI
- High blue-band reflectance
- Pixel located near a river

The probability becomes

$$
P(\text{Water}\mid\text{Observed Features})=0.85
$$

This is exactly how many remote sensing classifiers work.

---

# Machine Learning Example 3 – Medical Diagnosis

Suppose

$$
P(\text{Disease})=0.02
$$

A patient's medical test returns positive.

The probability may increase dramatically.

The model estimates

$$
P(\text{Disease}\mid\text{Positive Test})
$$

rather than

$$
P(\text{Disease})
$$

This principle forms the basis of Bayesian medical diagnosis.

---

# Why Is Conditional Probability Important in Machine Learning?

Most supervised machine learning algorithms make predictions **after observing input features**.

For example,

- **Logistic Regression** estimates the probability of a class given the input features.
- **Naive Bayes** directly uses conditional probabilities for classification.
- **Image classifiers** estimate the probability that an image belongs to each class given its pixels.
- **Remote sensing models** estimate the probability that a satellite pixel belongs to a land-cover class given its spectral bands, NDVI, elevation, texture, and other measurements.

In general, many supervised learning models answer the following question:

> **What is the probability of this class given the observed data?**

---

# Things to Remember

- Conditional probability updates probabilities after receiving new information.
- It is written as \(P(A\mid B)\).
- Learning new information changes the sample space.
- Most supervised machine learning models estimate conditional probabilities.
- Conditional probability is the foundation of Naive Bayes, Bayesian inference, probabilistic classifiers, and many AI algorithms.

---

# Summary

Conditional probability is the probability of an event occurring **given that another event has already occurred**.

Unlike ordinary probability, conditional probability incorporates new information, allowing us to update our beliefs as additional evidence becomes available.

In machine learning, models rarely estimate probabilities in isolation. Instead, they estimate probabilities **conditioned on observed features**.

This idea lies at the heart of spam detection, medical diagnosis, image classification, remote sensing, and countless other machine learning applications.

# Final Example

A land-cover classification model initially estimates

$$
P(\text{Water})=0.10
$$

After observing

- Low NDVI
- High blue reflectance
- Location near a river

the model updates its estimate to

$$
P(\text{Water}\mid\text{Observed Features})=0.85

The probability changes because new evidence has reduced the set of plausible outcomes.

This is the essence of **conditional probability**.
