# Why Should You Care?

Probability measures **how likely an event is to happen**.

For example:

- What is the probability that it will rain tomorrow?
- What is the probability that a satellite pixel represents water?
- What is the probability that an email is spam?
- What is the probability that a patient has a disease?

Machine learning models often output **probabilities** rather than simple **Yes/No** answers.

For example, an image classifier might predict:

| Class | Probability |
|-------|------------:|
| Forest | 0.85 |
| Agriculture | 0.10 |
| Water | 0.04 |
| Urban | 0.01 |

The model predicts **Forest** because it has the highest probability.

Understanding the basic probability rules allows us to determine whether these predictions are mathematically valid and correctly interpret what they mean.

---

# Rule 1: Probability Is Always Between 0 and 1

The most fundamental rule of probability is

$$
0 \le P(A) \le 1
$$

where:

- **0** means the event is impossible.
- **1** means the event is certain.
- Any value between **0** and **1** represents some degree of uncertainty.

---

## Examples

| Event | Probability |
|-------|------------:|
| The sun rises tomorrow | 1 |
| Rolling a 7 on a six-sided die | 0 |
| Rain tomorrow | 0.40 |

---

## Machine Learning Example

Suppose a classifier predicts

| Class | Probability |
|-------|------------:|
| Cat | 1.20 |
| Dog | -0.30 |
| Rabbit | 0.10 |

These predictions are impossible because probabilities cannot be negative or greater than one.

A valid prediction would be

| Class | Probability |
|-------|------------:|
| Cat | 0.70 |
| Dog | 0.20 |
| Rabbit | 0.10 |

---

# Rule 2: The Probability of the Sample Space Is 1

The probability that **one of the possible outcomes occurs** is always

$$
P(S)=1
$$

Why?

Because every experiment must produce one outcome from the sample space.

---

## Example

Roll a die.

Sample space

$$
S=\{1,2,3,4,5,6\}
$$

Since one of these six numbers must appear,

$$
P(S)=1
$$

---

## Machine Learning Example

Suppose a land-cover classifier predicts one of the following classes:

- Forest
- Water
- Urban
- Agriculture

Then

$$
P(\text{Forest})+
P(\text{Water})+
P(\text{Urban})+
P(\text{Agriculture})=1
$$

Example:

| Class | Probability |
|-------|------------:|
| Forest | 0.50 |
| Water | 0.20 |
| Urban | 0.10 |
| Agriculture | 0.20 |

Total:

$$
1.00
$$

---

# Rule 3: Complement Rule

Sometimes it is easier to calculate the probability that an event **does not happen**.

The complement rule states

$$
P(A^c)=1-P(A)
$$

where

- \(A\) = the event
- \(A^c\) = the complement ("not A")

---

## Example

Suppose

$$
P(\text{Rain})=0.30
$$

Then

$$
P(\text{No Rain})=1-0.30=0.70
$$

---

## Machine Learning Example

Suppose a medical AI predicts

| Outcome | Probability |
|----------|------------:|
| Disease | 0.92 |

Then

$$
P(\text{No Disease})=1-0.92=0.08
$$

The complement rule is used constantly in binary classification problems.

---

# Rule 4: Addition Rule for Mutually Exclusive Events

If two events **cannot happen at the same time**, then

$$
P(A \cup B)=P(A)+P(B)
$$

---

## Example

Roll a die.

Event

$$
A=\{1\}
$$

Event

$$
B=\{6\}
$$

A die cannot land on both **1** and **6** simultaneously.

Therefore,

$$
P(A\cup B)=\frac16+\frac16=\frac26=\frac13
$$

---

## Machine Learning Example

Suppose a classifier predicts

| Class | Probability |
|-------|------------:|
| Forest | 0.60 |
| Water | 0.15 |

If each pixel belongs to only one class,

$$
P(\text{Forest or Water})=0.60+0.15=0.75
$$

---

# Rule 5: Total Probability Equals 1

The probabilities of **all possible outcomes** must always sum to one.

---

## Example

| Weather | Probability |
|---------|------------:|
| Sunny | 0.50 |
| Cloudy | 0.30 |
| Rainy | 0.20 |

Total:

$$
0.50+0.30+0.20=1
$$

---

## Machine Learning Example

| Class | Probability |
|-------|------------:|
| Forest | 0.62 |
| Water | 0.15 |
| Urban | 0.13 |
| Agriculture | 0.10 |

Adding every probability gives

$$
1.00
$$

If they do **not** sum to one, then something is wrong with the probability output.

# Why Are These Rules Important in Machine Learning?

Many machine learning algorithms predict probabilities rather than labels.

For example:

- **Logistic Regression** estimates the probability that an email is spam.
- **Naive Bayes** uses probability theory to classify observations.
- **Neural Networks** often use the **Softmax** function so that the probabilities of all classes sum to one.
- In **remote sensing**, land-cover classifiers estimate the probability that each satellite pixel belongs to classes such as Forest, Water, or Urban.

Understanding these rules helps you interpret model predictions correctly instead of treating them as arbitrary numbers.

---

# Things to Remember

- Every probability lies between **0** and **1**.
- The probability of the entire sample space is **1**.
- An event and its complement always add up to **1**.
- If events are mutually exclusive, their probabilities can be added.
- The probabilities of all possible outcomes always sum to **1**.

---

# Summary

Probability measures how likely an event is to occur and always lies between **0** (impossible) and **1** (certain).

The probability of the entire sample space is always **1**, because one outcome must occur. The complement rule states that the probability of an event not occurring equals **1 − P(A)**. For mutually exclusive events, the probability of either event occurring is the sum of their individual probabilities. Finally, the probabilities of all possible outcomes must always add up to **1**.

These rules form the mathematical foundation of probabilistic machine learning and are essential for interpreting predictions from models such as Logistic Regression, Naive Bayes, Decision Trees, and Neural Networks.

A satellite image classifier predicts:

| Class | Probability |

| Forest | 0.65 |
