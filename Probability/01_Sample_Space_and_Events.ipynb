# Sample Space and Events

---

## Why Should You Care?

Machine learning is built on **data** and **uncertainty**.

For example:

- Will it rain tomorrow?
- Is this satellite pixel **forest** or **agriculture**?
- Will this customer buy a product?
- Is this email spam?

We do not know the answer with certainty, so we use **probability**.

Before studying probability, we need two fundamental concepts:

- **Sample Space** → All possible outcomes.
- **Event** → One or more outcomes we care about.

You can think of them as:

- **Sample Space** = the entire universe of possible outcomes.
- **Event** = the particular outcomes we are interested in.

---

# 1. Sample Space (S or Ω)

A **sample space** is:

> **The set of every possible outcome of an experiment.**

An experiment can be anything that produces an outcome.

---

## Example 1 – Tossing a Coin

Possible outcomes:

- Heads
- Tails

Sample space:

$$
S=\{H,T\}
$$

---

## Example 2 – Rolling One Die

Possible outcomes:

- 1
- 2
- 3
- 4
- 5
- 6

Sample space:

$$
S=\{1,2,3,4,5,6\}
$$

---

## Example 3 – Satellite Image Classification

Suppose a model classifies every pixel as:

- Forest
- Water
- Urban
- Agriculture

The sample space is

$$
S=\{\text{Forest},\text{Water},\text{Urban},\text{Agriculture}\}
$$

Every prediction must belong to this set.

---

## Example 4 – Temperature Prediction

Suppose tomorrow's temperature is classified into:

- Cold
- Mild
- Hot

Sample space:

$$
S=\{\text{Cold},\text{Mild},\text{Hot}\}
$$

---

## Machine Learning View

Imagine you build an image classifier.

Possible predictions:

- Cat
- Dog
- Rabbit

The sample space is

$$
S=\{\text{Cat},\text{Dog},\text{Rabbit}\}
$$

Since **Elephant** is not part of the sample space, the model cannot predict it.

---

# 2. Event

An **event** is:

> **A subset of the sample space.**

In other words, we usually care about only some of the possible outcomes.

---

## Example 1 – Even Numbers

Roll a die.

Sample space

$$
S=\{1,2,3,4,5,6\}
$$

Question:

> What is the probability of rolling an even number?

Event:

$$
A=\{2,4,6\}
$$

Notice that the event is part of the sample space.

---

## Example 2 – Numbers Greater Than 4

Roll a die.

Question:

> What is the probability of rolling a number greater than 4?

Event:

$$
A=\{5,6\}
$$

---

## Example 3 – Coin Toss

Sample space:

$$
S=\{H,T\}
$$

Event:

Getting Heads

$$
A=\{H\}
$$

---

## Example 4 – Remote Sensing

Suppose land-cover classes are

- Forest
- Urban
- Water
- Agriculture

Sample space:

$$
S=\{\text{Forest},\text{Urban},\text{Water},\text{Agriculture}\}
$$

Suppose we only care whether a pixel represents **vegetation**.

Then the event is

$$
A=\{\text{Forest},\text{Agriculture}\}
$$

Two different outcomes are grouped into a single event.

This happens frequently in environmental monitoring.

---

## Example 5 – Disease Classification

Suppose an AI predicts:

- Healthy
- Flu
- COVID
- Pneumonia

Sample space:

$$
S=\{\text{Healthy},\text{Flu},\text{COVID},\text{Pneumonia}\}
$$

Question:

> What is the probability that the patient has an infectious disease?

Event:

$$
A=\{\text{Flu},\text{COVID}\}
$$

---

# Visualizing It

Imagine the sample space as a large box containing every possible outcome.

```text
+--------------------------------+
| Forest                         |
| Water                          |
| Urban                          |
| Agriculture                    |
+--------------------------------+
```

Now highlight only:

- Forest
- Agriculture

The highlighted outcomes form an **event**.

---

# Why Is This Important in Machine Learning?

Nearly every machine learning algorithm estimates the probability of **events**.

Examples include:

- Probability this email is spam.
- Probability this image contains a car.
- Probability this satellite pixel is water.
- Probability tomorrow's rainfall exceeds 20 mm.
- Probability a customer will churn.

A machine learning model always starts with the **sample space** (all possible predictions) and estimates the probability of the **event** that interests us.

---

# Things to Remember

- Every experiment has exactly one sample space.
- Every event is a subset of the sample space.
- An event may contain one outcome or many outcomes.
- Probabilities are assigned to events rather than individual outcomes.
- Machine learning models often predict the probability of an event (e.g., "this pixel is forest").

---

# Summary

A **sample space** is the complete set of all possible outcomes of an experiment and is usually denoted by **S** or **Ω**.

An **event** is any subset of the sample space that represents the outcomes of interest.

These concepts form the foundation of probability theory and are essential for understanding many machine learning algorithms.

In machine learning, the sample space usually corresponds to all possible classes that a model can predict, while an event represents one or more specific outcomes we want to study.

For example, a land-cover classification model may have the sample space

$$
S=\{\text{Forest},\text{Water},\text{Urban},\text{Agriculture}\}
$$

If we are only interested in vegetation, the event is

$$
A=\{\text{Forest},\text{Agriculture}\}
$$

---

# Final Example

A die is rolled once.

Sample space:

$$
S=\{1,2,3,4,5,6\}
$$

Event:

"The result is an even number."

$$
A=\{2,4,6\}
$$

The sample space contains **all possible outcomes**, while the event contains only the outcomes that satisfy the desired condition.

Understanding this distinction is the first step toward learning probability, statistics, and machine learning.
