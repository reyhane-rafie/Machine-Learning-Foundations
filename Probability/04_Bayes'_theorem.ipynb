# Why Should You Care?

Imagine making a prediction, and then receiving new information.

Should your prediction change?

Absolutely.

For example:

- A medical test comes back **positive**.
- A satellite pixel has a **high NDWI value**.
- An email contains the word **"Congratulations!"**.
- A customer suddenly stops using an app.

Each new piece of evidence should change what we believe.

Bayes' Theorem gives us the mathematical rule for updating those beliefs.

This idea appears throughout Machine Learning.

Examples include:

- Spam detection
- Medical diagnosis
- Remote sensing
- Recommendation systems
- Speech recognition
- Robotics

---

# Intuition

Imagine you're a detective.

At first, you know very little.

Perhaps you believe there is only a **10% chance** that a particular suspect committed the crime.

Then you discover:

- Fingerprints
- Security camera footage
- DNA evidence

Each clue changes your belief.

You don't ignore your original belief, nor do you trust one piece of evidence blindly.

Instead, you **combine what you already believed with the new evidence**.

That is exactly what Bayes' Theorem does.

A simple way to think about it is

> **Old belief + New evidence = Updated belief**

Machine learning models perform this kind of reasoning constantly.

# Bayes' Theorem


::contentReference[oaicite:0]{index=0}


Bayes' Theorem is written as

$$
P(A\mid B)=\frac{P(B\mid A)\,P(A)}{P(B)}
$$

where

- **A** is the event we care about.
- **B** is the observed evidence.

---

# Understanding Each Part

## 1. Prior — \(P(A)\)

The **prior** is your belief **before** observing any new evidence.

Example:

Before opening an email,

only **5%** of all emails are spam.

That is your prior belief.

---

## 2. Likelihood — \(P(B\mid A)\)

The likelihood answers:

> If **A is true**, how likely is it that we would observe **B**?

Example:

If an email is spam,

how likely is it to contain the word

**"Free"**?

Perhaps

90%.

---

## 3. Evidence — \(P(B)\)

Evidence measures

how common the observed information is overall.

Example:

How often does the word

"Free"

appear in **all** emails,

including both spam and non-spam?

---

## 4. Posterior — \(P(A\mid B)\)

The posterior is the quantity we want.

It answers

> After observing the evidence, what should I believe now?

This is the updated probability.

---

# A Simple Example

Suppose

Only **1%** of people have a certain disease.

$$
P(\text{Disease})=0.01
$$

The medical test correctly detects the disease **99%** of the time.

$$
P(\text{Positive}\mid\text{Disease})=0.99
$$

However,

**5%** of healthy people also test positive.

Now someone receives a **positive** test result.

Many people immediately think

> "The test is 99% accurate, so there is a 99% chance the person has the disease."

This is **incorrect**.

The disease is very rare.

Bayes' Theorem combines

- how common the disease is (prior),
- how reliable the test is (likelihood),
- and the observed evidence,

to calculate the true probability.

This is why Bayes' Theorem is so powerful—it prevents us from ignoring how common or rare an event actually is.

---

# Machine Learning Example 1 – Spam Detection

Suppose

only **10%** of emails are spam.

An email contains the words

- Congratulations
- Winner
- Claim Prize

These words are much more common in spam emails.

Instead of estimating

$$
P(\text{Spam})
$$

the classifier estimates

$$
P(\text{Spam}\mid\text{Observed Words})
$$

If the updated probability becomes

95%,

the email is classified as spam.

This is exactly how the **Naive Bayes** classifier works.

---

# Machine Learning Example 2 – Remote Sensing

Suppose a land-cover classifier predicts

- Forest
- Water
- Urban
- Agriculture

Initially,

$$
P(\text{Water})=0.10
$$

Now the model observes

- High NDWI
- Low NDVI
- Pixel located near a river

Bayes' Theorem updates

$$
P(\text{Water})
$$

to

$$
P(\text{Water}\mid\text{Observed Features})
$$

The probability may increase to

92%.

The model updates its belief using new evidence.

# Machine Learning Example 3 – Image Classification

Initially,

an image classifier predicts

| Class | Probability |
|-------|------------:|

The model then detects

- Whiskers

The probabilities change as new evidence is extracted from the image.

This continual updating of beliefs is a Bayesian way of thinking.

# Common Mistakes

### Mistake 1

Confusing

$$
P(A\mid B)
$$

with

$$
P(B\mid A)
$$

These are usually **not** the same.

Example:

- Probability of coughing **given** you have the flu.
- Probability of having the flu **given** you are coughing.

These answer different questions.

---

### Mistake 2

Ignoring the prior probability.

Even a highly accurate medical test can produce many false positives if the disease is extremely rare.

Always consider how common an event is before seeing the evidence.

---

### Mistake 3

Thinking Bayes' Theorem only applies to medicine.

It is widely used in

- Machine Learning
- Robotics
- Finance
- Natural Language Processing
- Remote Sensing
- Artificial Intelligence

---

# Things to Remember

- Bayes' Theorem updates probabilities after observing new evidence.
- It combines a prior belief with observed evidence.
- The updated probability is called the **posterior**.
- Bayes' Theorem is built on conditional probability.
- \(P(A\mid B)\) is generally **not** equal to \(P(B\mid A)\).
- It forms the mathematical foundation of the Naive Bayes classifier.

---

# Summary

Bayes' Theorem is a mathematical rule for updating probabilities when new information becomes available.

Instead of relying only on our initial belief or only on new evidence, Bayes' Theorem combines both to produce a more informed estimate.

In machine learning, many models estimate the probability of a class **given the observed features**. This makes Bayes' Theorem one of the most important ideas in probabilistic machine learning.

# Final Example

Suppose a land-cover classifier initially estimates

| Class | Prior Probability |
|-------|------------------:|
| Forest | 0.50 |
| Water | 0.10 |
| Urban | 0.20 |
| Agriculture | 0.20 |

After observing

- High NDWI
- Low NDVI
- Location near a river

the probability of **Water** increases to

$$
P(\text{Water}\mid\text{Observed Features})=0.92
$$

The model predicts **Water** because the observed evidence strongly supports that class.

This illustrates the central idea of Bayes' Theorem:

> **New evidence should update our beliefs.**
