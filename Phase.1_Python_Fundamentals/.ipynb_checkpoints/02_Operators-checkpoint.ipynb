# Why Should You Care?

Variables store information, but **operators allow Python to do something with that information**.

Operators enable Python to:

- Perform calculations.
- Compare values.
- Make decisions.

In **Machine Learning**, **GIS**, and **Remote Sensing**, operators are used constantly.

Examples:

- Calculate NDVI.
- Compare environmental measurements.
- Filter satellite pixels.
- Combine multiple conditions before making predictions.

Without operators, Python could only store data—it could not process it.

---

# Core Concept

An **operator** is a symbol that tells Python to perform an operation on one or more values.

Python operators can be grouped into three main categories:

1. Arithmetic Operators
2. Comparison Operators
3. Logical Operators

---

# 1. Arithmetic Operators

Arithmetic operators perform mathematical calculations.

| Operator | Meaning | Example | Result |
|----------|---------|---------|-------:|
| `+` | Addition | `3 + 2` | `5` |
| `-` | Subtraction | `7 - 4` | `3` |
| `*` | Multiplication | `5 * 2` | `10` |
| `/` | Division | `8 / 2` | `4.0` |
| `//` | Floor Division | `7 // 2` | `3` |
| `%` | Modulo (Remainder) | `7 % 2` | `1` |
| `**` | Power (Exponentiation) | `2 ** 3` | `8` |

---

## Examples

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

Output:

```text
13
7
30
3.3333333333333335
3
1
1000
```

---

# 2. Comparison Operators

Comparison operators compare two values.

The result is **always** either:

```python
True
```

or

```python
False
```

| Operator | Meaning |
|----------|---------|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

---

## Examples

```python
temperature = 28

print(temperature > 20)

print(temperature == 30)

print(temperature != 30)

print(temperature <= 28)
```

Output:

```text
True
False
True
True
```

Comparison operators are commonly used to ask questions such as:

- Is the temperature above 30°C?
- Is NDVI greater than 0.5?
- Is rainfall below the threshold?

---

# 3. Logical Operators

Logical operators combine multiple conditions.

| Operator | Meaning |
|----------|---------|
| `and` | Both conditions must be `True` |
| `or` | At least one condition must be `True` |
| `not` | Reverses `True` and `False` |

---

## Examples

```python
temperature = 28

is_raining = False

print(temperature > 20 and is_raining == False)

print(temperature > 30 or is_raining == False)

print(not is_raining)
```

Output:

```text
True
True
True
```

---

# Remote Sensing Example

Suppose a satellite pixel has the following properties:

```python
ndvi = 0.72

cloud = False

temperature = 31
```

### Arithmetic

```python
temperature - 5
```

Result:

```text
26
```

---

### Comparison

```python
ndvi > 0.5
```

Result:

```text
True
```

---

### Logical

```python
ndvi > 0.5 and cloud == False
```

Meaning:

- Vegetation is healthy.
- The pixel is cloud-free.

Only pixels satisfying both conditions may be used for further analysis or machine learning.

This type of filtering is very common in **Google Earth Engine** and **Python** workflows.

Output:

```text
False
```

---

```python
print(rainfall < 20)
```

```text
True
```

---

```python
print(temperature > 20 and rainfall < 20)
```

Both are true.

Output:

```text
True
```

- Operators perform actions on variables and values.

# Common Beginner Mistakes

## Mistake 1: Confusing `=` and `==`

❌ Incorrect:

```python
temperature == 25
```

when trying to assign a value.

✅ Correct:

```python
temperature = 25
```

Remember:

- `=` assigns a value.
- `==` compares two values.

---

## Mistake 2: Forgetting That Comparison Operators Return Booleans

```python
temperature > 30
```

does **not** return a number.

It returns:

```python
True
```

or

```python
False
```

---

## Mistake 3: Using `and` When You Mean `or`

Example:

```python
temperature > 30 and rainfall < 10
```

Both conditions must be true.

If you only need one condition to be true, use:

```python
temperature > 30 or rainfall < 10
```

---

# Summary

Operators allow Python to work with data instead of simply storing it.

- **Arithmetic operators** perform mathematical calculations.
- **Comparison operators** compare values and return Boolean results.
- **Logical operators** combine multiple conditions.

These operators form the basis of calculations, decision-making, and data filtering in Python, making them essential tools for AI, Machine Learning, GIS, and Remote Sensing.
