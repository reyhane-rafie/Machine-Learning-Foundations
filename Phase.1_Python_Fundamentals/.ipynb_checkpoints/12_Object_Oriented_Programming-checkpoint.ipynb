---

# 2. Core Concept (Short)

Imagine you're designing cars.

Before making a real car, you create a **blueprint**.

That blueprint describes:

- Color
- Engine
- Wheels
- Doors

The blueprint itself isn't a real car.

When you build a car from the blueprint, you get a real car.

In Python:

- **Class** → The blueprint.
- **Object** → A real thing created from the blueprint.
- **Method** → An action the object can perform.

Think of it like this:

```
Class
   ↓
Create
   ↓
Object
   ↓
Use its methods
```

---

## Class

A **class** defines what an object should have.

Example:

```python
class Dog:
    pass
```

This creates a blueprint called `Dog`.

---

## Object

An **object** is created from a class.

```python
dog1 = Dog()
```

Now `dog1` is a real object.

You can create many objects from one class.

```python
dog1 = Dog()
dog2 = Dog()
dog3 = Dog()
```

One blueprint.

Many objects.

---

## Method

A **method** is a function that belongs to an object.

Example:

```python
name = "Rey"

print(name.upper())
```

Here:

- `"Rey"` is an object (a string object).
- `upper()` is a method.

It performs an action on that object.

Result:

```
REY
```

You've actually been using methods since the **Strings** lesson!

---

# 3. Simple Example

Imagine a cookie cutter.

The cookie cutter is the **class**.

Every cookie made from it is an **object**.

The cookie can be:

- Eaten
- Decorated
- Broken

These are like **methods**.

```
Cookie Cutter
      ↓
Class

Cookies
      ↓
Objects

Decorate()
Eat()
      ↓
Methods
```

# 4. Remote Sensing / Environmental Example

Imagine a satellite image.

Every satellite image has:

- Width
- Height
- Bands

And it can perform actions like:

- Clip
- Display
- Calculate NDVI

Conceptually:

```
Class
SatelliteImage

        ↓

Object
Sentinel Image from July 2025

        ↓

Methods
clip()
select()
show()
```

Later in **Google Earth Engine**, you'll write code like:

```python
image.clip(studyArea)
```

or

```python
image.select("B4")
```

Here:

- `image` is an object.
- `clip()` and `select()` are methods.

Now you know what those terms mean!

# 5. Important Points

- A class is a blueprint.
- An object is a real instance created from a class.
- A method is a function that belongs to an object.
- You have already used methods like `.upper()` and `.strip()`.
- Most AI and Remote Sensing libraries are built using classes and objects.

# 7. Mini Coding Example

Let's create a very simple class.

```python
class Satellite:

    def info(self):
        print("This is a satellite.")
```

Create an object:

```python
image = Satellite()
```

Call its method:

```python
image.info()
```

Output:

```
This is a satellite.
```

---

# Line by Line Explanation

## Creating the class

```python
class Satellite:
```

Creates a blueprint.

---

## Creating a method

```python
def info(self):
```

Creates a method.

For now, don't worry about `self`.

Just remember:

It is required when defining methods inside a class.

We'll study it properly later.

---

## Creating an object

```python
image = Satellite()
```

Creates an object.

---

## Calling the method

```python
image.info()
```

Calls the object's method.
