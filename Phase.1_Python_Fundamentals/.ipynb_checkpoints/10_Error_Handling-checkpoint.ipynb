# Lesson 10 — Error Handling (try, except)

## 1. Why Should You Care?

Real-world programs rarely run perfectly.

Users enter wrong information, files may be missing, and datasets can contain unexpected values.

In **AI, Machine Learning, GIS, and Remote Sensing**, you will work with large and messy datasets.

**Error handling** helps your programs continue running and respond properly instead of crashing.

Examples:

- A satellite file does not exist.
- A CSV file has missing values.
- A user enters text instead of a number.

---

# 2. Core Concept (Short)

Normally, when Python finds an error, it stops the program.

Example:

```python
temperature = float(input("Enter temperature: "))
```

If the user enters:

```
abc
```

Python cannot convert `"abc"` into a float.

It creates an error:

```
ValueError
```

---

## Error Handling

Error handling allows us to try something and handle the problem if it fails.

The basic structure:

```python
try:
    # Code that might cause an error

except:
    # Code that runs if an error happens
```

Think of it like:

```
Try → Something might go wrong
        ↓
Error happens?
        ↓
Except → Handle the problem
```

---

# 3. Simple Example

## Without Error Handling

```python
age = int(input("Enter your age: "))

print(age)
```

If the user enters:

```
twenty
```

Python crashes.

---

## With Error Handling

```python
try:
    age = int(input("Enter your age: "))
    print(age)

except:
    print("Please enter a number.")
```

Now:

Input:

```
twenty
```

Output:

```
Please enter a number.
```

The program does not crash.

---

# 4. Remote Sensing / Environmental Example

Imagine you are reading a satellite measurement.

```python
ndvi = float(input("Enter NDVI value: "))
```

A valid input:

```
0.72
```

works.

But someone enters:

```
unknown
```

Python cannot convert it.

---

## Using Error Handling

```python
try:
    ndvi = float(input("Enter NDVI value: "))
    print("NDVI:", ndvi)

except:
    print("Invalid NDVI value.")
```

Now your program can handle incorrect data.

---

## Another Example: Opening a File

Without error handling:

```python
file = open("sentinel_data.csv", "r")
```

If the file does not exist:

```
FileNotFoundError
```

---

With error handling:

```python
try:
    file = open("sentinel_data.csv", "r")
    data = file.read()

except:
    print("File not found.")
```

---

# 5. Important Points

- Errors are normal in real programs.
- `try` contains code that might fail.
- `except` handles the error.
- Error handling prevents programs from stopping unexpectedly.
- It is especially important when working with user input and real-world datasets.
