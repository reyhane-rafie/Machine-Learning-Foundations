# Core Concept

A **function** is a reusable block of code that performs a specific task.

Instead of writing the same instructions repeatedly, you define them once and call them whenever needed.

---

# Creating a Function

Basic syntax:

```python
def function_name():
    print("Hello!")
```

Calling the function:

```python
function_name()
```

Output:

```text
Hello!
```

# Parameters

Sometimes a function needs information to perform its task.

That information is called a **parameter**.

Example:

```python
def greet(name):
    print("Hello", name)
```

Calling the function:

```python
greet("Rey")
```

Output:

```text
Hello Rey
```

---

## Parameter vs Argument

Example:

```python
def greet(name):
    print("Hello", name)


greet("Rey")
```

Here:

Parameter:

```python
name
```

The variable inside the function definition.

Argument:

```python
"Rey"
```

The actual value passed to the function.

Easy way to remember:

```
Parameter = placeholder variable

Argument = actual value
```

---

# Return

Some functions do not only display information.

They produce a value that can be used later.

This is done with:

```python
return
```

Example:

```python
def add(a, b):
    return a + b
```

Using the function:

```python
result = add(3, 5)

print(result)
```

Output:

```text
8
```

---

# `print()` vs `return`

This is one of the most important concepts in Python.

## Using print()

```python
def add(a, b):
    print(a + b)
```

This only displays the result.

The value is not returned to the program.

---

## Using return()

```python
def add(a, b):
    return a + b
```

This sends the value back.

Now you can store it:

```python
x = add(3, 5)
```

and use it later.

---

# Scope

**Scope** means where a variable exists and can be accessed.

A variable created inside a function only exists inside that function.

Example:

```python
def greet():

    message = "Hello"

print(message)
```

This produces an error.

Why?

Because:

```python
message
```

only exists inside:

```python
greet()
```

This type of variable is called a:

```
Local variable
```

---

# Global Variables

A variable created outside a function is called a global variable.

Example:

```python
name = "Rey"


def greet():
    print(name)
```

Here:

```python
name
```

can be accessed inside the function.

---

## Simple Rule

Remember:

> Variables created inside a function stay inside that function.

# Remote Sensing / Environmental Example

Suppose you frequently check whether vegetation is healthy.

Without a function:

```python
if ndvi > 0.5:
    print("Healthy")
```

You would have to repeat this logic many times.

Instead, create a function:

```python
def check_ndvi(ndvi):

    if ndvi > 0.5:
        return "Healthy vegetation"

    else:
        return "Sparse vegetation"
```

Now you can reuse it:

```python
print(check_ndvi(0.72))

print(check_ndvi(0.31))

print(check_ndvi(0.60))
```

Output:

```text
Healthy vegetation
Sparse vegetation
Healthy vegetation
```

The code is:

- Cleaner.
- Easier to modify.
- Easier to reuse.

# Mini Coding Example

```python
def area(length, width):

    return length * width


room_area = area(5, 4)


print(room_area)
```

---

# Step-by-Step Explanation

## Step 1

```python
def area(length, width):
```

Creates a function.

It has two parameters:

```
length
width
```

---

## Step 2

```python
return length * width
```

Calculates the area.

The result is returned.

---

## Step 3

```python
room_area = area(5, 4)
```

Calls the function.

Python assigns:

```
length = 5

width = 4
```

The function calculates:

```
5 × 4 = 20
```

and returns:

```
20
```

This value is stored in:

```python
room_area
```

---

## Step 4

```python
print(room_area)
```

Output:

```text
20
```

# Common Beginner Mistakes

## Mistake 1: Confusing print and return

Incorrect understanding:

> "If I print a value, I can use it later."

Not true.

`print()` only displays information.

`return` sends the value back.

---

## Mistake 2: Forgetting to call the function

Creating:

```python
def greet():
    print("Hello")
```

does not run the function.

You must call:

```python
greet()
```

---

## Mistake 3: Confusing Parameters and Arguments

Example:

```python
def calculate(value):
```

`value` is the parameter.

Example:

```python
calculate(100)
```

`100` is the argument.
