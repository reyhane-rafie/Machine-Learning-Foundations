# Why Should You Care?

Variables are one of the most fundamental concepts in programming.

Almost everything you do in **Artificial Intelligence (AI)**, **Machine Learning (ML)**, **GIS**, and **Remote Sensing (RS)** involves storing and manipulating data using variables.

For example:

- A satellite image stores pixel values as numbers.
- A machine learning model stores features, labels, and predictions.
- Environmental datasets store values such as temperature, rainfall, and NDVI.

Before building AI models or analyzing satellite imagery, you need to understand how Python stores information.

---

# Core Concept

A **variable** is a name that stores a piece of information.

Think of it as a labeled box.

```
Variable Name  →  Stored Value
```

Example:

```python
temperature = 25
```

Here:

- `temperature` → Variable name
- `25` → Stored value

Python automatically determines the type of data stored inside the variable.

---

# The Four Basic Data Types

Python provides many data types, but the four most important for beginners are:

1. Integer (`int`)
2. Float (`float`)
3. String (`str`)
4. Boolean (`bool`)

---

# 1. Integers (`int`)

Integers are **whole numbers** without decimal points.

Examples:

```python
year = 2026

number_of_satellites = 5

pixel_value = 120
```

Typical uses:

- Counting objects.
- Years.
- Image pixel values.
- Number of observations.

---

# 2. Floats (`float`)

Floats are **numbers with decimal values**.

Examples:

```python
temperature = 23.5

rainfall = 12.7

ndvi = 0.65
```

Typical uses:

- Temperature.
- Rainfall.
- NDVI.
- Coordinates.
- Continuous measurements.

Remote Sensing data frequently uses floats because environmental measurements are rarely whole numbers.

---

# 3. Strings (`str`)

Strings store **text**.

A string must be enclosed in quotation marks.

Examples:

```python
name = "Sentinel-2"

location = "Iran"
```

Strings can represent:

- Names.
- Labels.
- Categories.
- File paths.
- Descriptions.

---

# 4. Booleans (`bool`)

Booleans represent **True** or **False** values.

Examples:

```python
cloud_present = True

water_detected = False
```

Booleans are commonly used for decision-making.

Examples:

- Is this pixel water?
- Is the image cloudy?
- Did the model classify the sample correctly?

---

# Simple Example

Imagine you are creating a student profile.

```python
name = "Rey"

age = 25

height = 165.5

is_student = True
```

The variables contain different types of information.

| Variable | Value | Data Type |
|----------|-------|-----------|
| `name` | `"Rey"` | `str` |
| `age` | `25` | `int` |
| `height` | `165.5` | `float` |
| `is_student` | `True` | `bool` |

Python stores each data type differently.

---

# Remote Sensing Example

Suppose we analyze a satellite pixel.

```python
latitude = 32.5

longitude = 51.6

ndvi_value = 0.72

land_cover = "vegetation"

cloud = False
```

Meaning:

| Variable | Description | Data Type |
|----------|-------------|-----------|
| `latitude` | Geographic coordinate | `float` |
| `longitude` | Geographic coordinate | `float` |
| `ndvi_value` | Vegetation index | `float` |
| `land_cover` | Land cover category | `str` |
| `cloud` | Cloud mask | `bool` |

A machine learning model could use variables such as:

```
NDVI + Rainfall + Temperature
            ↓
Predict Vegetation Health
```

---

# Naming Variables

Choose names that clearly describe the stored information.

Good examples:

```python
temperature

ndvi_value

rainfall

city_name
```

Poor examples:

```python
x

abc

data1
```

Meaningful names make your code easier to read and maintain.

---

# Key Takeaways

- A variable stores information.
- Python automatically determines a variable's data type.
- The four basic data types are:
  - `int` → Whole numbers.
  - `float` → Decimal numbers.
  - `str` → Text.
  - `bool` → `True` or `False`.
- Remote Sensing and Machine Learning use all four data types.
- Meaningful variable names improve code readability.

# Common Beginner Mistakes

## Mistake 1: Forgetting Quotation Marks Around Text

❌ Incorrect:

```python
city = Isfahan
```

✅ Correct:

```python
city = "Isfahan"
```

---

## Mistake 2: Confusing Integers and Floats

```python
25
```

is an integer.

```python
25.0
```

is a float.

Even though they represent the same numerical value, they are different data types.

---

## Mistake 3: Writing Boolean Values as Strings

❌ Incorrect:

```python
cloud = "True"
```

This is a string.

✅ Correct:

```python
cloud = True
```

This is a Boolean value.
