# Lesson 11 — Modules (`import`, `pip`)

## 1. Why Should You Care?

Imagine you want to calculate complex mathematical functions, analyze satellite images, or build a machine learning model.

You could write thousands of lines of code yourself—but you don't have to.

Python has **modules** (also called **libraries** or **packages**) that contain code written by other programmers. By importing them, you can use powerful tools with just a few lines of code.

In **AI, Remote Sensing, GIS, and Google Earth Engine**, you'll use libraries like **NumPy, Pandas, Matplotlib, and Scikit-learn** almost every day.

---

# 2. Core Concept (Short)

A **module** is a file that contains Python code (functions, classes, or variables) that you can reuse.

Instead of writing everything yourself, you **import** a module.

Example:

```python
import math

print(math.sqrt(16))
```

Output:

```
4.0
```

Here:

- `math` is the module.
- `sqrt()` is a function inside that module.

---

## `import`

The basic syntax is:

```python
import module_name
```

Example:

```python
import random

print(random.randint(1, 10))
```

This prints a random integer between **1 and 10**.

---

You can also import only one function:

```python
from math import sqrt

print(sqrt(25))
```

Output:

```
5.0
```

Now you don't need to write `math.sqrt()`.

---

## `pip`

Some modules are built into Python, such as:

- `math`
- `random`
- `csv`

Others must be installed first.

Python uses **pip** (**Python Package Installer**) to install them.

Example:

```bash
pip install numpy
```

After installation:

```python
import numpy
```

Now you can use **NumPy** in your programs.

Think of **pip** like an app store:

```
pip
   ↓
Downloads a library
   ↓
Installs it
   ↓
Python can import it
```

Or you could buy ready-made bricks.

```python
import math

print(math.sqrt(49))
```

# 4. Remote Sensing / Environmental Example

Suppose you have thousands of NDVI values.

Without libraries, calculations would be slow and difficult.

With **NumPy**:

```python
import numpy as np

ndvi = np.array([0.32, 0.58, 0.71])

print(ndvi.mean())
```

NumPy quickly calculates the average NDVI.

---

Or imagine reading a CSV file:

```python
import pandas as pd

data = pd.read_csv("ndvi.csv")
```

With one line, **Pandas** reads an entire dataset.

These libraries save hours of work.

---

# 5. Important Points

- A module is reusable Python code.
- `import` allows your program to use a module.
- Built-in modules come with Python.
- `pip` installs additional modules.
- Libraries like **NumPy** and **Pandas** are essential for AI and Remote Sensing.

# 🎓 Mentor Tip

One point that often confuses beginners is the difference between **module**, **library**, and **package**.

At your current stage, it's perfectly fine to think of them all as **collections of reusable Python code**.

Later, when you have more experience, we'll look at the technical differences.

For now, focus on understanding:

- **`pip`** installs a library.
- **`import`** lets your Python program use that library.
