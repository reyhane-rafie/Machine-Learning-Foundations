# Why Should You Care?

Computers are extremely good at performing repetitive tasks.

Instead of writing the same code many times, loops allow you to tell Python:

> "Repeat this task."

Loops make programs:

- Shorter.
- Easier to maintain.
- More powerful.

In **AI**, **Machine Learning**, **GIS**, and **Remote Sensing**, loops are used for tasks such as:

- Processing thousands of satellite images.
- Analyzing pixels in an image.
- Reading rows from datasets.
- Training models through many iterations (epochs).

Without loops, many real-world data processing tasks would be almost impossible.

# Core Concept

A **loop** repeats a block of code multiple times.

Python has two main types of loops:

1. `for` loop
2. `while` loop

Python also provides:

- `break`
- `continue`

to control how loops behave.

---

# 1. `for` Loop

A `for` loop is used when you know how many times you want to repeat something.

Example:

```python
for i in range(5):
    print(i)
```

Output:

```text
0
1
2
3
4
```

---

## Understanding `range()`

The function:

```python
range(5)
```

generates:

```text
0, 1, 2, 3, 4
```

Important:

The ending number is **not included**.

So:

```python
range(5)
```

means:

```
Start: 0
Stop: 5 (not included)
```

---

# 2. `while` Loop

A `while` loop repeats code as long as a condition is `True`.

Example:

```python
count = 1

while count <= 5:
    print(count)
    count = count + 1
```

Output:

```text
1
2
3
4
5
```

---

## Why Update the Variable?

Notice:

```python
count = count + 1
```

This changes the value of `count`.

Without this update:

```python
while count <= 5:
    print(count)
```

the condition would always remain true.

The loop would never stop.

This is called an:

```
Infinite loop
```

---

# 3. `break`

`break` immediately stops a loop.

Example:

```python
for i in range(10):
    if i == 4:
        break

    print(i)
```

Output:

```text
0
1
2
3
```

When:

```python
i == 4
```

Python exits the loop completely.

---

# 4. `continue`

`continue` skips the current iteration and moves to the next one.

Example:

```python
for i in range(5):

    if i == 2:
        continue

    print(i)
```

Output:

```text
0
1
3
4
```

When:

```python
i == 2
```

Python skips the print statement and continues with the next value.

---

# Simple Example

Imagine you need to water five plants.

Without a loop:

```python
print("Water plant")
print("Water plant")
print("Water plant")
print("Water plant")
print("Water plant")
```

The code works, but it is repetitive.

---

Using a loop:

```python
for i in range(5):
    print("Water plant")
```

Same result with much less code.

# Remote Sensing / Environmental Example

Suppose you have NDVI values from five locations.

```python
ndvi_values = [0.32, 0.61, 0.45, 0.71, 0.80]
```

You want to print each value.

Using a loop:

```python
for ndvi in ndvi_values:
    print(ndvi)
```

Output:

```text
0.32
0.61
0.45
0.71
0.80
```

Later, loops will allow you to:

- Process satellite images one by one.
- Analyze environmental measurements.
- Apply calculations to large datasets.
- Train machine learning models through many iterations.

# Mini Coding Example

```python
total = 0

for i in range(1, 6):
    total = total + i

print(total)
```

---

# Step-by-Step Explanation

## Step 1

```python
total = 0
```

Creates a variable to store the running total.

Current value:

```
total = 0
```

---

## Step 2

```python
for i in range(1, 6):
```

`range(1, 6)` produces:

```text
1, 2, 3, 4, 5
```

Important:

The starting value is included.

The ending value is not included.

---

## Step 3

```python
total = total + i
```

Python adds the current value of `i` to `total`.

Step-by-step:

| i | total |
|---|-------|
| 1 | 1 |
| 2 | 3 |
| 3 | 6 |
| 4 | 10 |
| 5 | 15 |

---

## Step 4

```python
print(total)
```

Output:

```text
15
```

This concept of keeping a running total is very common in:

- Data analysis.
- Statistics.
- Machine Learning.

# Key Takeaways

- A loop repeats code.
- Use a `for` loop when you know the number of repetitions.
- Use a `while` loop when repetition depends on a condition.
- `break` stops a loop immediately.
- `continue` skips the current iteration.
- Loops are essential for processing large datasets.

# Common Beginner Mistakes

## Mistake 1: Forgetting That `range()` Stops Before the Ending Number

Example:

```python
range(5)
```

does not produce:

```text
0,1,2,3,4,5
```

It produces:

```text
0,1,2,3,4
```

---

## Mistake 2: Creating an Infinite `while` Loop

Incorrect:

```python
count = 1

while count <= 5:
    print(count)
```

The value of `count` never changes.

Correct:

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

---

## Mistake 3: Confusing `break` and `continue`

`break`:

```
Stop the entire loop.
```

`continue`:

```
Skip only the current iteration.
```
