# Core Concept

A **collection** stores multiple pieces of data inside one variable.

Python has four main collection types:

| Collection | Ordered | Can Change? | Allows Duplicates? |
|---|---|---|---|
| List | Yes | Yes | Yes |
| Tuple | Yes | No | Yes |
| Set | No | Yes | No |
| Dictionary | Yes | Yes | Keys must be unique |

---

# 1. Lists (`list`)

A list stores multiple items in order.

Example:

```python
numbers = [10, 20, 30, 40]
```

Lists are:

- Ordered.
- Changeable.
- Allow duplicates.

---

## Changing List Values

Example:

```python
numbers[1] = 25

print(numbers)
```

Output:

```text
[10, 25, 30, 40]
```

The second value changed from:

```
20 → 25
```

---

## List Indexing

Python starts counting from:

```
0
```

Example:

```python
temperatures = [24, 27, 30]
```

| Index | Value |
|---|---|
| 0 | 24 |
| 1 | 27 |
| 2 | 30 |

Access values:

```python
print(temperatures[0])
```

Output:

```text
24
```

---

# 2. Tuples (`tuple`)

A tuple is similar to a list.

The main difference:

> A tuple cannot be changed after creation.

Example:

```python
coordinates = (51.5, 32.7)
```

You can read values:

```python
print(coordinates[0])
```

Output:

```text
51.5
```

But this is not allowed:

```python
coordinates[0] = 60
```

Python gives an error.

---

## When to Use Tuples

Tuples are useful for data that should remain fixed.

Examples:

- Coordinates.
- Dates.
- Constant values.

Example:

```python
location = (32.5, 51.6)
```

The latitude and longitude usually should not accidentally change.

---

# 3. Sets (`set`)

A set stores unique values.

Example:

```python
classes = {"forest", "water", "forest", "urban"}
```

Output:

```text
{'forest', 'water', 'urban'}
```

The duplicate value:

```
forest
```

is automatically removed.

---

## When to Use Sets

Sets are useful when you only care about unique values.

Examples:

- Unique land-cover classes.
- Unique satellite names.
- Removing duplicates from data.

---

# 4. Dictionaries (`dict`)

A dictionary stores:

```
key → value
```

pairs.

Example:

```python
pixel = {
    "ndvi": 0.72,
    "temperature": 29,
    "cloud": False
}
```

Each value has a label.

Access values using the key:

```python
print(pixel["ndvi"])
```

Output:

```text
0.72
```

# Common Beginner Mistakes

## Mistake 1: Confusing Lists and Tuples

List:

```python
coordinates = [32.5, 51.6]
```

Can change.

Tuple:

```python
coordinates = (32.5, 51.6)
```

Cannot change.

---

## Mistake 2: Thinking Sets Keep Order

Example:

```python
values = {3, 1, 2}
```

A set does not focus on order.

It focuses on uniqueness.

---

## Mistake 3: Forgetting Dictionary Keys

Incorrect:

```python
print(pixel[0])
```

Correct:

```python
print(pixel["ndvi"])
```

Dictionaries are accessed using keys.
