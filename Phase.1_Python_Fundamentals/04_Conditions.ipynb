# Why Should You Care?

Conditions allow a program to **make decisions**.

Without conditions, a program would always execute the same instructions, regardless of the input or data.

In **Artificial Intelligence (AI)**, **Machine Learning (ML)**, **GIS**, and **Remote Sensing (RS)**, conditions are used constantly.

Examples:

- Is this pixel water or land?
- Is the NDVI high enough to indicate healthy vegetation?
- Should this satellite image be discarded because it is too cloudy?
- Is today's temperature above a certain threshold?

Conditions are one of the building blocks of intelligent programs.

---

# Core Concept

A **condition** allows Python to decide which code to execute based on whether an expression is **True** or **False**.

Python provides three keywords for decision-making:

- `if` → Checks the first condition.
- `elif` → Checks another condition if the previous one was `False`.
- `else` → Executes if none of the previous conditions are `True`.

---

# Basic Syntax

```python
if condition:
    # Runs if the condition is True

elif another_condition:
    # Runs if the previous condition was False
    # and this condition is True

else:
    # Runs if all previous conditions are False
```

Notice two important features:

- A colon (`:`) follows each condition.
- The code inside each block is **indented**.

Indentation tells Python which statements belong to each condition.

---

# How an `if` Statement Works

An `if` statement asks a question.

If the answer is **True**, Python executes the indented block.

If the answer is **False**, Python skips it and checks the next condition.

Simple flow:

```text
Condition?

      │
   True│False
      │
 Run code
      │
      ▼
 Continue program
```

---

# Simple Example

Imagine you are deciding whether to take an umbrella.

```python
is_raining = True

if is_raining:
    print("Take an umbrella.")
else:
    print("Enjoy the sunshine!")
```

Output:

```text
Take an umbrella.
```

Python evaluates:

```python
is_raining
```

Since it is `True`, Python executes:

```python
print("Take an umbrella.")
```

The `else` block is skipped.

---

# Using `elif`

Sometimes there are more than two possible outcomes.

Example:

```python
temperature = 25

if temperature >= 30:
    print("Hot")

elif temperature >= 20:
    print("Warm")

else:
    print("Cold")
```

Output:

```text
Warm
```

Python checks the conditions from top to bottom.

- `temperature >= 30` → False
- `temperature >= 20` → True

Since the second condition is true, Python prints:

```text
Warm
```

The `else` block is ignored.

---

# Remote Sensing Example

Suppose a satellite image has:

```python
cloud_percentage = 12
```

You only want images with **less than 20% cloud cover**.

```python
if cloud_percentage < 20:
    print("Use this image.")
else:
    print("Find another image.")
```

Output:

```text
Use this image.
```

This type of filtering is common in **Google Earth Engine** and other Remote Sensing workflows.

---

Another example:

```python
ndvi = 0.68

if ndvi > 0.5:
    print("Healthy vegetation")
else:
    print("Sparse vegetation")
```

Output:

```text
Healthy vegetation
```

The condition:

```python
ndvi > 0.5
```

returns:

```python
True
```

Therefore, the first block executes.

# Common Beginner Mistakes

## Mistake 1: Forgetting the Colon (`:`)

❌ Incorrect:

```python
if temperature > 30
    print("Hot")
```

✅ Correct:

```python
if temperature > 30:
    print("Hot")
```

Every `if`, `elif`, and `else` statement must end with a colon.

---

## Mistake 2: Incorrect Indentation

❌ Incorrect:

```python
if temperature > 30:
print("Hot")
```

Python raises an error because the code inside the condition is not indented.

✅ Correct:

```python
if temperature > 30:
    print("Hot")
```

Indentation tells Python which statements belong to the condition.

---

## Mistake 3: Confusing `=` with `==`

❌ Incorrect:

```python
if temperature = 30:
```

`=` assigns a value and cannot be used for comparison.

✅ Correct:

```python
if temperature == 30:
```

Remember:

- `=` → Assignment.
- `==` → Comparison.
