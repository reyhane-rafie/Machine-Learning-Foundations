# 2. Core Concept (Short)

A **file** is a place where information is stored permanently.

Python can:

- **Read** → Get information from a file.
- **Write** → Save information into a file.

The basic syntax:

```python
file = open("filename.txt", "mode")
```

---

## Common File Modes

| Mode | Purpose |
|---|---|
| `"r"` | Read a file |
| `"w"` | Write/create a file |
| `"a"` | Add information to a file |

After opening a file, you should close it:

```python
file.close()
```

A better modern way is using `with`:

```python
with open("data.txt", "r") as file:
    data = file.read()
```

Python automatically closes the file.

# 4. Remote Sensing / Environmental Example

Imagine you have NDVI measurements.

A CSV file:

```
location,ndvi,year
Forest_A,0.72,2025
Forest_B,0.55,2025
Forest_C,0.31,2025
```

---

## What is CSV?

CSV means:

**Comma-Separated Values**

It is one of the most common formats for storing data.

Example:

| location | ndvi | year |
|---|---|---|
| Forest_A | 0.72 | 2025 |
| Forest_B | 0.55 | 2025 |
| Forest_C | 0.31 | 2025 |

Python can read it:

```python
import csv

with open("ndvi_data.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

Output:

```
['location', 'ndvi', 'year']
['Forest_A', '0.72', '2025']
['Forest_B', '0.55', '2025']
['Forest_C', '0.31', '2025']
```

Each row becomes a **list**.

---

# Writing CSV Files

Example:

```python
import csv

data = [
    ["location", "ndvi"],
    ["Forest_A", 0.72],
    ["Forest_B", 0.55]
]

with open("output.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerows(data)
```

Creates:

```
output.csv
```

with:

```
location,ndvi
Forest_A,0.72
Forest_B,0.55
```

---

# 5. Important Points

- Files allow data to be stored permanently.
- `"r"` reads data from files.
- `"w"` writes new data and can overwrite old data.
- CSV files store tabular data using rows and columns.
- File handling is the bridge between Python and real datasets.
