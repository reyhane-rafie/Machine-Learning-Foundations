# 1. Input

Input means **receiving information** into your program.

Python uses the `input()` function to ask the user for information.

Example:

```python
name = input("What is your name? ")
```

When Python reaches this line:

- It pauses.
- Waits for the user to type.
- Stores the entered value.

If the user types:

```text
Rey
```

then:

```python
name
```

contains:

```python
"Rey"
```

---

# 2. Output

Output means **displaying information** to the user.

Python uses the `print()` function.

Example:

```python
print("Hello!")
```

Output:

```text
Hello!
```

You can also print variables.

```python
name = "Rey"

print(name)
```

Output:

```text
Rey
```

---

# 3. Type Conversion

Sometimes data needs to be converted from one type to another.

Example:

```python
age = "25"
```

Although `"25"` looks like a number, it is actually a **string**.

To convert it into an integer:

```python
age = int(age)
```

Now:

```python
age
```

contains:

```python
25
```

which is an integer.

# Remote Sensing Example

Suppose your program asks for an NDVI threshold.

```python
threshold = input("Enter NDVI threshold: ")
```

The user enters:

```text
0.5
```

You might expect this to be a number.

However, Python stores:

```python
"0.5"
```

which is a **string**.

To use it in calculations, convert it to a float.

```python
threshold = float(threshold)
```

Now Python treats it as:

```python
0.5
```

Example:

```python
ndvi = 0.72

print(ndvi > threshold)
```

Output:

```text
True
```

The same concept applies when reading environmental data from:

- CSV files.
- Configuration files.
- User settings.
- Sensor measurements.

# Common Beginner Mistakes

## Mistake 1: Assuming `input()` Returns a Number

❌ Incorrect assumption:

```python
age = input("Age: ")
```

If the user types:

```text
25
```

`age` is **not** an integer.

It is:

```python
"25"
```

To perform calculations:

```python
age = int(input("Age: "))
```

---

## Mistake 2: Forgetting Type Conversion

Example:

```python
age = input("Age: ")

print(age + 1)
```

This causes an error because Python cannot add an integer to a string.

Correct version:

```python
age = int(input("Age: "))

print(age + 1)
```

---

## Mistake 3: Using the Wrong Conversion Function

If the user enters:

```text
12.8
```

This will fail:

```python
int("12.8")
```

Instead, use:

```python
float("12.8")
```

Choose the conversion function that matches the data you expect.

Because `input()` always returns a string, **type conversion** is often necessary before performing calculations.

Mastering these concepts is essential for writing programs that work with users, files, environmental measurements, and satellite data.
