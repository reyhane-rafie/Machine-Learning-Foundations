## Lesson 5 — TensorFlow vs PyTorch

### 1. Why should I care?

TensorFlow and PyTorch are two popular frameworks for building and training deep learning models. They handle much of the complicated work behind neural networks, including tensor calculations, automatic differentiation, and GPU acceleration. For your goal of applying AI to satellite imagery, Remote Sensing, GIS, and environmental data, learning one framework will let you move from understanding neural networks to building real models.

## 2. Core concept

Both frameworks help you implement the concepts you've been learning:

- Neural networks
- Activation functions
- Loss functions
- Backpropagation
- Gradient descent

Instead of manually calculating every gradient, you define a model and let the framework calculate gradients automatically.

Think of them as toolkits for building neural networks.

|                          | PyTorch                   | TensorFlow                          |
| ------------------------ | ------------------------- | ----------------------------------- |
| Main ecosystem           | PyTorch                   | TensorFlow, often with Keras        |
| Learning style           | Python-friendly, flexible | Keras offers a high-level interface |
| Automatic gradients      | Yes                       | Yes                                 |
| GPU support              | Yes                       | Yes                                 |
| Computer vision          | Strong ecosystem          | Strong ecosystem                    |
| Geospatial deep learning | Commonly used             | Also usable                         |

Both can build powerful deep learning models. Neither is universally better.

## 3. Simple example

Imagine you want to teach a computer to predict whether a student passes an exam based on study hours.

You have already learned how a neural network works. Now, a framework helps you implement it:

```
Study hours
     ↓
Neural network
     ↓
Prediction: pass or fail
     ↓
Calculate loss
     ↓
Calculate gradients automatically
     ↓
Update weights
```

Without a framework, you would need to implement many calculations yourself. With PyTorch or TensorFlow, you can focus more on designing experiments and understanding your results.

My recommendation for your learning journey is PyTorch first.

- Its automatic differentiation system connects directly to the backpropagation lesson you just studied.

### What is a tensor?

Before coding with either framework, understand this word.

A tensor is a general structure for storing numerical data.

You've already learned about NumPy arrays. Tensors are similar, but deep learning frameworks provide additional capabilities, such as automatic gradient calculation and GPU computation.

| Data structure            | Example                                   |
| ------------------------- | ----------------------------------------- |
| Scalar                    | One temperature value: `25.0`             |
| Vector                    | Temperatures from 5 locations             |
| Matrix                    | A grid of temperatures                    |
| Higher-dimensional tensor | A batch of multispectral satellite images |

For example, a batch of 32 satellite image patches, each with 10 bands and a size of 64 × 64 pixels, could have the shape:

```
(32, 10, 64, 64)
```

Here, the dimensions represent:

- `32` — number of image patches
- `10` — spectral bands per patch
- `64 × 64` — pixels in each patch

This kind of data structure is essential for training deep learning models on imagery.

### What does automatic differentiation mean?

Remember backpropagation?

PyTorch can track mathematical operations and automatically calculate gradients when you call `backward()`. TensorFlow can do the same using `tf.GradientTape`.&#x20;



PyTorch Tutorials 1.7.1 documentation





You won't need to calculate every derivative by hand.

## 6. Important points

1. PyTorch and TensorFlow are frameworks for building and training neural networks.
2. Both support automatic differentiation, backpropagation, and GPU acceleration.
3. Tensors store numerical data in one or more dimensions.
4. PyTorch is my recommended starting point for your current learning journey.
5. Understanding the underlying concepts matters more than memorizing framework commands.
