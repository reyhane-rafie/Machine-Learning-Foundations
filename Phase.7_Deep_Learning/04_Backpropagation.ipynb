## Lesson 4 — Backpropagation

---

## 1. Why should I care?

Backpropagation is one of the most important ideas in deep learning. It allows a neural network to figure out which weights contributed to its prediction error and how those weights should change.

Whether you're classifying Sentinel-2 imagery, mapping vegetation, or predicting environmental conditions, backpropagation is a fundamental part of how many neural networks learn.

---

## 2. Core Concept

Let's recall what you already learned:

- **Forward propagation:** The network uses its weights to make a prediction.
- **Loss function:** Measures how far the prediction is from the correct answer.
- **Backpropagation:** Calculates how the loss changes with respect to the network's weights.
- **Gradient descent:** Uses those gradients to update the weights.

The complete learning cycle looks like this:

### 1. Forward propagation

```text
Input → Neural network → Prediction
```

### 2. Calculate the loss

Compare the prediction with the true answer.

### 3. Backpropagation

Calculate gradients for the weights.

### 4. Gradient descent

Update weights to reduce the loss.

Repeat over many training examples.

**The key distinction:** Backpropagation calculates the gradients; gradient descent uses them to update the weights.

### What is a gradient?

You encountered gradients in your mathematics lessons.

A gradient tells us how sensitive the loss is to a change in a parameter.

For example:

- A large positive gradient means increasing that weight would increase the loss locally.
- A negative gradient means increasing that weight would decrease the loss locally.
- A gradient near zero means a small change in that weight has little immediate effect on the loss.

Backpropagation calculates these gradients efficiently throughout the network.

---

## 3. Simple Example

Imagine you're learning to throw a ball into a basket.

1. You throw the ball and miss.
2. You measure how far the ball landed from the basket.
3. You work out how your throwing angle and force contributed to the miss.
4. You adjust your next throw.

Backpropagation is similar to step 3: it works backward from the error to determine how the model's parameters influenced it.

Gradient descent is similar to step 4: it uses that information to adjust the parameters.

One important difference: backpropagation uses mathematical derivatives, not guesswork, to calculate the local effects of parameter changes.

## 4. Remote Sensing / Environmental Example

Imagine training a neural network to classify satellite pixels as vegetation or non-vegetation.

### Input features

```text
NDVI
Red band
Near-infrared band
Land surface temperature
```

### Prediction

The model predicts vegetation with a probability of 0.60.

- **True label:** Vegetation
- **Predicted probability:** 0.60

The prediction is correct, but the model may still improve its confidence. A suitable loss function measures the prediction error.

### Backpropagation

Calculates how the loss depends on the network's weights.

### Gradient descent

Adjusts those weights. Training repeats across many labeled pixels.

Over time, the network may learn spectral patterns that distinguish vegetation from other land-cover types.

It doesn't simply memorize a rule such as "high NDVI means vegetation." It learns combinations of features through training, although the quality of what it learns depends on the data and model.

## 5. Important Details

### How does backpropagation work?

Consider a network with two hidden layers.

During the **forward pass**, information travels from the inputs toward the output.

During **backpropagation**, the loss information travels in the reverse direction. Using the chain rule, the network calculates how each parameter contributes to the final loss through the connected layers.

You already learned the chain rule in calculus. Backpropagation is a practical application of that idea.

For now, remember:

$$
\text{Gradient} = \frac{\partial L}{\partial w}
$$

Here:

- \(L\) is the loss.
- \(w\) is a weight.

The expression tells us how the loss changes when that weight changes slightly.

### How are weights updated?

Gradient descent uses this update rule:

$$
w_{\text{new}} = w_{\text{old}} - \eta \frac{\partial L}{\partial w}
$$

Where:

- \(w\): The weight being updated.
- \(L\): The loss.
- \(\eta\): The learning rate.
- \(\frac{\partial L}{\partial w}\): The gradient calculated by backpropagation.

The learning rate controls the size of each update.

- **Too large:** Training may become unstable or fail to converge.
- **Too small:** Training may be very slow.

### Is backpropagation the same as training?

Not exactly.

| Concept | Role |
|---|---|
| Forward propagation | Calculates predictions |
| Loss function | Measures prediction error |
| Backpropagation | Calculates gradients |
| Gradient descent | Updates parameters |
| Training | Repeats these steps to learn |

Backpropagation itself doesn't guarantee that the network finds the best possible solution. Training can depend on initialization, learning rate, data quality, and other factors.

---

## 6. Important Points

1. Backpropagation calculates gradients of the loss with respect to network parameters.
2. It works backward from the loss through the network.
3. It relies on the chain rule from calculus.
4. Gradient descent uses the calculated gradients to update weights.
5. Forward propagation, loss calculation, backpropagation, and parameter updates form the core training cycle of many neural networks.
