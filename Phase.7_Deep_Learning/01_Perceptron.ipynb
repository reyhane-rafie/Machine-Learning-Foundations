# Phase 7 — Deep Learning

## Lesson 1 — Perceptron

---

### 1. Why should I care?

The **Perceptron** is one of the simplest building blocks of neural networks. Understanding it will make later topics—**neural networks, activation functions, backpropagation, and deep learning**—much easier to understand.

In Remote Sensing, the same basic idea can eventually be extended to models that classify satellite pixels, detect objects, or learn complex patterns from imagery.

---

### 2. Core Concept

A Perceptron is a simple model that takes several inputs, gives each input a **weight**, combines them, and produces an output.

Think of it as:

```text
Inputs
  ↓
Multiply by weights
  ↓
Add them together
  ↓
Apply an activation function
  ↓
Output
````

For example:

```text
x₁ ──× w₁ ──┐
x₂ ──× w₂ ──┤
x₃ ──× w₃ ──┤ → weighted sum → activation → output
             │
           bias
```

The basic calculation is:

$$
z = w_1x_1 + w_2x_2 + ... + w_nx_n + b
$$

where:

* `x` = input feature
* `w` = weight
* `b` = bias
* `z` = weighted sum

The Perceptron then uses an **activation function** to decide the output.

A classic Perceptron uses a step function:

```text
if z ≥ 0 → output 1
if z < 0 → output 0
```

So the Perceptron is essentially learning:

> **Which combination of input values should lead to class 0 or class 1?**

---

### 3. Simple Example

Imagine predicting whether a person will **go running**.

We have:

```text
x₁ = Is it sunny?
x₂ = Is it warm?
```

The Perceptron might learn:

```text
sunny → weight = 2
warm  → weight = 1
bias  → -2
```

For:

```text
sunny = 1
warm = 1
```

we calculate:

$$
z = (2)(1) + (1)(1) - 2 = 1
$$

Since `1 ≥ 0`:

```text
Output = 1 → Go running
```

The important idea is that **the model learns the weights** rather than us manually choosing them.

---

### 4. Remote Sensing / Environmental Example

Imagine classifying satellite pixels as:

```text
0 = non-vegetation
1 = vegetation
```

We could give a Perceptron features such as:

```text
NDVI
B4
B8
LST
```

For example:

```text
NDVI ──× weight
B4   ──× weight
B8   ──× weight
LST  ──× weight
          ↓
      weighted sum
          ↓
       activation
          ↓
Vegetation / Non-vegetation
```

During training, the Perceptron adjusts its **weights and bias** to improve its classifications.

This is a very simple version of what much more sophisticated neural networks eventually do.

---

### 5. Important Details

#### What problem does it solve?

Primarily **binary classification**.

```text
Class 0
   vs
Class 1
```

#### What does it learn?

It learns:

* **Weights** → determine how each input contributes to the decision
* **Bias** → shifts the decision boundary

#### Why is it called a neural-network building block?

Because its structure resembles a very simplified artificial neuron:

```text
inputs → weighted combination → activation → output
```

Multiple artificial neurons can be connected together to create a **neural network**.

#### Important Limitation

A basic Perceptron can only learn **linearly separable** problems.

Imagine two classes that can be separated with a straight line:

```text
● ● ● | ○ ○ ○
● ● ● | ○ ○ ○
```

A Perceptron can handle this.

But if the classes require a curved or more complicated boundary, a single Perceptron cannot solve it.

This limitation is one reason we eventually need **multiple neurons and nonlinear activation functions**.

---

### 6. Important Points

Remember these **5 things**:

1. A Perceptron is a simple **binary classifier**.
2. It calculates a **weighted sum + bias**.
3. An activation function converts that value into an output.
4. Training adjusts the **weights and bias**.
5. A single Perceptron can only learn **linear decision boundaries**.

---

### 7. Practice Together — Question 1

Suppose we have:

```text
x₁ = 2
x₂ = 3

w₁ = 1
w₂ = 2

b = -5
```

The Perceptron calculates:

$$
z = w_1x_1 + w_2x_2 + b
$$

Using the classic step activation:

```text
z ≥ 0 → output 1
z < 0 → output 0
```

**What is the final output, and how did you calculate it?**

```
```
