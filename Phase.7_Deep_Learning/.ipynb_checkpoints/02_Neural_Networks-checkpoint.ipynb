## Lesson 2 — Neural Networks

---

## 1. Why should I care?

A **neural network** takes the simple idea of the Perceptron and connects many artificial neurons together. This allows the model to learn much more complex patterns than a single linear decision boundary.

In Remote Sensing, neural networks are fundamental for tasks such as **land-cover classification, image segmentation, object detection, crop mapping, and environmental prediction**.

---

## 2. Core Concept

A neural network is essentially a collection of connected neurons organized into **layers**.

A simple neural network looks like:

```text
Input Layer        Hidden Layer        Output Layer

x₁ ──────────────→ ● ───────────────→
                    ↘                  ●
x₂ ──────────────→ ● ───────────────→
                    ↗
x₃ ──────────────→ ● ───────────────→
````

More commonly:

```text
Inputs
  ↓
Hidden Layer
  ↓
Hidden Layer
  ↓
Output
```

Each neuron performs something similar to the Perceptron:

$$
z = w_1x_1 + w_2x_2 + ... + b
$$

Then:

$$
a = f(z)
$$

where `f` is an **activation function**.

The crucial difference is that neurons are connected in layers. This allows the network to gradually transform the input.

For example:

```text
Raw satellite features
        ↓
Simple patterns
        ↓
More complex patterns
        ↓
Very complex patterns
        ↓
Prediction
```

A network with **multiple hidden layers** is called a **deep neural network**.

---

## 3. Simple Example

Imagine trying to recognize a handwritten number.

A computer receives many pixel values:

```text
Pixel 1
Pixel 2
Pixel 3
...
Pixel 784
```

The first hidden layer might learn simple patterns such as:

```text
edges
lines
curves
```

Later layers can combine these:

```text
edges + curves
      ↓
parts of numbers
      ↓
complete number
```

So instead of manually telling the model:

> "Look for this curve and this line."

the neural network **learns useful patterns from the data**.

This ability to learn increasingly complex representations is one of the most important ideas in deep learning.

---

## 4. Remote Sensing / Environmental Example

Imagine classifying a Sentinel-2 pixel into:

```text
Forest
Agriculture
Water
Urban
Bare soil
```

Our input could contain:

```text
B2
B3
B4
B8
B11
B12
NDVI
NDWI
LST
Elevation
```

A neural network could learn relationships between these features:

```text
Satellite features
       ↓
   Hidden Layer
       ↓
   Hidden Layer
       ↓
  Class probabilities
       ↓
Forest / Agriculture / Water / Urban / Soil
```

For example, it might learn that a combination of:

```text
high NIR
low red
certain SWIR response
specific temperature
```

is characteristic of vegetation.

Importantly, **we don't explicitly program those rules**. The network learns the weights that produce useful relationships from training examples.

---

## 5. Important Details

### What problem does it solve?

Neural networks can handle:

* Classification
* Regression
* Image classification
* Segmentation
* Object detection
* Time-series prediction

They can work with many types of data:

```text
Tabular data
Images
Satellite imagery
Time series
Text
Audio
```

---

### What Are the Main Components?

#### 1. Input Layer

Receives the features.

```text
NDVI
LST
Elevation
B4
B8
```

#### 2. Hidden Layers

Transform the information.

```text
Input → Hidden → Hidden → ...
```

These are where the network learns increasingly useful representations.

#### 3. Output Layer

Produces the final prediction.

For binary classification:

```text
0 or 1
```

For multi-class classification:

```text
Forest:      0.72
Agriculture: 0.15
Water:       0.03
Urban:       0.10
```

The model would predict **Forest**.

---

### What Does Training Mean?

Initially, the weights are generally not useful.

The network makes a prediction:

```text
Input
 ↓
Neural network
 ↓
Prediction
```

We compare it with the true answer:

```text
Prediction
     ↓
Loss
```

Then an optimization process adjusts the weights to reduce the loss.

Later, we will learn exactly how this happens through:

**forward propagation → loss → backpropagation → gradient descent**

You already learned gradient descent in the mathematics phase, so this will connect directly to what you already know.

---

### Neural Network vs Perceptron

Think of it this way:

```text
Perceptron
    ↓
one simple artificial neuron

Neural Network
    ↓
many connected artificial neurons

Deep Neural Network
    ↓
many neurons + multiple hidden layers
```

The Perceptron is therefore a useful conceptual starting point for understanding neural networks.

---

## 6. Important Points

1. A neural network consists of **connected artificial neurons arranged in layers**.
2. Each neuron performs a weighted sum followed by an **activation function**.
3. Hidden layers allow the network to learn increasingly complex representations.
4. Training adjusts weights to reduce a **loss function**.
5. Multiple hidden layers make the network **deep**, which is the foundation of deep learning.

---

## 7. Practice Together — Question 1

Let's test the most important conceptual distinction.

Suppose we have:

```text
NDVI
LST
Elevation
B4
B8
```

and we want to classify each pixel as:

```text
Forest
Agriculture
Water
Urban
```

A single Perceptron is quite limited because it produces a relatively simple linear decision boundary.

**Why can a neural network with hidden layers learn more complex relationships than a single Perceptron?**

Explain it in your own words. Don't worry about using perfect terminology.

```
```
