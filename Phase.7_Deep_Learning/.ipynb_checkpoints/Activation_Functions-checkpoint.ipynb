# Phase 7 — Deep Learning

## Lesson 3 — Activation Functions

---

## 1. Why should I care?

Activation functions are what allow neural networks to learn **nonlinear patterns**. Without them, even a network with many layers would behave essentially like one linear model.

This is especially important for Remote Sensing, where relationships between spectral bands, vegetation, temperature, soil, and land-cover classes are rarely simple straight-line relationships.

---

## 2. Core Concept

Recall what a neuron does:

$$
z = w_1x_1 + w_2x_2 + ... + b
$$

Then the neuron applies an **activation function**:

$$
a = f(z)
$$

The activation function determines how the neuron's value is transformed before being passed to the next layer.

Think of it as a **decision gate**:

```text
Inputs
  ↓
Weighted sum
  ↓
Activation function
  ↓
Output
  ↓
Next neuron/layer
````

### Why do we need it?

Suppose we have:

```text
Input
 ↓
Linear calculation
 ↓
Linear calculation
 ↓
Linear calculation
```

Even if we stack many layers, the overall relationship remains linear.

But real problems are often nonlinear.

For example:

```text
Temperature + rainfall + NDVI
              ↓
       complex relationship
              ↓
       crop condition
```

An activation function introduces the **nonlinearity** that allows neural networks to learn these complex relationships.

---

# 3. The Main Activation Functions

You don't need to memorize every activation function. For now, focus on **three important ones**.

---

### A. Step Function

This is the classic Perceptron activation.

```text
z < 0  → 0
z ≥ 0 → 1
```

It creates a hard decision:

```text
        output
          1 |       ─────────
            |
          0 |────────
            +----------------
                 z
```

Useful for understanding the original Perceptron, but **not commonly used for training modern neural networks**.

Why?

Because it doesn't provide a useful gradient for gradient descent.

---

### B. Sigmoid

The sigmoid transforms any value into a number between **0 and 1**.

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

You don't need to memorize the formula yet. Understand the behavior:

```text
Large negative → close to 0

0              → 0.5

Large positive → close to 1
```

For example:

```text
z = -5 → ~0.007
z =  0 → 0.5
z = +5 → ~0.993
```

This makes sigmoid useful for **binary classification output layers**.

For example:

```text
0.92 → 92% model probability of class 1
```

More precisely, it represents the model's estimated probability under the binary-classification setup; it is not automatically a perfectly calibrated real-world probability.

---

### C. ReLU

ReLU stands for:

**Rectified Linear Unit**

Its rule is extremely simple:

$$
ReLU(z)=\max(0,z)
$$

So:

```text
z < 0 → 0
z > 0 → z
```

Examples:

```text
ReLU(-3) = 0
ReLU(-1) = 0
ReLU( 0) = 0
ReLU( 2) = 2
ReLU( 7) = 7
```

Visually:

```text
output
  |
  |          /
  |         /
  |        /
  |_______/________
          0       z
```

ReLU is extremely common in **hidden layers** of neural networks because it introduces nonlinearity while remaining computationally simple.

---

# 4. Simple Example

Imagine a security system deciding whether someone should enter.

The model produces:

```text
z = 2.5
```

A sigmoid could transform that into something like:

```text
0.92
```

You could interpret this as:

> The model estimates a high probability of class 1.

Now imagine a hidden neuron produces:

```text
z = -4
```

ReLU gives:

```text
0
```

So that neuron doesn't pass a positive activation forward.

If:

```text
z = 4
```

ReLU gives:

```text
4
```

---

# 5. Remote Sensing / Environmental Example

Imagine a neural network is learning whether a pixel represents **healthy vegetation**.

Input features might include:

```text
NDVI
NDWI
B4
B8
B11
LST
```

A hidden neuron might calculate:

$$
z =
w_1(NDVI)
+ w_2(LST)
+ w_3(B8)
+ ...
+ b
$$

Then:

```text
z
↓
ReLU
↓
activation
↓
next layer
```

Different neurons can respond to different patterns.

For example, one neuron might become strongly activated by a combination associated with:

* high vegetation signal
* appropriate moisture
* relatively favorable temperature

while another may respond to a different spectral pattern.

**The important point:** we don't manually tell the neurons what these patterns are. During training, the network learns weights that make useful internal patterns emerge.

---

# 6. Important Details

### Where are activation functions used?

Typically:

```text
Input
 ↓
Hidden layer → ReLU
 ↓
Hidden layer → ReLU
 ↓
Output layer → activation appropriate to task
```

The output activation depends on the problem.

### Common Choices

| Task                       | Typical output activation           |
| -------------------------- | ----------------------------------- |
| Binary classification      | Sigmoid                             |
| Multi-class classification | Softmax                             |
| Regression                 | Often no activation / linear output |
| Hidden layers              | Often ReLU or related functions     |

We will study **softmax** more carefully later.

---

### Why not use sigmoid everywhere?

Sigmoid can suffer from the **vanishing gradient** problem.

When sigmoid gets very close to 0 or 1, its gradient becomes very small. During backpropagation, this can make learning slow, especially in deeper networks.

ReLU generally avoids this problem for positive inputs and is therefore commonly preferred in hidden layers.

However, ReLU has its own issue:

### Dead ReLU

If a neuron consistently receives negative values:

```text
z < 0
```

then:

```text
ReLU(z) = 0
```

and its gradient can become zero, potentially causing that neuron to stop learning.

This is one reason variants such as **Leaky ReLU** exist.

You don't need to go deeply into those yet.

---

# 7. Important Points

Remember these **5 ideas**:

1. An activation function transforms a neuron's weighted sum.
2. **Nonlinear activation functions allow neural networks to learn nonlinear relationships.**
3. **ReLU** is commonly used in hidden layers.
4. **Sigmoid** is commonly used for binary-classification outputs.
5. The activation function used in the output layer depends on the prediction task.

---

# 8. Practice Together — Question 1

Imagine we remove all activation functions from a neural network.

The network still has:

```text
Input
 ↓
Hidden Layer 1
 ↓
Hidden Layer 2
 ↓
Output
```

Each layer still performs weighted sums and adds biases.

**Question:**

Why would this network still be unable to learn genuinely nonlinear relationships, even though it has multiple layers?

Explain the idea in your own words.

```
```
